####entrenamiento y selección del modelo

````python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

clientes_df = pd.read_parquet('clientes.parquet')
productos_df = pd.read_parquet('productos.parquet')

universe_size = len(clientes_df) * len(productos_df)

def get_batch_performance(batch_id):
    # Cargar datos reales
    data = pd.read_parquet(f"data/batch_t{batch_id}.parquet")
    data = data.groupby(['customer_id', 'product_id']).size().reset_index()
    true_pairs = set(map(tuple, data[['customer_id', 'product_id']].values))

    # Cargar predicciones
    preds = pd.read_csv(f"predictions/batch_t{batch_id}.csv",
                        names=['customer_id', 'product_id'])
    pred_pairs = set(map(tuple, preds[['customer_id', 'product_id']].values))

    # TP = predicho y real
    tp = len(pred_pairs & true_pairs)
    # FP = predicho pero no real
    fp = len(pred_pairs - true_pairs)
    # FN = real pero no predicho
    fn = len(true_pairs - pred_pairs)
    # TN = universo total - todo lo demás
    tn = universe_size - tp - fp - fn

    y_true = [1]*tp + [0]*fp + [1]*fn + [0]*tn
    y_pred = [1]*tp + [1]*fp + [0]*fn + [0]*tn

    metrics = {
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred),
        "recall": recall_score(y_true, y_pred),
        "f1": f1_score(y_true, y_pred)
    }

    return metrics
````

####entrenamiento del modelo


````python
import pandas as pd

print("--- Cargando Datos ---")
# --- Carga de Datos ---

try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    print("Archivo transacciones.parquet cargado exitosamente.")
    print("Primeras 5 filas de transacciones:")
    print(df_transacciones.head())
    print("\nInformación del DataFrame de transacciones:")
    df_transacciones.info()
except FileNotFoundError:
    print("Error: El archivo transacciones.parquet no se encontró.")

try:
    df_clientes = pd.read_parquet('clientes.parquet')
    print("\nArchivo clientes.parquet cargado exitosamente.")
    print("Primeras 5 filas de clientes:")
    print(df_clientes.head())
    print("\nInformación del DataFrame de clientes:")
    df_clientes.info()
except FileNotFoundError:
    print("Error: El archivo clientes.parquet no se encontró.")

try:
    df_productos = pd.read_parquet('productos.parquet')
    print("\nArchivo productos.parquet cargado exitosamente.")
    print("Primeras 5 filas de productos:")
    print(df_productos.head())
    print("\nInformación del DataFrame de productos:")
    df_productos.info()
except FileNotFoundError:
    print("Error: El archivo productos.parquet no se encontró.")

# --- Transformaciones Iniciales y Agregación Diaria ---
if 'df_transacciones' in locals() and 'df_clientes' in locals() and 'df_productos' in locals():
    print("\n--- Realizando Transformaciones y Agregación Diaria ---")

    # Convertir 'purchase_date' a tipo datetime y extraer solo la fecha (sin la hora)
    df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date

    # Agrupar transacciones por cliente, producto y día.
    # Se suman los 'items' para netear compras y devoluciones del mismo día.
    print("Agregando transacciones a nivel diario para netear compras y devoluciones...")
    df_trans_agg = df_transacciones.groupby(
        ['customer_id', 'product_id', 'purchase_date']
    ).agg(
        items_neto=('items', 'sum')
    ).reset_index()

    print(f"Filas antes de la agregación: {len(df_transacciones)}")
    print(f"Filas después de la agregación diaria: {len(df_trans_agg)}")

    # Filtrar para mantener solo las compras efectivas (items_neto > 0)
    df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
    print(f"Filas después de filtrar por compras efectivas (items_neto > 0): {len(df_compras_efectivas)}")

    # --- Cruce de Información (Merge) con los datos ya agregados ---
    print("\n--- Realizando Cruce de Datos (Merge) ---")
    # Renombrar columna para claridad en los pasos siguientes
    df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)
    df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])

    # Crear columnas 'year' y 'week' a partir de la fecha de compra
    df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
    df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week

    # Cruzar con información de clientes
    print("Cruzando con clientes...")
    df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')

    # Cruzar el resultado con información de productos
    print("Cruzando con productos...")
    df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')

    print(f"Cruce finalizado. Dimensiones del DataFrame 'df_merged': {df_merged.shape}")

    # --- Verificación Final de Nulos y Duplicados Post-Proceso ---
    print("\n--- Verificación Post-Merge ---")
    print(f"Número de filas duplicadas encontradas: {df_merged.duplicated().sum()}")
    nulos = df_merged.isnull().sum()
    print("\nValores nulos por columna:")
    print(nulos[nulos > 0])
    print("Primeras 5 filas del DataFrame combinado:")
    print(df_merged.head())
    print("\nInformación del DataFrame combinado:")
    df_merged.info()

    print("\n--- Pre-procesamiento Completado ---")
    print(f"Dimensiones finales: {df_merged.shape}")

else:
    print("\nNo se pudieron cargar todos los DataFrames necesarios. El pre-procesamiento no pudo continuar.")
````

````python
import pandas as pd
import plotly.express as px

# --- Holdout Split (Train / Validation / Test) ---

print("\n--- División Holdout (Train / Validation / Test) ---")

# Ensure 'year_week' column is created before splitting
df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)

# Definir semanas para cada conjunto
weeks_total = sorted(df_merged['year_week'].unique())
num_weeks_total = len(weeks_total)
num_weeks_test = 4
num_weeks_val = 4

# Calcular puntos de corte (índices)
idx_val_start = num_weeks_total - num_weeks_test - num_weeks_val
idx_test_start = num_weeks_total - num_weeks_test

# Obtener listas de semanas para cada conjunto
weeks_train = weeks_total[:idx_val_start]
weeks_val = weeks_total[idx_val_start:idx_test_start]
weeks_test = weeks_total[idx_test_start:]

print(f"Semanas para Entrenamiento ({len(weeks_train)}): {weeks_train[0]} a {weeks_train[-1]}")
print(f"Semanas para Validación ({len(weeks_val)}): {weeks_val[0]} a {weeks_val[-1]}")
print(f"Semanas para Prueba ({len(weeks_test)}): {weeks_test[0]} a {weeks_test[-1]}")

# Crear los DataFrames para cada conjunto a partir de df_merged
df_train = df_merged[df_merged['year_week'].isin(weeks_train)].copy()
df_val = df_merged[df_merged['year_week'].isin(weeks_val)].copy()
df_test = df_merged[df_merged['year_week'].isin(weeks_test)].copy()

# Verificar las dimensiones y los rangos temporales
print(f"\nDimensiones df_train: {df_train.shape}")
print(f"Semanas en df_train: {df_train['year_week'].min()} a {df_train['year_week'].max()} ({df_train['year_week'].nunique()} semanas)")
print(f"Fechas en df_train: {df_train['purchase_date'].min().strftime('%Y-%m-%d')} a {df_train['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_val: {df_val.shape}")
print(f"Semanas en df_val: {df_val['year_week'].min()} a {df_val['year_week'].max()} ({df_val['year_week'].nunique()} semanas)")
print(f"Fechas en df_val: {df_val['purchase_date'].min().strftime('%Y-%m-%d')} a {df_val['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_test: {df_test.shape}")
print(f"Semanas en df_test: {df_test['year_week'].min()} a {df_test['year_week'].max()} ({df_test['year_week'].nunique()} semanas)")
print(f"Fechas en df_test: {df_test['purchase_date'].min().strftime('%Y-%m-%d')} a {df_test['purchase_date'].max().strftime('%Y-%m-%d')}")

# --- Visualización de la División Temporal ---
print("\nVisualizando la división temporal...")

# Crear un DataFrame para la visualización
set_map = {week: 'Train' for week in weeks_train}
set_map.update({week: 'Validation' for week in weeks_val})
set_map.update({week: 'Test' for week in weeks_test})

viz_df = pd.DataFrame(list(set_map.items()), columns=['year_week', 'dataset'])
viz_df = viz_df.sort_values('year_week')
# Contar semanas por dataset para el gráfico
viz_counts = viz_df['dataset'].value_counts().reindex(['Train', 'Validation', 'Test'])

fig_split = px.bar(x=viz_counts.index, y=viz_counts.values,
                   title='Distribución Temporal de los Conjuntos de Datos (Número de Semanas)',
                   labels={'x': 'Conjunto', 'y': 'Número de Semanas'},
                   color=viz_counts.index,
                   text_auto=True)
fig_split.update_layout(showlegend=False)
fig_split.show()
````

````python
from itertools import product

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    """
    Genera la tabla base con todas las combinaciones (customer_id, product_id, target_week)
    para las semanas objetivo dadas y calcula el target (si hubo compra o no).

    Args:
        transactions_df (pd.DataFrame): DataFrame con las transacciones (debe tener
                                         customer_id, product_id, year_week).
                                         Usado para determinar compras reales.
        target_weeks (list): Lista de semanas (formato 'YYYY-WNN') para las cuales
                             se generará la base y el target.

    Returns:
        pd.DataFrame: DataFrame con columnas [customer_id, product_id, target_week, target].
                      Contiene todas las combinaciones relevantes y si la compra ocurrió.
    """
    if not target_weeks:
        print("Warning: No target weeks provided.")
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])

    print(f"Generando base para target_weeks: {target_weeks[0]} a {target_weeks[-1]}")

    # Filtrar transacciones relevantes para el target
    # Seleccionar solo las columnas necesarias y eliminar duplicados a nivel de semana
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1 # Marcar que estas compras ocurrieron

    # --- Determinar el universo de clientes y productos activos ---
    # Se usan todos los clientes/productos presentes en el dataframe de transacciones de entrada
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()

    print(f"Clientes activos considerados: {len(active_customers)}")
    print(f"Productos activos considerados: {len(active_products)}")

    # --- Crear todas las combinaciones posibles para las semanas objetivo ---
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])

    print(f"Tamaño base inicial (combinaciones): {df_base.shape}")

    # --- Unir con las transacciones reales para obtener el target ---
    # Usar left_on y right_on para manejar nombres de columna diferentes
    df_final = pd.merge(
        df_base,
        target_transactions,
        left_on=['customer_id', 'product_id', 'target_week'], # Claves de df_base
        right_on=['customer_id', 'product_id', 'year_week'],   # Claves correspondientes en target_transactions
        how='left'
    )

    # Eliminar la columna 'year_week' redundante que viene del merge
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])

    # Rellenar con 0 donde no hubo compra
    df_final['target'] = df_final['target'].fillna(0).astype(int)

    print(f"Tamaño final base + target: {df_final.shape}")
    # Asegurarse que el target existe antes de calcular value_counts
    if 'target' in df_final.columns and not df_final.empty:
        print(f"Distribución del target:\n{df_final['target'].value_counts(normalize=True)}")
    else:
        print("Target column not found or DataFrame is empty after merge.")


    return df_final

# --- Aplicar la función para crear las bases de Validación y Test ---

print("\n--- Creando Base de Validación ---")
target_weeks_val = sorted(df_val['year_week'].unique())
df_base_val = create_target_base(df_val, target_weeks_val)

print("\n--- Creando Base de Test ---")
target_weeks_test = sorted(df_test['year_week'].unique())
df_base_test = create_target_base(df_test, target_weeks_test)


# Se muestran las primeras filas de una de las bases creadas
print("\nEjemplo df_base_val:")
# Asegurarse que df_base_val no está vacío antes de llamar a head()
if not df_base_val.empty:
    print(df_base_val.head())
else:
    print("df_base_val está vacío.")
````

````python
from itertools import product

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    """
    Genera la tabla base con todas las combinaciones (customer_id, product_id, target_week)
    para las semanas objetivo dadas y calcula el target (si hubo compra o no).

    Args:
        transactions_df (pd.DataFrame): DataFrame con las transacciones (debe tener
                                         customer_id, product_id, year_week).
                                         Usado para determinar compras reales.
        target_weeks (list): Lista de semanas (formato 'YYYY-WNN') para las cuales
                             se generará la base y el target.

    Returns:
        pd.DataFrame: DataFrame con columnas [customer_id, product_id, target_week, target].
                      Contiene todas las combinaciones relevantes y si la compra ocurrió.
    """
    if not target_weeks:
        print("Warning: No target weeks provided.")
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])

    print(f"Generando base para target_weeks: {target_weeks[0]} a {target_weeks[-1]}")

    # Filtrar transacciones relevantes para el target
    # Seleccionar solo las columnas necesarias y eliminar duplicados a nivel de semana
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1 # Marcar que estas compras ocurrieron

    # --- Determinar el universo de clientes y productos activos ---
    # Se usan todos los clientes/productos presentes en el dataframe de transacciones de entrada
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()

    print(f"Clientes activos considerados: {len(active_customers)}")
    print(f"Productos activos considerados: {len(active_products)}")

    # --- Crear todas las combinaciones posibles para las semanas objetivo ---
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])

    print(f"Tamaño base inicial (combinaciones): {df_base.shape}")

    # --- Unir con las transacciones reales para obtener el target ---
    # Usar left_on y right_on para manejar nombres de columna diferentes
    df_final = pd.merge(
        df_base,
        target_transactions,
        left_on=['customer_id', 'product_id', 'target_week'], # Claves de df_base
        right_on=['customer_id', 'product_id', 'year_week'],   # Claves correspondientes en target_transactions
        how='left'
    )

    # Eliminar la columna 'year_week' redundante que viene del merge
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])

    # Rellenar con 0 donde no hubo compra
    df_final['target'] = df_final['target'].fillna(0).astype(int)

    print(f"Tamaño final base + target: {df_final.shape}")
    # Asegurarse que el target existe antes de calcular value_counts
    if 'target' in df_final.columns and not df_final.empty:
        print(f"Distribución del target:\n{df_final['target'].value_counts(normalize=True)}")
    else:
        print("Target column not found or DataFrame is empty after merge.")


    return df_final

# --- Aplicar la función para crear las bases de Validación y Test ---

print("\n--- Creando Base de Validación ---")
target_weeks_val = sorted(df_val['year_week'].unique())
df_base_val = create_target_base(df_val, target_weeks_val)

print("\n--- Creando Base de Test ---")
target_weeks_test = sorted(df_test['year_week'].unique())
df_base_test = create_target_base(df_test, target_weeks_test)


# Se muestran las primeras filas de una de las bases creadas
print("\nEjemplo df_base_val:")
# Asegurarse que df_base_val no está vacío antes de llamar a head()
if not df_base_val.empty:
    print(df_base_val.head())
else:
    print("df_base_val está vacío.")
````

````python
from itertools import product

# --- Auxiliary Functions (moved here for completeness if not defined previously) ---
def year_week_to_date(yw_str):
    """Convierte 'YYYY-WNN' a la fecha del lunes de esa semana ISO."""
    try:
        # %G: Año ISO, %V: Semana ISO (01-53), %u: Día semana ISO (1=Lunes)
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        # Manejar posibles errores de formato
        print(f"Error convirtiendo: {yw_str}")
        return pd.NaT

def get_lag_week(target_week_str, lag):
    """Calcula la semana 'YYYY-WNN' correspondiente a 'lag' semanas antes."""
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    # Convertir de nuevo a formato 'YYYY-WNN'
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    """
    Añade features de lag (semanas anteriores) a la tabla base.

    Args:
        base_df (pd.DataFrame): Tabla base (customer_id, product_id, target_week, [target]).
        transactions_hist_df (pd.DataFrame): Transacciones históricas para buscar datos de lag.
                                            Debe tener customer_id, product_id, year_week,
                                            y las columnas en feature_cols.
        lag_weeks (list): Lista de enteros para los lags (ej., [1, 2, 4]).
        feature_cols (list): Lista de nombres de columnas de transactions_hist_df
                             cuyos valores se usarán como features de lag (ej., ['purchased', 'items']).

    Returns:
        pd.DataFrame: base_df con las nuevas columnas de features de lag añadidas.
    """
    df_with_lags = base_df.copy()

    # Preparar datos históricos (seleccionar columnas relevantes)
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()

    for lag in lag_weeks:
        print(f"  Generando lags de {lag} semana(s)...")
        # Calcular la semana de búsqueda (lookup_week) para esta pasada de lag
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))

        # Preparar datos históricos para el merge de este lag
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})

        # Definir nombres para las nuevas columnas de lag
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)

        # Columnas a seleccionar de hist_lag para el merge
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())

        # Unir los datos históricos del lag correspondiente
        df_with_lags = pd.merge(
            df_with_lags,
            hist_lag[merge_cols],
            on=['customer_id', 'product_id', 'lookup_week'],
            how='left'
        )

        # Rellenar NaNs para las nuevas columnas de lag con 0
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
            else:
                 print(f"Advertencia: La columna de lag '{lag_col}' no se encontró después del merge para lag {lag}.")


        # Eliminar columna temporal
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])

    return df_with_lags



# --- Combine train and validation for full historical context for Test set ---
print("Combinando df_train y df_val para historial de Test...")
df_train_val = pd.concat([df_train, df_val], ignore_index=True)


# --- REFINAR DATOS HISTÓRICOS: Agregar 'items' por semana ---
print("Refinando datos históricos (agregando 'items' por semana)...")

# Para df_train_hist
df_train_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index()
df_train_agg['purchased'] = 1 # Añadir indicador de compra
df_train_hist_agg = df_train_agg.copy()

# Para df_train_val_hist
df_train_val_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index()
df_train_val_agg['purchased'] = 1 # Añadir indicador de compra
df_train_val_hist_agg = df_train_val_agg.copy()

print("Datos históricos agregados.")

# Define lags and features to compute for the `add_lag_features` function
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']


# --- RE-GENERAR Features de Lag con datos agregados ---

# --- Generar features de lag para el Validation set ---
print("\nRe-Generando Features de Lag para Validation Set (con items agregados)...")
df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)

# --- Generar features de lag para el Test set ---
print("\nRe-Generando Features de Lag para Test Set (con items agregados)...")
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

print("\nFeatures de Lag re-generadas.")

# --- Mostrar Resultados (verificar items enteros o suma correcta) ---
print("\nTest Set con Features de Lag Re-Generadas (primeras filas):")
print(df_feat_test.head())


# Features de Ventana Móvil
def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    """
    Añade features de ventana móvil a la tabla base.
    Calcula agregados sobre N semanas ANTERIORES a la target_week.

    Args:
        base_df (pd.DataFrame): Tabla base (customer_id, product_id, target_week, ...).
        transactions_hist_agg_df (pd.DataFrame): Transacciones históricas AGREGADAS
                                                 (1 fila por cust-prod-week). Debe tener
                                                 customer_id, product_id, year_week,
                                                 y las columnas en feature_cols.
        windows (list): Lista de enteros para tamaños de ventana (ej., [4, 8]).
        feature_cols (list): Columnas sobre las que calcular los agregados (ej., ['purchased', 'items']).
        agg_funcs (list): Lista de funciones de agregación a aplicar (ej., ['sum', 'mean']).

    Returns:
        pd.DataFrame: base_df con las nuevas columnas de features de ventana móvil.
    """
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()

    # Asegurar orden temporal para rolling calculations
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])

    print("\nGenerando Features de Ventana Móvil...")

    for window in windows:
        print(f"  Calculando para ventana de {window} semana(s)...")

        # Crear multi-indíce temporal para rolling por grupo
        hist_data_indexed = hist_data.set_index('date')

        # Agrupar por cliente y producto
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)

        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                print(f"    Generando: {col_name}")

                # --- Enfoque simplificado con shift en semanas ---
                # Calcular rolling sin shift
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                # Aplanar el multi-index resultante del groupby y rolling
                rolling_col = rolling_col.reset_index()
                 # Renombrar la columna resultante
                rolling_col = rolling_col.rename(columns={feature_col: col_name})

                # SHIFT para evitar Data Leakage
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(),
                                       on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)

                # Preparar para merge con base_df
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})

                # Unir a la tabla base
                df_with_rolling = pd.merge(
                    df_with_rolling,
                    rolling_col[['customer_id', 'product_id', 'target_week', col_name]],
                    on=['customer_id', 'product_id', 'target_week'],
                    how='left'
                )
                # Rellenar NaNs
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)


    return df_with_rolling

# --- Aplicación para generar Features de Ventana Móvil ---

# Definir ventanas, features y agregaciones
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

# --- Generar features de ventana móvil para Validation set ---
# Usar df_feat_val como base y df_train_hist_agg como historial
df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

# --- Generar features de ventana móvil para Test set ---
# Usar df_feat_test como base y df_train_val_hist_agg como historial
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)


# --- Mostrar Resultados ---
print("\nValidation Set con Features de Lag y Rolling (primeras filas):")
print(df_feat_val.head())
print("\nTest Set con Features de Lag y Rolling (primeras filas):")
print(df_feat_test.head())

# Verificar columnas nuevas y NaNs
print("\nColumnas en df_feat_val:")
print(df_feat_val.columns)
print("\nNaN check en df_feat_val:")
print(df_feat_val.isnull().sum().tail(15))
````

````python
# --- Añadir Features Estáticas ---

# 1. Features de Producto
print("\nAñadiendo features de producto...")
# Seleccionar columnas relevantes de df_productos
product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()

# Convertir a categóricas explícitamente
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

# Unir con df_feat_val
df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
# Unir con df_feat_test
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')

# 2. Features de Cliente
print("Añadiendo features de cliente...")
# Seleccionar columnas relevantes de df_clientes
client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()

# Convertir a categóricas
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

# Unir con df_feat_val
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
# Unir con df_feat_test
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

print("Features estáticas añadidas.")

# --- Añadir Features Temporales ---
print("\nAñadiendo features temporales...")

def add_temporal_features(df):
    df_temp = df.copy()
    # Convertir target_week a datetime para extraer componentes
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    # Extraer número de semana (ISO)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    # Extraer mes
    df_temp['month'] = df_temp['date'].dt.month.astype(int)

    # Eliminar columna de fecha auxiliar
    df_temp = df_temp.drop(columns=['date'])
    return df_temp

df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

print("Features temporales añadidas.")

# --- Revisión Final ---
print("\n--- Revisión Final de Features ---")
print("\nColumnas finales en df_feat_val:")
print(df_feat_val.columns)
print("\nTipos de datos en df_feat_val:")
print(df_feat_val.info())

print("\nPrimeras filas de df_feat_val con todas las features:")
print(df_feat_val.head())

# Verificar NaNs después de añadir features estáticas
print("\nNaN check final en df_feat_val:")
nan_check_val = df_feat_val.isnull().sum()
print(nan_check_val[nan_check_val > 0])

print("\nNaN check final en df_feat_test:")
nan_check_test = df_feat_test.isnull().sum()
print(nan_check_test[nan_check_test > 0])
````

````python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

print("--- Definiendo Pipelines de Preprocesamiento ---")

# 1. Identificar Tipos de Columnas para Features

# Columnas Numéricas
numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]

# Columnas Categóricas
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]

all_features = numeric_features + categorical_features
expected_cols = ['customer_id', 'product_id', 'target_week', 'target'] + all_features
missing_cols = set(expected_cols) - set(df_feat_val.columns)
extra_cols = set(df_feat_val.columns) - set(expected_cols)
if not missing_cols and not extra_cols:
    print("Verificación de columnas OK.")
elif missing_cols:
    print(f"WARN: Columnas faltantes en df_feat_val: {missing_cols}")
    # Attempt to use only existing columns if some are missing
    numeric_features = [f for f in numeric_features if f in df_feat_val.columns]
    categorical_features = [f for f in categorical_features if f in df_feat_val.columns]
    all_features = numeric_features + categorical_features
else:
    print(f"WARN: Columnas extras en df_feat_val: {extra_cols}")

# 2. Crear Pipeline Numérico
#    - Imputación: 'median' porque es más robusta a outliers que 'mean'.
#    - Escalado: StandardScaler centra los datos (media 0, std 1).
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
print("Pipeline numérico definido (Imputer(median) -> StandardScaler).")

# 3. Crear Pipeline Categórico
#    - Imputación: 'most_frequent' para categóricas
#    - Codificación: OneHotEncoder es estándar para variables nominales sin orden inherente.
#                   handle_unknown='ignore' evita errores si aparecen nuevas categorías en test.
#                   sparse_output=False, fácil de inspeccionar.
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])
print("Pipeline categórico definido (Imputer(most_frequent) -> OneHotEncoder).")


# 4. Combinar con ColumnTransformer
#    Aplica el transformer correspondiente a cada grupo de columnas.
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ],
    remainder='drop'
)

print("ColumnTransformer definido para combinar pipelines numérico y categórico.")
print("\nEstructura del Preprocesador:")
print(preprocessor)

# 5. Preparar los datos X e y
print("\n--- Preparando Conjuntos X, y ---")

# Columnas a usar como features
feature_columns = numeric_features + categorical_features

# Conjunto de Validación
X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()

# Conjunto de Test
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()
print(f"Dimensiones X_val: {X_val.shape}, y_val: {y_val.shape}")
print(f"Dimensiones X_test: {X_test.shape}, y_test: {y_test.shape}")

# 6. Probar el Preprocesador
print("\n--- Probando el Preprocesador (fit en val, transform en val y test) ---")
try:
    # Ajustar el preprocesador con los datos de validación/entrenamiento
    preprocessor.fit(X_val)
    print("Preprocessor 'fit' ejecutado en X_val.")

    # Transformar los datos de validación
    X_val_processed = preprocessor.transform(X_val)
    print("Preprocessor 'transform' ejecutado en X_val.")
    print(f"Dimensiones de X_val procesado: {X_val_processed.shape}")

    # Transformar los datos de test (usando el preprocesador ya ajustado)
    X_test_processed = preprocessor.transform(X_test)
    print("Preprocessor 'transform' ejecutado en X_test.")
    print(f"Dimensiones de X_test procesado: {X_test_processed.shape}")

    # Intentar obtener nombres de features después de la transformación
    try:
        feature_names_out = preprocessor.get_feature_names_out()
        print(f"\nNúmero de features después del preprocesamiento: {len(feature_names_out)}")
        print("Algunos nombres de features procesadas:")
        print(feature_names_out[:10])
        print("...")
        print(feature_names_out[-10:])

    except Exception as e:
        print(f"\nNo se pudieron obtener los nombres de las features procesadas: {e}")

except Exception as e:
    print(f"\nError durante la prueba del preprocesador: {e}")
````

````python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt # Para graficar la matriz de confusión

print("--- Construyendo y Evaluando Modelo Baseline ---")

# Primer clasificador
#    LogisticRegression por su simplicidad y ser un estándar como baseline.
#    solver='liblinear'.
#    random_state para reproducibilidad.
baseline_classifier = LogisticRegression(solver='liblinear', random_state=42, class_weight='balanced')
#    Se añade class_weight='balanced' para que el modelo preste más atención a la clase minoritaria
print("Clasificador Baseline: LogisticRegression(solver='liblinear', random_state=42, class_weight='balanced')")

# Crear el Pipeline Baseline Completo
baseline_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', baseline_classifier)
])
print("Pipeline Baseline creado (preprocessor -> classifier).")

# Entrenar el Pipeline Baseline
print("\nEntrenando el Pipeline Baseline con X_val y y_val...")
baseline_pipeline.fit(X_val, y_val)
print("Entrenamiento completado.")

# Evaluar el Pipeline Baseline en el conjunto de Validación (X_val)
print("\nEvaluando el Pipeline Baseline en X_val...")

# Predicciones de clase
y_pred_val = baseline_pipeline.predict(X_val)

# Predicciones de probabilidad (para AUC)
y_pred_proba_val = baseline_pipeline.predict_proba(X_val)[:, 1]

# Calcular Métricas
print("\n--- Resultados del Modelo Baseline en Validación ---")

# ROC AUC Score
try:
    roc_auc = roc_auc_score(y_val, y_pred_proba_val)
    print(f"ROC AUC Score: {roc_auc:.4f}")
except ValueError as e:
    print(f"No se pudo calcular ROC AUC: {e}")

# Classification Report
print("\nClassification Report:")
print(classification_report(y_val, y_pred_val, zero_division=0))

# Matriz de Confusión
print("\nMatriz de Confusión:")
cm = confusion_matrix(y_val, y_pred_val)
print(cm)

# Visualización de la Matriz de Confusión
try:
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=baseline_pipeline.classes_)
    disp.plot(cmap=plt.cm.Blues)
    plt.title("Matriz de Confusión - Baseline (Validación)")
    plt.show()
except Exception as e:
    print(f"\nNo se pudo graficar la matriz de confusión: {e}")

print("\n--- Evaluación Baseline Completada ---")
````

Esto ayudará a comprar el desemeño del modelo frente al modelo base e identifiar a un modelo candidato para realizar la optimizacón de hiperparámetros, el cual en última instancia será seleccionado para definir el objeto`final_pipeline`.

````python
# --- Importaciones Adicionales ---
from sklearn.tree import DecisionTreeClassifier
try:
    from lightgbm import LGBMClassifier
    lgbm_available = True
except ImportError:
    print("WARN: LightGBM no está instalado.")
    LGBMClassifier = None
    lgbm_available = False
import time
import pandas as pd
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

# --- Modelos a Probar ---
models_to_try = {}
# class_weight='balanced' para ayudar a los modelos con el desbalance inicial
models_to_try['Decision Tree'] = DecisionTreeClassifier(random_state=42, class_weight='balanced')
if lgbm_available:
    models_to_try['LightGBM'] = LGBMClassifier(random_state=42, n_jobs=-1, class_weight='balanced')

print(f"Modelos a probar: {list(models_to_try.keys())}")

# --- Almacenar Resultados ---
# Baseline para una comparación directa
results = {}
report_baseline_dict = classification_report(y_val, y_pred_val, output_dict=True, zero_division=0)
results['Logistic Regression (Baseline)'] = {
    'model_instance': baseline_pipeline.named_steps['classifier'],
    'pipeline': baseline_pipeline,
    'report': report_baseline_dict,
    'roc_auc': roc_auc_score(y_val, y_pred_proba_val),
    'confusion_matrix': confusion_matrix(y_val, y_pred_val),
    'training_time': None
}
print("Resultados del Baseline almacenados para comparación.")

# --- Bucle de Entrenamiento y Evaluación ---
for model_name, model_instance in models_to_try.items():
    print(f"\n--- Entrenando y Evaluando: {model_name} ---")

    current_pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', model_instance)
    ])

    start_time = time.time()
    current_pipeline.fit(X_val, y_val)
    end_time = time.time()
    training_time = end_time - start_time
    print(f"Entrenamiento completado en {training_time:.2f} segundos.")

    # Evaluar en el conjunto de validación
    print("Evaluando en X_val...")
    y_pred_current = current_pipeline.predict(X_val)
    y_pred_proba_current = current_pipeline.predict_proba(X_val)[:, 1]

    # Calcular Métricas
    current_roc_auc = roc_auc_score(y_val, y_pred_proba_current)
    current_report_dict = classification_report(y_val, y_pred_current, output_dict=True, zero_division=0)
    current_cm = confusion_matrix(y_val, y_pred_current)

    # Extraer y destacar el F1-Score para la clase 1 como métrica principal
    f1_class_1 = current_report_dict['1']['f1-score']
    print(f"\n**Métrica principal -> F1-Score (Clase 1 - Compra): {f1_class_1:.4f}**")
    print(f"ROC AUC Score (referencial): {current_roc_auc:.4f}")

    print("\nClassification Report completo:")
    print(classification_report(y_val, y_pred_current, zero_division=0))

    # Almacenar resultados
    results[model_name] = {
        'model_instance': model_instance,
        'pipeline': current_pipeline,
        'report': current_report_dict,
        'roc_auc': current_roc_auc,
        'confusion_matrix': current_cm,
        'training_time': training_time
    }
    print(f"Resultados de '{model_name}' almacenados.")

print("\n--- Todos los modelos han sido entrenados y evaluados ---")

# --- Resumen Comparativo de Métricas ---
metrics_summary = []
for name, res in results.items():
    if res:
        metrics_summary.append({
            'Model': name,
            'ROC_AUC': res['roc_auc'],
            'Precision_1': res['report']['1']['precision'],
            'Recall_1': res['report']['1']['recall'],
            'F1_1': res['report']['1']['f1-score'],
            'Accuracy': res['report']['accuracy'],
            'Training Time (s)': res['training_time']
        })

df_results = pd.DataFrame(metrics_summary)

print("\n--- Resumen Comparativo de Métricas (Ordenado por F1-Score en Clase 1) ---")
# Ordenar los resultados por la métrica de interés para facilitar la elección
print(df_results.sort_values(by='F1_1', ascending=False).to_string())
````

````python
# --- Importaciones (se completan importaciones anteriores) ---
import optuna
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from sklearn.pipeline import make_pipeline, Pipeline
import numpy as np
import warnings
from sklearn.impute import SimpleImputer # Assuming SimpleImputer is needed here
from lightgbm import LGBMClassifier # Assuming LGBMClassifier is needed here

print("--- Iniciando Optimización de Hiperparámetros con Optuna para LightGBM ---")

# Ocultar warnings de Optuna/Sklearn
warnings.filterwarnings('ignore', category=optuna.exceptions.ExperimentalWarning)
warnings.filterwarnings('ignore', category=UserWarning)

# --- Función Objetivo para Optuna ---
def objective(trial):
    # ---- Hiperparámetros del Preprocesador ----
    num_imputer_strategy = trial.suggest_categorical('num_imputer_strategy', ['mean', 'median'])
    numeric_imputer = SimpleImputer(strategy=num_imputer_strategy)

    scaler_choice = trial.suggest_categorical('scaler', ['StandardScaler', 'MinMaxScaler', 'RobustScaler', 'None'])
    if scaler_choice == 'StandardScaler':
        scaler = StandardScaler()
    elif scaler_choice == 'MinMaxScaler':
        scaler = MinMaxScaler()
    elif scaler_choice == 'RobustScaler':
        scaler = RobustScaler()
    else: # 'None'
        scaler = 'passthrough'

    numeric_transformer = make_pipeline(numeric_imputer, scaler) if scaler != 'passthrough' else make_pipeline(numeric_imputer)

    categorical_transformer = make_pipeline(
        SimpleImputer(strategy='most_frequent'),
        OneHotEncoder(handle_unknown='ignore', sparse_output=False)
    )

    preprocessor = ColumnTransformer(
        transformers=[
            ('num', numeric_transformer, numeric_features),
            ('cat', categorical_transformer, categorical_features)
        ],
        remainder='drop'
    )

    # ---- Hiperparámetros del Modelo (LGBMClassifier) ----
    lgbm_params = {
        'objective': 'binary',
        'metric': 'f1',
        'random_state': 42,
        'n_jobs': -1,
        'n_estimators': trial.suggest_int('n_estimators', 200, 1500, step=100),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 20, 100),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-3, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'is_unbalance': trial.suggest_categorical('is_unbalance', [True, False])
    }
    model = LGBMClassifier(**lgbm_params)

    # ---- Crear Pipeline ----
    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    # ---- Evaluación con Validación Cruzada ----
    cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    try:
        # Optimizar usando 'f1' score
        scores = cross_val_score(pipeline, X_val, y_val, scoring='f1', cv=cv, n_jobs=-1)
        f1_mean = np.mean(scores)
        return f1_mean

    except Exception as e:
        print(f"Trial fallido con error: {e}. Retornando 0.0")
        return 0.0

# --- Ejecutar Estudio Optuna ---
study_name = 'lgbm_f1_optimization_v1'
study = optuna.create_study(direction='maximize')

n_trials = 50
print(f"Iniciando optimización con {n_trials} trials para maximizar el F1-Score...")
study.optimize(objective, n_trials=n_trials, timeout=1800)

# Restaurar warnings
warnings.filterwarnings('default', category=optuna.exceptions.ExperimentalWarning)
warnings.filterwarnings('default', category=UserWarning)

# --- Resultados de la Optimización ---
print("\n--- Optimización Completada ---")
print(f"Número de trials finalizados: {len(study.trials)}")
best_trial = study.best_trial
print(f"Mejor trial:")
# El valor ahora es F1-Score
print(f"  Valor (F1-Score): {best_trial.value:.6f}")
print(f"  Mejores Hiperparámetros:")
for key, value in best_trial.params.items():
    print(f"    {key}: {value}")

# --- Visualizaciones de Optuna ---
try:
    fig1 = optuna.visualization.plot_optimization_history(study)
    fig1.update_layout(title="Historial de Optimización (F1-Score)")
    fig1.show()
    fig2 = optuna.visualization.plot_param_importances(study)
    fig2.update_layout(title="Importancia de Hiperparámetros para F1-Score")
    fig2.show()
except ImportError:
    print("\nWARN: Plotly no instalado. Omitiendo gráficos de Optuna.")
except Exception as e:
     print(f"\nWARN: No se pudieron generar gráficos de Optuna: {e}")

# --- Entrenar y Evaluar Modelo Final con Mejores Hiperparámetros ---
print("\n--- Entrenando Modelo Final con Mejores Hiperparámetros ---")

# Obtener los mejores parámetros y construir el pipeline final
best_params = best_trial.params
num_imputer_final = SimpleImputer(strategy=best_params['num_imputer_strategy'])

scaler_choice_final = best_params['scaler']
if scaler_choice_final == 'StandardScaler':
    scaler_final = StandardScaler()
elif scaler_choice_final == 'MinMaxScaler':
    scaler_final = MinMaxScaler()
elif scaler_choice_final == 'RobustScaler':
    scaler_final = RobustScaler()
else:
    scaler_final = 'passthrough'

numeric_transformer_final = make_pipeline(num_imputer_final, scaler_final) if scaler_final != 'passthrough' else make_pipeline(num_imputer_final)
categorical_transformer_final = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
preprocessor_final = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_final, numeric_features),
        ('cat', categorical_transformer_final, categorical_features)
    ],
    remainder='drop'
)
lgbm_best_params = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_final = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params)

final_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_final),
    ('classifier', lgbm_final)
])

# Entrenar en el conjunto de validación completo
print("Entrenando pipeline final en X_val completo...")
final_pipeline.fit(X_val, y_val)
print("Entrenamiento final completado.")

# --- Evaluación Final del Modelo Optimizado ---
print("\n--- Evaluación del Modelo Optimizado en X_val ---")
y_pred_val_opt = final_pipeline.predict(X_val)
y_pred_proba_val_opt = final_pipeline.predict_proba(X_val)[:, 1]
roc_auc_val_opt = roc_auc_score(y_val, y_pred_proba_val_opt)
print(f"ROC AUC Score (X_val): {roc_auc_val_opt:.4f} (Métrica referencial)")
print("Classification Report (X_val):")
print(classification_report(y_val, y_pred_val_opt, zero_division=0))

print("\n--- Evaluación del Modelo Optimizado en X_test (Rendimiento Final) ---")
y_pred_test_opt = final_pipeline.predict(X_test)
y_pred_proba_test_opt = final_pipeline.predict_proba(X_test)[:, 1]
roc_auc_test_opt = roc_auc_score(y_test, y_pred_proba_test_opt)
print(f"ROC AUC Score (X_test): {roc_auc_test_opt:.4f} (Métrica referencial)")
print("Classification Report (X_test):")
print(classification_report(y_test, y_pred_test_opt, zero_division=0))
print("Matriz de Confusión (X_test):")
cm_test = confusion_matrix(y_test, y_pred_test_opt)
print(cm_test)

# Visualización Matriz Test
try:
    disp_test = ConfusionMatrixDisplay(confusion_matrix=cm_test, display_labels=final_pipeline.classes_)
    disp_test.plot(cmap=plt.cm.Blues)
    plt.title("Matriz de Confusión - Modelo Optimizado (Test)")
    plt.show()
except Exception as e:
    print(f"\nNo se pudo graficar la matriz de confusión de test: {e}")
````

````python
from sklearn.pipeline import Pipeline

print("--- Confirmando disponibilidad del modelo final ---\n")

try:
    # 1. Check if the `final_pipeline` object exists
    if 'final_pipeline' in locals() or 'final_pipeline' in globals():
        print("1. El objeto `final_pipeline` existe en el entorno.")

        # 2. Verify that `final_pipeline` is an instance of `sklearn.pipeline.Pipeline`
        if isinstance(final_pipeline, Pipeline):
            print("2. `final_pipeline` es una instancia de `sklearn.pipeline.Pipeline`.")

            # 3. Confirm that the `final_pipeline` object has been fitted (trained)
            # Scikit-learn estimators usually have a `_is_fitted` attribute or checkable internal state.
            # For a pipeline, we can check if its steps are fitted.
            fitted_status = True
            for step_name, step_estimator in final_pipeline.steps:
                # Check if the estimator has a _is_fitted method or a similar attribute
                if hasattr(step_estimator, 'fit') and not hasattr(step_estimator, 'classes_') and not hasattr(step_estimator, 'n_features_in_'):
                    # This is a heuristic for checking if an estimator within a pipeline is fitted
                    # For transformers like preprocessor, checking for 'n_features_in_' is common.
                    # For classifiers, checking for 'classes_' is common.
                    # If a step is meant to be fitted and doesn't have these after fit, it might not be fitted.
                    # A more robust check might involve trying to transform/predict or checking specific attributes.
                    # For this context, assuming if the previous fitting steps ran without error, it's fitted.
                    pass # Relying on previous execution, as direct `_is_fitted` isn't universal or easily accessible for all pipeline components.
                elif hasattr(step_estimator, 'n_features_in_') and not hasattr(step_estimator, '_sklearn_version'):
                     # Check for fitted transformers in the preprocessor
                     if step_name == 'preprocessor':
                         # Check if the ColumnTransformer itself has been fitted
                         if not hasattr(step_estimator, 'n_features_in_'):
                             fitted_status = False
                             break
                     else:
                        # For other simple estimators, this might indicate not fitted
                        fitted_status = False
                        break

            # A more direct check for an sklearn pipeline is to check if its last step is fitted
            # or if it has internal attributes set during fit (like classes_ for classifiers).
            if hasattr(final_pipeline.named_steps['classifier'], 'classes_') or hasattr(final_pipeline.named_steps['preprocessor'], 'n_features_in_'):
                 print("3. El objeto `final_pipeline` parece haber sido entrenado (fitted) ya que los componentes internos muestran signos de haber sido ajustados.")
            else:
                print("3. No se puede confirmar el estado 'fitted' de `final_pipeline` de manera concluyente a través de sus atributos internos. (Basado en heurísticas).")

        else:
            print("2. ERROR: `final_pipeline` NO es una instancia de `sklearn.pipeline.Pipeline`.")

    else:
        print("1. ERROR: El objeto `final_pipeline` NO se encuentra en el entorno actual. Por favor, asegúrese de ejecutar las celdas previas.")

except NameError:
    print("ERROR: El objeto `final_pipeline` no está definido. Por favor, ejecute las celdas donde se define y entrena.")
except Exception as e:
    print(f"Ocurrió un error inesperado al verificar `final_pipeline`: {e}")
````

### Resumen Detallado del Proyecto

El proyecto tuvo como objetivo desarrollar un modelo predictivo para estimar la probabilidad semanal de compra de cada producto por cada cliente activo de **SodAI Drinks**. El trabajo se ha estructurado en varias etapas clave:

1.  **Carga de Datos**:
    *   Inicialmente, se cargaron tres conjuntos de datos en formato Parquet: `transacciones.parquet`, `clientes.parquet` y `productos.parquet`. Estos DataFrames contenían el historial de compras, las características demográficas y de comportamiento de los clientes, y los atributos descriptivos de los productos, respectivamente.

2.  **Preprocesamiento Inicial**:
    *   La columna `purchase_date` del DataFrame de transacciones se convirtió a formato de fecha y se agregó a nivel diario. Esta agregación incluyó la suma neta de la columna `items` para consolidar compras y devoluciones del mismo día, asegurando que solo las transacciones efectivas (con `items_neto > 0`) fueran consideradas. Posteriormente, se crearon las columnas `year` y `week` a partir de `purchase_date`. Finalmente, los tres DataFrames se fusionaron (`df_merged`) utilizando `customer_id` y `product_id` para centralizar toda la información de transacciones con sus respectivos atributos de cliente y producto. Se confirmó la ausencia de valores nulos o duplicados tras esta fase.

3.  **División Holdout**:
    *   Se implementó una estrategia de división holdout temporal para los datos, fundamental para evitar la fuga de información y asegurar una evaluación realista del modelo. El conjunto total de semanas (`year_week`) se dividió en tres partes:
        *   **Entrenamiento**: Semanas `2024-W01` a `2024-W45` (45 semanas).
        *   **Validación**: Semanas `2024-W46` a `2024-W49` (4 semanas).
        *   **Prueba**: Semanas `2024-W50` a `2025-W01` (4 semanas).
    *   Esta partición generó `df_train`, `df_val` y `df_test`, cada uno con datos de transacciones de sus respectivas ventanas temporales.

4.  **Ingeniería de Características**:
    *   Se crearon DataFrames base (`df_base_val`, `df_base_test`) que incluían todas las combinaciones posibles de `customer_id`, `product_id` y `target_week` para las semanas de validación y prueba, junto con una variable `target` binaria (1 si hubo compra, 0 si no). La creación de `df_train_val` (concatenación de `df_train` y `df_val`) fue crucial para generar características históricas para el conjunto de prueba.
    *   **Features de Lag**: Se calcularon `lag_N_purchased` y `lag_N_items` para N = 1, 2, 3, 4 semanas anteriores a la semana objetivo, indicando si hubo una compra y la cantidad de ítems comprados. Esto se hizo de forma agregada a nivel semanal.
    *   **Features de Ventana Móvil**: Se calcularon sumas y promedios de `purchased` e `items` en ventanas de 4 y 8 semanas (`rolling_4w_sum/mean_purchased/items`, `rolling_8w_sum/mean_purchased/items`), para capturar tendencias más amplias. Se aplicó un `shift` para evitar el *data leakage*.
    *   **Features Estáticas**: Se añadieron características de `df_productos` (`brand`, `package`, `sub_category`) y `df_clientes` (`customer_type`).
    *   **Features Temporales**: Se extrajeron `week_of_year` y `month` de la `target_week` para capturar estacionalidad.

5.  **Definición del Preprocesador**:
    *   Se definió un `ColumnTransformer` (`preprocessor`) dentro de un pipeline para gestionar el preprocesamiento de características. Este incluía:
        *   Un pipeline numérico con `SimpleImputer(strategy='median')` y `StandardScaler()` para las características numéricas.
        *   Un pipeline categórico con `SimpleImputer(strategy='most_frequent')` y `OneHotEncoder(handle_unknown='ignore', sparse_output=False)` para las características categóricas.
    *   El preprocesador fue ajustado (`fit`) con los datos de `X_val` y luego usado para transformar `X_val` y `X_test`, asegurando consistencia.

6.  **Selección de Modelo**:
    *   Se entrenaron y evaluaron varios modelos en el conjunto de validación (`X_val`) utilizando el `preprocessor`:
        *   **Logistic Regression (Baseline)**: Obtuvo un F1-Score de 0.3819 para la clase de compra.
        *   **Decision Tree**: Mejoró el F1-Score a 0.4384.
        *   **LightGBM (parámetros por defecto)**: Alcanzó un F1-Score de 0.4079.
    *   El **F1-Score para la clase 1 (compra)** fue la métrica principal para la selección debido al desbalance de clases. Aunque el Decision Tree tuvo el mejor F1-Score con parámetros por defecto, LightGBM fue elegido para la optimización debido a su mayor eficiencia y potencial de rendimiento.

7.  **Optimización de Hiperparámetros**:
    *   Se utilizó `Optuna` para optimizar los hiperparámetros del `LGBMClassifier` y del preprocesador (estrategia de imputación y tipo de escalador). El objetivo de la optimización fue maximizar el F1-Score para la clase positiva a través de validación cruzada estratificada (`StratifiedKFold`) en `X_val`.
    *   La optimización, tras 50 trials, identificó un conjunto de hiperparámetros óptimos que resultaron en un F1-Score de 0.4606 en validación cruzada.

8.  **Modelo Final**:
    *   Se construyó el `final_pipeline` utilizando el `preprocessor` y el `LGBMClassifier` con los mejores hiperparámetros encontrados por Optuna. Este pipeline final fue entrenado en el conjunto de validación completo (`X_val`).
    *   La evaluación final del modelo optimizado en el conjunto de prueba (`X_test`) arrojó un **F1-Score de 0.49** para la clase de compra y un **ROC AUC de 0.9125**. Esto representó una mejora significativa sobre el baseline y los modelos no optimizados.

9.  **Generación de Predicciones**:
    *   Para la semana objetivo `2025-W01`, se filtró el `df_feat_test` para obtener `X_predict_2025_W01`, conteniendo 99,900 combinaciones cliente-producto.
    *   El `final_pipeline` se utilizó para generar las probabilidades de compra para `X_predict_2025_W01` utilizando el método `predict_proba()`.
    *   Finalmente, las probabilidades predichas se combinaron con los `customer_id` y `product_id` correspondientes, se ordenaron de mayor a menor y se mostraron las 20 principales predicciones, proporcionando una lista priorizada de oportunidades de venta.

#### Encontrar Punto de Corte Óptimo para F1-Score

Utilizar el modelo optimizado (`final_pipeline`) para predecir las probabilidades de compra en el conjunto de validación (`X_val`). Luego, se iterará a través de un rango de posibles puntos de corte (thresholds) y se calculará el F1-Score para cada uno. El punto de corte que maximice el F1-Score será seleccionado para convertir las probabilidades en predicciones binarias.


````python
from sklearn.metrics import f1_score
import numpy as np

print("--- Encontrando el punto de corte (threshold) óptimo para F1-Score ---")

# 1. Predecir las probabilidades de compra en el conjunto de validación (X_val)
y_pred_proba_val = final_pipeline.predict_proba(X_val)[:, 1]
print(f"Probabilidades predichas para X_val generadas. Dimensión: {y_pred_proba_val.shape}")

# 2. Crear un rango de posibles puntos de corte (thresholds)
thresholds = np.arange(0.0, 1.01, 0.01)

# Inicializar variables para almacenar el mejor F1-Score y su umbral
best_f1_score = 0
optimal_threshold = 0

# 3. y 4. Iterar a través de los umbrales y calcular el F1-Score
print("Calculando F1-Score para diferentes puntos de corte...")
for threshold in thresholds:
    y_pred_binary = (y_pred_proba_val >= threshold).astype(int)
    f1 = f1_score(y_val, y_pred_binary, pos_label=1, zero_division=0)

    if f1 > best_f1_score:
        best_f1_score = f1
        optimal_threshold = threshold

# 6. Imprimir el F1-Score máximo encontrado y el umbral óptimo correspondiente
print(f"\nProceso de búsqueda de umbral completado.")
print(f"F1-Score máximo para la clase positiva: {best_f1_score:.4f}")
print(f"Umbral óptimo correspondiente: {optimal_threshold:.2f}")

print("--- Búsqueda de Umbral Óptimo Finalizada ---")
````

#### Generar Predicciones y Aplicar Punto de Corte para 01/01/25 - 05/01/25 (2025-W01)

Generar predicciones binarias de compra para la semana '2025-W01' utilizando el modelo optimizado y el punto de corte óptimo de F1-Score.


````python
print("--- Generando predicciones binarias de compra para 2025-W01 ---")

# El `predicted_purchase_proba` ya fue calculado en un paso anterior
# y `predictions_df` ya contiene 'customer_id', 'product_id' y 'predicted_proba'.

# 1. Aplicar el `optimal_threshold` (0.32) para convertir probabilidades a predicciones binarias
# Se asume que `optimal_threshold` ya está definido desde la búsqueda del umbral óptimo.
predictions_df['predicted_purchase_binary'] = (predictions_df['predicted_proba'] >= optimal_threshold).astype(int)

# 2. Mostrar las top N predicciones con la nueva columna binaria
N = 20
print(f"\nTop {N} predicciones de compra binarias para la semana '2025-W01' (con umbral {optimal_threshold:.2f}):")
print(predictions_df.head(N).to_string())

# 3. Mostrar la distribución de las predicciones binarias (opcional, para verificación)
print("\nDistribución de 'predicted_purchase_binary':")
print(predictions_df['predicted_purchase_binary'].value_counts(normalize=True))
````

#### Cargar y Preprocesar Compras Reales de batch_t1

Cargar el archivo `batch_t1.parquet` y preprocesarlo para extraer las compras efectivas de cliente-producto dentro del período 01/01/25 - 05/01/25, que servirá como la verdad fundamental (ground truth) para evaluar las predicciones.


````python
import pandas as pd

print("--- Cargando y preprocesando compras reales de batch_t1 ---")

# 1. Cargar el archivo `batch_t1.parquet`
try:
    df_batch_t1_raw = pd.read_parquet('data/batch_t1.parquet')
    print(f"Archivo batch_t1.parquet cargado exitosamente. Dimensiones: {df_batch_t1_raw.shape}")
except FileNotFoundError:
    print("Error: El archivo data/batch_t1.parquet no se encontró.")
    exit() # Terminate if file not found, as subsequent steps depend on it.

# 2. Convertir la columna `purchase_date` a tipo datetime y extraer solo la fecha
df_batch_t1_raw['purchase_date'] = pd.to_datetime(df_batch_t1_raw['purchase_date']).dt.date

# 3. Agrupar el DataFrame resultante por customer_id, product_id, y la fecha de compra, sumando la columna items para obtener un items_neto
df_batch_t1_agg = df_batch_t1_raw.groupby(
    ['customer_id', 'product_id', 'purchase_date']
).agg(
    items_neto=('items', 'sum')
).reset_index()
print(f"Transacciones de batch_t1 agregadas a nivel diario. Dimensiones: {df_batch_t1_agg.shape}")

# 4. Filtrar para incluir solo las filas donde items_neto sea mayor que cero (compras efectivas)
df_batch_t1_effective_purchases = df_batch_t1_agg[df_batch_t1_agg['items_neto'] > 0].copy()
print(f"Compras efectivas filtradas. Dimensiones: {df_batch_t1_effective_purchases.shape}")

# 5. Convertir la columna `purchase_date` a formato datetime y extraer year ISO y week ISO
df_batch_t1_effective_purchases['purchase_date'] = pd.to_datetime(df_batch_t1_effective_purchases['purchase_date'])
df_batch_t1_effective_purchases['year'] = df_batch_t1_effective_purchases['purchase_date'].dt.isocalendar().year
df_batch_t1_effective_purchases['week'] = df_batch_t1_effective_purchases['purchase_date'].dt.isocalendar().week

# 6. Crear la columna `year_week`
df_batch_t1_effective_purchases['year_week'] = df_batch_t1_effective_purchases['year'].astype(str) + '-W' + df_batch_t1_effective_purchases['week'].astype(str).str.zfill(2)

# 7. Filtrar para la semana '2025-W01'
ground_truth_2025_W01 = df_batch_t1_effective_purchases[df_batch_t1_effective_purchases['year_week'] == '2025-W01'].copy()
print(f"Ground truth para 2025-W01 filtrado. Dimensiones: {ground_truth_2025_W01.shape}")

# 8. Crear un conjunto (set) de tuplas (customer_id, product_id)
ground_truth_pairs_2025_W01 = set(map(tuple, ground_truth_2025_W01[['customer_id', 'product_id']].values))

# 9. Imprimir el número de pares cliente-producto ónicos y mostrar las primeras 5 filas
print(f"\nNúmero de pares cliente-producto ónicos en ground_truth_pairs_2025_W01: {len(ground_truth_pairs_2025_W01)}")
print("\nPrimeras 5 filas de ground_truth_2025_W01:")
print(ground_truth_2025_W01.head())
````

#### Evaluar Predicciones de 2025-W01 vs. Compras Reales de batch_t1

Cargar las predicciones binarias previamente generadas para '2025-W01' y compararlas con las compras reales de `batch_t1` preprocesadas, calculando métricas de rendimiento.


````python
print("--- Evaluando Predicciones de 2025-W01 ---\n")

# 1. Obtener los pares (customer_id, product_id) de las predicciones binarias positivas
# Asegurarse de que predictions_df existe y tiene la columna binaria
if 'predictions_df' not in locals() or predictions_df.empty:
    print("Error: 'predictions_df' no está disponible o está vacío. No se puede evaluar.")
else:
    predicted_positive_pairs = set(tuple(row) for row in predictions_df[
        predictions_df['predicted_purchase_binary'] == 1
    ][['customer_id', 'product_id']].to_numpy())

    # Asegurarse de que ground_truth_pairs_2025_W01 existe
    if 'ground_truth_pairs_2025_W01' not in locals():
        print("Error: 'ground_truth_pairs_2025_W01' no está disponible. No se puede evaluar.")
    else:
        # 2. Calcular Verdaderos Positivos (TP), Falsos Positivos (FP) y Falsos Negativos (FN)
        tp = len(predicted_positive_pairs.intersection(ground_truth_pairs_2025_W01))
        fp = len(predicted_positive_pairs - ground_truth_pairs_2025_W01)
        fn = len(ground_truth_pairs_2025_W01 - predicted_positive_pairs)

        # 3. Calcular Precisión, Recall y F1-Score
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

        # 4. Imprimir los valores calculados
        print(f"Verdaderos Positivos (TP): {tp}")
        print(f"Falsos Positivos (FP): {fp}")
        print(f"Falsos Negativos (FN): {fn}")
        print(f"\n--- Métricas de Rendimiento (Clase Positiva) ---")
        print(f"Precisión: {precision:.4f}")
        print(f"Recall: {recall:.4f}")
        print(f"F1-Score: {f1:.4f}")

print("\n--- Evaluación Completada ---")
````

#### Resumen de la Evaluación de Predicciones para '2025-W01'

Las predicciones binarias generadas para la semana '2025-W01' (utilizando el umbral óptimo de F1-Score de 0.32) han sido evaluadas contra las compras reales (`ground_truth_pairs_2025_W01`) obtenidas del preprocesamiento de `batch_t1.parquet`.

**Resultados Clave de la Evaluación:**

*   **Verdaderos Positivos (TP):** 612
    *   Estas son las combinaciones cliente-producto que el modelo predijo correctamente como compra y que realmente ocurrieron.
*   **Falsos Positivos (FP):** 631
    *   Estas son las combinaciones que el modelo predijo como compra, pero que no ocurrieron en la realidad. Un alto número de FP puede llevar a esfuerzos de venta o marketing ineficientes.
*   **Falsos Negativos (FN):** 2652
    *   Estas son las combinaciones que el modelo predijo incorrectamente como no compra, pero que sí ocurrieron. Un alto número de FN significa oportunidades de venta perdidas.

**Métricas de Rendimiento (para la Clase Positiva 'Compra'):**

*   **Precisión:** 0.4924
    *   De todas las predicciones de compra realizadas por el modelo, aproximadamente el 49.24% fueron correctas. Esto indica que casi la mitad de las recomendaciones del modelo resultarían en una compra real.
*   **Recall:** 0.1875
    *   El modelo fue capaz de identificar aproximadamente el 18.75% de todas las compras reales que ocurrieron. Un recall bajo sugiere que el modelo está perdiendo una cantidad significativa de compras reales.
*   **F1-Score:** 0.2716
    *   Esta métrica, que es la media armónica de precisión y recall, muestra el equilibrio general del modelo en la identificación de la clase positiva. El valor de 0.2716 es bajo, reflejando el compromiso entre una precisión moderada y un recall bajo.

### Conclusión de la Evaluación:

El modelo LightGBM optimizado, al aplicar el umbral de 0.32, muestra una **Precisión** aceptable, lo que es bueno para no molestar a los clientes con recomendaciones irrelevantes. Sin embargo, su **Recall** es relativamente bajo. Esto significa que, aunque las predicciones positivas son razonablemente fiables, el modelo aún no está capturando una gran parte de las compras reales que ocurren. Para el negocio, esto implica que se identificarán oportunidades de venta con cierta confianza, pero se seguirán perdiendo muchas otras.

Integrar las compras efectivas de `batch_t1.parquet` con el DataFrame histórico consolidado (`df_merged`) para actualizar el conjunto de datos históricos.


````python
print("--- Integrando compras efectivas de batch_t1 en el historial consolidado ---")

# 1. Renombrar 'items_neto' a 'items' en df_batch_t1_effective_purchases
df_batch_t1_effective_purchases_renamed = df_batch_t1_effective_purchases.rename(columns={'items_neto': 'items'})
print("Columna 'items_neto' renombrada a 'items' en df_batch_t1_effective_purchases.")

# 2. Expandir df_batch_t1_effective_purchases_renamed con información de clientes y productos
# Se asume que df_clientes y df_productos están disponibles desde pasos anteriores.

df_batch_t1_expanded = pd.merge(
    df_batch_t1_effective_purchases_renamed,
    df_clientes,
    on='customer_id',
    how='left'
)
df_batch_t1_expanded = pd.merge(
    df_batch_t1_expanded,
    df_productos,
    on='product_id',
    how='left'
)
print("df_batch_t1 expandido con features de clientes y productos.")

# 3. Asegurarse de que las columnas y el orden coincidan con df_merged
# Obtener la lista de columnas de df_merged
merged_cols = df_merged.columns.tolist()

# Filtrar df_batch_t1_expanded para que tenga solo las columnas presentes en df_merged
# y reordenarlas para que coincidan
df_batch_t1_aligned = df_batch_t1_expanded[merged_cols].copy()
print("Columnas de df_batch_t1_expanded alineadas con df_merged.")

# 4. Concatenar df_merged y df_batch_t1_aligned para crear df_merged_updated
df_merged_updated = pd.concat([df_merged, df_batch_t1_aligned], ignore_index=True)
print(f"DataFrames concatenados. Dimensiones iniciales: {df_merged_updated.shape}")

# 5. Eliminar duplicados si los hubiera (en caso de solapamiento de fechas, aunque aquí no debería haber) y ordenar
df_merged_updated.drop_duplicates(inplace=True)
df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
df_merged_updated.reset_index(drop=True, inplace=True)
print(f"Duplicados eliminados y DataFrame ordenado. Dimensiones finales: {df_merged_updated.shape}")

# 6. Verificar las dimensiones y el rango de fechas de df_merged_updated
print(f"\nNuevas dimensiones de df_merged_updated: {df_merged_updated.shape}")
print(f"Rango de fechas de df_merged_updated: {df_merged_updated['purchase_date'].min().strftime('%Y-%m-%d')} a {df_merged_updated['purchase_date'].max().strftime('%Y-%m-%d')}")

print("--- Integración de batch_t1 completada ---")
````

#### Re-dividir Datos con Historial Actualizado

Redefinir las particiones de entrenamiento, validación y prueba (`df_train`, `df_val`, `df_test`) utilizando el `df_merged_updated` que incluye el historial de `batch_t1`.


````python
print("--- Redefiniendo particiones de datos con historial actualizado ---")

# 1. Obtener las semanas totales únicas de `df_merged_updated` y ordénelas.
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_total_updated = len(weeks_total_updated)

# 2. Mantener las definiciones de num_weeks_test y num_weeks_val como 4 semanas cada una.
num_weeks_test = 4
num_weeks_val = 4

# 3. Calcular los índices de inicio para los conjuntos de validación y prueba
idx_val_start_updated = num_weeks_total_updated - num_weeks_test - num_weeks_val
idx_test_start_updated = num_weeks_total_updated - num_weeks_test

# 4. Crear las listas de semanas para entrenamiento, validación y prueba
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:]

print(f"Semanas para Entrenamiento actualizadas ({len(weeks_train)}): {weeks_train[0]} a {weeks_train[-1]}")
print(f"Semanas para Validación actualizadas ({len(weeks_val)}): {weeks_val[0]} a {weeks_val[-1]}")
print(f"Semanas para Prueba actualizadas ({len(weeks_test)}): {weeks_test[0]} a {weeks_test[-1]}")

# 5. Generar los nuevos DataFrames df_train, df_val y df_test
df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()

# 6. Imprimir dimensiones y rangos de fechas para verificación
print(f"\nDimensiones df_train actualizado: {df_train.shape}")
print(f"Fechas en df_train actualizado: {df_train['purchase_date'].min().strftime('%Y-%m-%d')} a {df_train['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_val actualizado: {df_val.shape}")
print(f"Fechas en df_val actualizado: {df_val['purchase_date'].min().strftime('%Y-%m-%d')} a {df_val['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_test actualizado: {df_test.shape}")
print(f"Fechas en df_test actualizado: {df_test['purchase_date'].min().strftime('%Y-%m-%d')} a {df_test['purchase_date'].max().strftime('%Y-%m-%d')}")

print("--- Redefinición de particiones completada ---")
````

#### Re-generar Feature Engineering

recalcular todos los atributos (lags, rolling windows, static, temporal) para los set de datos de entrenamiento, validación y testeo, inluyendo actualizar `batch_t1`.


````python
print("--- Iniciando regeneración de Feature Engineering con historial actualizado ---")

# 1. Recrear el df_train_val DataFrame con los datos actualizados
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
print(f"df_train_val recreado. Dimensiones: {df_train_val.shape}")

# 2. Recrear df_train_hist_agg (historial para el conjunto de validación)
df_train_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index()
df_train_agg['purchased'] = 1 # Añadir indicador de compra
df_train_hist_agg = df_train_agg.copy()
print(f"df_train_hist_agg recreado. Dimensiones: {df_train_hist_agg.shape}")

# 3. Recrear df_train_val_hist_agg (historial para el conjunto de prueba)
df_train_val_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index()
df_train_val_agg['purchased'] = 1 # Añadir indicador de compra
df_train_val_hist_agg = df_train_val_agg.copy()
print(f"df_train_val_hist_agg recreado. Dimensiones: {df_train_val_hist_agg.shape}")

# 4. Recrear df_base_val y df_base_test
# Se asume que `create_target_base` está definido en el contexto
# `target_weeks_val` y `target_weeks_test` se deben haber actualizado previamente en el paso de Holdout
print("\n--- Recreando bases para validación y prueba ---")
df_base_val = create_target_base(df_val, weeks_val) # Usar df_val y las semanas actualizadas
df_base_test = create_target_base(df_test, weeks_test) # Usar df_test y las semanas actualizadas

print("\n--- Aplicando Features de Lag ---")
# 5. Aplicar add_lag_features a df_base_val
df_feat_val = add_lag_features(
    base_df=df_base_val,
    transactions_hist_df=df_train_hist_agg, # Historial para validar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

# 6. Aplicar add_lag_features a df_base_test
df_feat_test = add_lag_features(
    base_df=df_base_test,
    transactions_hist_df=df_train_val_hist_agg, # Historial para probar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

print("\n--- Aplicando Features de Ventana Móvil ---")
# 7. Aplicar add_rolling_features a df_feat_val
df_feat_val = add_rolling_features(
    base_df=df_feat_val,
    transactions_hist_agg_df=df_train_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

# 8. Aplicar add_rolling_features a df_feat_test
df_feat_test = add_rolling_features(
    base_df=df_feat_test,
    transactions_hist_agg_df=df_train_val_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

print("\n--- Añadiendo Features Estáticas y Temporales ---")
# 9. Mergear features de producto con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')

# 10. Mergear features de cliente con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

# 11. Aplicar add_temporal_features a df_feat_val y df_feat_test
df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

print("--- Verificación de Features Regeneradas ---")
print(f"\nDimensiones de df_feat_val: {df_feat_val.shape}")
print("Primeras 5 filas de df_feat_val:")
print(df_feat_val.head().to_string())
print("\nNaN check en df_feat_val:")
nan_check_val = df_feat_val.isnull().sum()
print(nan_check_val[nan_check_val > 0].to_string())

print(f"\nDimensiones de df_feat_test: {df_feat_test.shape}")
print("Primeras 5 filas de df_feat_test:")
print(df_feat_test.head().to_string())
print("\nNaN check en df_feat_test:")
nan_check_test = df_feat_test.isnull().sum()
print(nan_check_test[nan_check_test > 0].to_string())

# 12. Actualizar X_val, y_val, X_test, y_test
X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()

print("\n--- Feature Engineering regenerado y conjuntos X, y actualizados ---")
````

#### Reentrenar el Modelo LightGBM Optimizado

Reentrenar el `final_pipeline` (el modelo LightGBM con los hiperparámetros óptimos ya identificados) utilizando los nuevos datos de entrenamiento (`X_val` y `y_val` generados con el historial actualizado).


````python
print("--- Reentrenando el Modelo LightGBM Optimizado ---")

# Obtener los mejores parámetros de la optimización anterior
best_params = study.best_trial.params

# 1. Recrear el preprocesador con los mejores parámetros
num_imputer_strategy_final = best_params['num_imputer_strategy']
numeric_imputer_re_fit = SimpleImputer(strategy=num_imputer_strategy_final)

scaler_choice_re_fit = best_params['scaler']
if scaler_choice_re_fit == 'StandardScaler':
    scaler_re_fit = StandardScaler()
elif scaler_choice_re_fit == 'MinMaxScaler':
    scaler_re_fit = MinMaxScaler()
elif scaler_choice_re_fit == 'RobustScaler':
    scaler_re_fit = RobustScaler()
else:
    scaler_re_fit = 'passthrough'

numeric_transformer_re_fit = make_pipeline(numeric_imputer_re_fit, scaler_re_fit) if scaler_re_fit != 'passthrough' else make_pipeline(numeric_imputer_re_fit)
categorical_transformer_re_fit = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))

preprocessor_re_fit = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_re_fit, numeric_features),
        ('cat', categorical_transformer_re_fit, categorical_features)
    ],
    remainder='drop'
)
print("Preprocesador recreado con los mejores parámetros.")

# 2. Recrear el clasificador LightGBM con los mejores parámetros
lgbm_best_params_re_fit = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_re_fit = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params_re_fit)
print("Clasificador LightGBM recreado con los mejores parámetros.")

# 3. Crear el final_pipeline recreado
final_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_re_fit),
    ('classifier', lgbm_re_fit)
])
print("Pipeline final recreado.")

# 4. Entrenar el final_pipeline con los datos de entrenamiento actualizados (X_val y y_val)
print("Entrenando el pipeline final con X_val (historial actualizado) y y_val...")
final_pipeline.fit(X_val, y_val)
print("Reentrenamiento completado con éxito.")

# Opcional: Evaluar rápidamente el modelo reentrenado en X_val
from sklearn.metrics import classification_report
y_pred_val_re_fit = final_pipeline.predict(X_val)
print("\nClassification Report para X_val (después del reentrenamiento):")
print(classification_report(y_val, y_pred_val_re_fit, zero_division=0))
````

Proporcionar un resumen de la evaluación de las predicciones binarias de compra para la semana '2025-W01' contra las compras reales del `batch_t1`.

### Resumen de la Evaluación

Las predicciones binarias generadas para la semana '2025-W01' (utilizando el umbral óptimo de F1-Score de 0.32) han sido evaluadas contra las compras reales (`ground_truth_pairs_2025_W01`) obtenidas del preprocesamiento de `batch_t1.parquet`.

**Resultados Clave de la Evaluación:**

*   **Verdaderos Positivos (TP):** 612
    *   Estas son las combinaciones cliente-producto que el modelo predijo correctamente como compra y que realmente ocurrieron.
*   **Falsos Positivos (FP):** 631
    *   Estas son las combinaciones que el modelo predijo como compra, pero que no ocurrieron en la realidad. Un alto número de FP puede llevar a esfuerzos de venta o marketing ineficientes.
*   **Falsos Negativos (FN):** 2652
    *   Estas son las combinaciones que el modelo predijo incorrectamente como no compra, pero que sí ocurrieron. Un alto número de FN significa oportunidades de venta perdidas.

**Métricas de Rendimiento (para la Clase Positiva 'Compra'):**

*   **Precisión:** 0.4924
    *   De todas las predicciones de compra realizadas por el modelo, aproximadamente el 49.24% fueron correctas. Esto indica que casi la mitad de las recomendaciones del modelo resultarían en una compra real.
*   **Recall:** 0.1875
    *   El modelo fue capaz de identificar aproximadamente el 18.75% de todas las compras reales que ocurrieron. Un recall bajo sugiere que el modelo está perdiendo una cantidad significativa de compras reales.
*   **F1-Score:** 0.2716
    *   Esta métrica, que es la media armónica de precisión y recall, muestra el equilibrio general del modelo en la identificación de la clase positiva. El valor de 0.2716 es bajo, reflejando el compromiso entre una precisión moderada y un recall bajo.

### Conclusión de la Evaluación:

El modelo LightGBM optimizado, al aplicar el umbral de 0.32, muestra una **Precisión** aceptable, lo que es bueno para no molestar a los clientes con recomendaciones irrelevantes. Sin embargo, su **Recall** es relativamente bajo. Esto significa que, aunque las predicciones positivas son razonablemente fiables, el modelo aún no está capturando una gran parte de las compras reales que ocurren. Para el negocio, esto implica que se identificarán oportunidades de venta con cierta confianza, pero se seguirán perdiendo muchas otras.

#### Concatenar batch_t1 a Datos Históricos

Integrar las compras efectivas de `batch_t1.parquet` con el DataFrame histórico consolidado (`df_merged`) para actualizar el conjunto de datos históricos.


````python
print("--- Integrando compras efectivas de batch_t1 en el historial consolidado ---")

# 1. Renombrar 'items_neto' a 'items' en df_batch_t1_effective_purchases
df_batch_t1_effective_purchases_renamed = df_batch_t1_effective_purchases.rename(columns={'items_neto': 'items'})
print("Columna 'items_neto' renombrada a 'items' en df_batch_t1_effective_purchases.")

# 2. Expandir df_batch_t1_effective_purchases_renamed con información de clientes y productos
# Se asume que df_clientes y df_productos están disponibles desde pasos anteriores.

df_batch_t1_expanded = pd.merge(
    df_batch_t1_effective_purchases_renamed,
    df_clientes,
    on='customer_id',
    how='left'
)
df_batch_t1_expanded = pd.merge(
    df_batch_t1_expanded,
    df_productos,
    on='product_id',
    how='left'
)
print("df_batch_t1 expandido con features de clientes y productos.")

# 3. Asegurarse de que las columnas y el orden coincidan con df_merged
# Obtener la lista de columnas de df_merged
merged_cols = df_merged.columns.tolist()

# Filtrar df_batch_t1_expanded para que tenga solo las columnas presentes en df_merged
# y reordenarlas para que coincidan
df_batch_t1_aligned = df_batch_t1_expanded[merged_cols].copy()
print("Columnas de df_batch_t1_expanded alineadas con df_merged.")

# 4. Concatenar df_merged y df_batch_t1_aligned para crear df_merged_updated
df_merged_updated = pd.concat([df_merged, df_batch_t1_aligned], ignore_index=True)
print(f"DataFrames concatenados. Dimensiones iniciales: {df_merged_updated.shape}")

# 5. Eliminar duplicados si los hubiera (en caso de solapamiento de fechas, aunque aquí no debería haber) y ordenar
df_merged_updated.drop_duplicates(inplace=True)
df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
df_merged_updated.reset_index(drop=True, inplace=True)
print(f"Duplicados eliminados y DataFrame ordenado. Dimensiones finales: {df_merged_updated.shape}")

# 6. Verificar las dimensiones y el rango de fechas de df_merged_updated
print(f"\nNuevas dimensiones de df_merged_updated: {df_merged_updated.shape}")
print(f"Rango de fechas de df_merged_updated: {df_merged_updated['purchase_date'].min().strftime('%Y-%m-%d')} a {df_merged_updated['purchase_date'].max().strftime('%Y-%m-%d')}")

print("--- Integración de batch_t1 completada ---")
````

#### Re-dividir Datos con Historial Actualizado

Redefinir las particiones de entrenamiento, validación y prueba (`df_train`, `df_val`, `df_test`) utilizando el `df_merged_updated` que incluye el historial de `batch_t1`.


````python
print("--- Redefiniendo particiones de datos con historial actualizado ---")

# 1. Obtener las semanas totales únicas de `df_merged_updated` y ordénelas.
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_total_updated = len(weeks_total_updated)

# 2. Mantener las definiciones de num_weeks_test y num_weeks_val como 4 semanas cada una.
num_weeks_test = 4
num_weeks_val = 4

# 3. Calcular los índices de inicio para los conjuntos de validación y prueba
idx_val_start_updated = num_weeks_total_updated - num_weeks_test - num_weeks_val
idx_test_start_updated = num_weeks_total_updated - num_weeks_test

# 4. Crear las listas de semanas para entrenamiento, validación y prueba
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:]

print(f"Semanas para Entrenamiento actualizadas ({len(weeks_train)}): {weeks_train[0]} a {weeks_train[-1]}")
print(f"Semanas para Validación actualizadas ({len(weeks_val)}): {weeks_val[0]} a {weeks_val[-1]}")
print(f"Semanas para Prueba actualizadas ({len(weeks_test)}): {weeks_test[0]} a {weeks_test[-1]}")

# 5. Generar los nuevos DataFrames df_train, df_val y df_test
df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()

# 6. Imprimir dimensiones y rangos de fechas para verificación
print(f"\nDimensiones df_train actualizado: {df_train.shape}")
print(f"Fechas en df_train actualizado: {df_train['purchase_date'].min().strftime('%Y-%m-%d')} a {df_train['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_val actualizado: {df_val.shape}")
print(f"Fechas en df_val actualizado: {df_val['purchase_date'].min().strftime('%Y-%m-%d')} a {df_val['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_test actualizado: {df_test.shape}")
print(f"Fechas en df_test actualizado: {df_test['purchase_date'].min().strftime('%Y-%m-%d')} a {df_test['purchase_date'].max().strftime('%Y-%m-%d')}")

print("--- Redefinición de particiones completada ---")
````

#### Re-generar Feature Engineering

Recalculate all features (lags, rolling windows, static, temporal) for the new train, validation, and test datasets, ensuring they reflect the updated historical data including `batch_t1`.


````python
print("--- Iniciando regeneración de Feature Engineering con historial actualizado ---")

# 1. Recrear el df_train_val DataFrame con los datos actualizados
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
print(f"df_train_val recreado. Dimensiones: {df_train_val.shape}")

# 2. Recrear df_train_hist_agg (historial para el conjunto de validación)
df_train_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index()
df_train_agg['purchased'] = 1 # Añadir indicador de compra
df_train_hist_agg = df_train_agg.copy()
print(f"df_train_hist_agg recreado. Dimensiones: {df_train_hist_agg.shape}")

# 3. Recrear df_train_val_hist_agg (historial para el conjunto de prueba)
df_train_val_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index()
df_train_val_agg['purchased'] = 1 # Añadir indicador de compra
df_train_val_hist_agg = df_train_val_agg.copy()
print(f"df_train_val_hist_agg recreado. Dimensiones: {df_train_val_hist_agg.shape}")

# 4. Recrear df_base_val y df_base_test
# Se asume que `create_target_base` está definido en el contexto
# `target_weeks_val` y `target_weeks_test` se deben haber actualizado previamente en el paso de Holdout
print("\n--- Recreando bases para validación y prueba ---")
df_base_val = create_target_base(df_val, weeks_val) # Usar df_val y las semanas actualizadas
df_base_test = create_target_base(df_test, weeks_test) # Usar df_test y las semanas actualizadas

print("\n--- Aplicando Features de Lag ---")
# 5. Aplicar add_lag_features a df_base_val
df_feat_val = add_lag_features(
    base_df=df_base_val,
    transactions_hist_df=df_train_hist_agg, # Historial para validar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

# 6. Aplicar add_lag_features a df_base_test
df_feat_test = add_lag_features(
    base_df=df_base_test,
    transactions_hist_df=df_train_val_hist_agg, # Historial para probar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

print("\n--- Aplicando Features de Ventana Móvil ---")
# 7. Aplicar add_rolling_features a df_feat_val
df_feat_val = add_rolling_features(
    base_df=df_feat_val,
    transactions_hist_agg_df=df_train_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

# 8. Aplicar add_rolling_features a df_feat_test
df_feat_test = add_rolling_features(
    base_df=df_feat_test,
    transactions_hist_agg_df=df_train_val_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

print("\n--- Añadiendo Features Estáticas y Temporales ---")
# 9. Mergear features de producto con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')

# 10. Mergear features de cliente con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

# 11. Aplicar add_temporal_features a df_feat_val y df_feat_test
df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

print("--- Verificación de Features Regeneradas ---")
print(f"\nDimensiones de df_feat_val: {df_feat_val.shape}")
print("Primeras 5 filas de df_feat_val:")
print(df_feat_val.head().to_string())
print("\nNaN check en df_feat_val:")
nan_check_val = df_feat_val.isnull().sum()
print(nan_check_val[nan_check_val > 0].to_string())

print(f"\nDimensiones de df_feat_test: {df_feat_test.shape}")
print("Primeras 5 filas de df_feat_test:")
print(df_feat_test.head().to_string())
print("\nNaN check en df_feat_test:")
nan_check_test = df_feat_test.isnull().sum()
print(nan_check_test[nan_check_test > 0].to_string())

# 12. Actualizar X_val, y_val, X_test, y_test
X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()

print("\n--- Feature Engineering regenerado y conjuntos X, y actualizados ---")
````

#### Reentrenar el Modelo LightGBM Optimizado

Reentrenar el `final_pipeline` (el modelo LightGBM con los hiperparámetros óptimos ya identificados) utilizando los nuevos datos de entrenamiento (`X_val` y `y_val` generados con el historial actualizado).


````python
print("--- Reentrenando el Modelo LightGBM Optimizado ---")

# Obtener los mejores parámetros de la optimización anterior
best_params = study.best_trial.params

# 1. Recrear el preprocesador con los mejores parámetros
num_imputer_strategy_final = best_params['num_imputer_strategy']
numeric_imputer_re_fit = SimpleImputer(strategy=num_imputer_strategy_final)

scaler_choice_re_fit = best_params['scaler']
if scaler_choice_re_fit == 'StandardScaler':
    scaler_re_fit = StandardScaler()
elif scaler_choice_re_fit == 'MinMaxScaler':
    scaler_re_fit = MinMaxScaler()
elif scaler_choice_re_fit == 'RobustScaler':
    scaler_re_fit = RobustScaler()
else:
    scaler_re_fit = 'passthrough'

numeric_transformer_re_fit = make_pipeline(numeric_imputer_re_fit, scaler_re_fit) if scaler_re_fit != 'passthrough' else make_pipeline(numeric_imputer_re_fit)
categorical_transformer_re_fit = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))

preprocessor_re_fit = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_re_fit, numeric_features),
        ('cat', categorical_transformer_re_fit, categorical_features)
    ],
    remainder='drop'
)
print("Preprocesador recreado con los mejores parámetros.")

# 2. Recrear el clasificador LightGBM con los mejores parámetros
lgbm_best_params_re_fit = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_re_fit = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params_re_fit)
print("Clasificador LightGBM recreado con los mejores parámetros.")

# 3. Crear el final_pipeline recreado
final_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_re_fit),
    ('classifier', lgbm_re_fit)
])
print("Pipeline final recreado.")

# 4. Entrenar el final_pipeline con los datos de entrenamiento actualizados (X_val y y_val)
print("Entrenando el pipeline final con X_val (historial actualizado) y y_val...")
final_pipeline.fit(X_val, y_val)
print("Reentrenamiento completado con éxito.")

# Opcional: Evaluar rápidamente el modelo reentrenado en X_val
from sklearn.metrics import classification_report
y_pred_val_re_fit = final_pipeline.predict(X_val)
print("\nClassification Report para X_val (después del reentrenamiento):")
print(classification_report(y_val, y_pred_val_re_fit, zero_division=0))
````

1.  **¿Cuál fue el resumen de los resultados de la evaluación de `batch_t1`?**
    El archivo `batch_t1.parquet` fue procesado para extraer las compras efectivas de cliente-producto para la semana `2025-W01`. Se identificaron 3264 pares únicos de cliente-producto que realizaron compras efectivas en ese período, estableciendo así el *ground truth* para la evaluación.

2.  **¿Cuál fue el resumen del proceso de reentrenamiento?**
    El modelo `LightGBM` fue reentrenado exitosamente utilizando los datos históricos actualizados que incluyen el `batch_t1`. Previo al reentrenamiento, los datos fueron redivididos en conjuntos de entrenamiento, validación y prueba, y se regeneraron todas las características (lags, ventanas móviles, estáticas, temporales) para asegurar que el modelo se entrenara con la información más reciente y relevante.

3.  **¿Cuáles fueron las nuevas predicciones generadas para `2025-W02`?**
    Se generaron predicciones binarias y de probabilidad para la semana `2025-W02` utilizando el modelo `LightGBM` reentrenado. Estas predicciones fueron guardadas en el archivo `purchase_predictions_2025_W02.csv`.

### Data Analysis Key Findings

*   **Preparación del Ground Truth de `batch_t1`**: Se identificaron 3264 pares únicos de cliente-producto como compras efectivas para la semana `2025-W01` (del 01/01/25 al 05/01/25), que sirvieron como el *ground truth* para la evaluación.
*   **Evaluación de Predicciones de `2025-W01`**:
    *   **Verdaderos Positivos (TP)**: 612
    *   **Falsos Positivos (FP)**: 631
    *   **Falsos Negativos (FN)**: 2652
    *   **Precisión**: 0.4924, indicando que aproximadamente el 49.24% de las predicciones positivas del modelo fueron correctas.
    *   **Recall**: 0.1875, lo que significa que el modelo logró identificar solo el 18.75% de todas las compras reales.
    *   **F1-Score**: 0.2716, reflejando un equilibrio bajo entre precisión y recall.
*   **Integración de Datos Históricos**: El `df_merged_updated` resultante de la integración de `batch_t1` contiene 246978 filas y 20 columnas, abarcando un rango de fechas desde 2024-01-01 hasta 2025-01-05.
*   **Redivisión de Datos**: Los conjuntos de entrenamiento, validación y prueba fueron actualizados:
    *   `df_train`: 45 semanas (2024-W01 a 2024-W45), con 195571 filas.
    *   `df_val`: 4 semanas (2024-W46 a 2024-W49), con 23800 filas.
    *   `df_test`: 4 semanas (2024-W50 a 2025-W01), con 27607 filas.
*   **Regeneración de Features**: Todas las características para los conjuntos de validación y prueba (`df_feat_val` con 378560 filas y `df_feat_test` con 403800 filas, ambas con 26 características) fueron exitosamente recalculadas y no presentaron valores nulos.
*   **Reentrenamiento y Predicción para `2025-W02`**: El modelo `LightGBM` fue reentrenado exitosamente con los datos actualizados y se generaron predicciones binarias y de probabilidad para la semana `2025-W02`, las cuales fueron guardadas en `purchase_predictions_2025_W02.csv`.

### Insights or Next Steps

*   El modelo actual, con un Recall del 18.75%, es deficiente para identificar la mayoría de las compras reales, lo que podría llevar a perder un número significativo de oportunidades de venta. Se recomienda investigar métodos para mejorar la capacidad del modelo para detectar más positivos reales, como el ajuste de umbrales, el re-balanceo de clases o la exploración de características adicionales que capturen mejor el comportamiento de compra.
*   Dada la baja F1-Score (0.2716), se debería considerar una optimización más profunda del modelo, posiblemente explorando diferentes arquitecturas de modelos, técnicas de ensamblaje o un *feature engineering* más sofisticado, para mejorar el equilibrio entre precisión y recall y aumentar la utilidad general del modelo para el negocio.

````python
print("--- Integrando compras efectivas de batch_t3 en el historial consolidado ---")

# Ensure df_clientes and df_productos are available (re-load if necessary)
# Assuming df_clientes and df_productos are already loaded in the environment from previous cells

# Initialize df_merged_updated if it's not defined (this handles the first batch integration)
if 'df_merged_updated' not in globals():
    df_merged_updated = df_merged.copy()
    print("df_merged_updated inicializado con df_merged (asumiendo primera integración de batch). Dims:", df_merged_updated.shape)

# --- Steps to integrate batch_t1 (re-executed for kernel consistency) ---
try:
    df_batch_t1_raw = pd.read_parquet('batch_t1.parquet')
    df_batch_t1_raw['purchase_date'] = pd.to_datetime(df_batch_t1_raw['purchase_date']).dt.date
    df_batch_t1_agg = df_batch_t1_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t1_effective_purchases = df_batch_t1_agg[df_batch_t1_agg['items_neto'] > 0].copy()
    df_batch_t1_effective_purchases_renamed = df_batch_t1_effective_purchases.rename(columns={'items_neto': 'items'})

    df_batch_t1_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t1_effective_purchases_renamed['purchase_date'])
    df_batch_t1_effective_purchases_renamed['year'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t1_effective_purchases_renamed['week'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t1_expanded = pd.merge(df_batch_t1_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t1_expanded = pd.merge(df_batch_t1_expanded, df_productos, on='product_id', how='left')
    df_batch_t1_expanded['year_week'] = df_batch_t1_expanded['year'].astype(str) + '-W' + df_batch_t1_expanded['week'].astype(str).str.zfill(2)

    columns_to_align_t1 = df_merged_updated.columns.tolist()
    missing_in_batch_t1_expanded = set(columns_to_align_t1) - set(df_batch_t1_expanded.columns)
    for col in missing_in_batch_t1_expanded:
        df_batch_t1_expanded[col] = np.nan
    df_batch_t1_aligned = df_batch_t1_expanded[columns_to_align_t1].copy()

    df_merged_updated = pd.concat([df_merged_updated, df_batch_t1_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t1 integrated into df_merged_updated. Dims:", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t1 (re-execution): {e}")

# --- Steps to integrate batch_t2 (re-executed for kernel consistency) ---
try:
    df_batch_t2_raw = pd.read_parquet('batch_t2.parquet')
    df_batch_t2_raw['purchase_date'] = pd.to_datetime(df_batch_t2_raw['purchase_date']).dt.date
    df_batch_t2_agg = df_batch_t2_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t2_effective_purchases = df_batch_t2_agg[df_batch_t2_agg['items_neto'] > 0].copy()
    df_batch_t2_effective_purchases_renamed = df_batch_t2_effective_purchases.rename(columns={'items_neto': 'items'})

    df_batch_t2_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t2_effective_purchases_renamed['purchase_date'])
    df_batch_t2_effective_purchases_renamed['year'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t2_effective_purchases_renamed['week'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t2_expanded = pd.merge(df_batch_t2_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t2_expanded = pd.merge(df_batch_t2_expanded, df_productos, on='product_id', how='left')
    df_batch_t2_expanded['year_week'] = df_batch_t2_expanded['year'].astype(str) + '-W' + df_batch_t2_expanded['week'].astype(str).str.zfill(2)

    columns_to_align_t2 = df_merged_updated.columns.tolist()
    missing_in_batch_t2_expanded = set(columns_to_align_t2) - set(df_batch_t2_expanded.columns)
    for col in missing_in_batch_t2_expanded:
        df_batch_t2_expanded[col] = np.nan
    df_batch_t2_aligned = df_batch_t2_expanded[columns_to_align_t2].copy()

    df_merged_updated = pd.concat([df_merged_updated, df_batch_t2_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t2 integrated into df_merged_updated. Dims:", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t2 (re-execution): {e}")


# --- Current subtask: Integrate batch_t3 ---
# 1. Renombrar 'items_neto' a 'items' en df_batch_t3_effective_purchases
df_batch_t3_effective_purchases_renamed = df_batch_t3_effective_purchases.rename(columns={'items_neto': 'items'})
print("Columna 'items_neto' renombrada a 'items' en df_batch_t3_effective_purchases.")

# 2. Realizar un merge del DataFrame df_batch_t3_effective_purchases_renamed con df_clientes y df_productos
df_batch_t3_expanded = pd.merge(
    df_batch_t3_effective_purchases_renamed,
    df_clientes,
    on='customer_id',
    how='left'
)
df_batch_t3_expanded = pd.merge(
    df_batch_t3_expanded,
    df_productos,
    on='product_id',
    how='left'
)
print("df_batch_t3 expandido con features de clientes y productos.")

# 3. Crear la columna 'year_week' en df_batch_t3_expanded
df_batch_t3_expanded['year_week'] = df_batch_t3_expanded['year'].astype(str) + '-W' + df_batch_t3_expanded['week'].astype(str).str.zfill(2)

# 4. Filtrar las columnas de df_batch_t3_expanded para que coincidan con las columnas de df_merged_updated y reordenarlas
merged_updated_cols = df_merged_updated.columns.tolist()

missing_in_batch_t3 = set(merged_updated_cols) - set(df_batch_t3_expanded.columns)
for col in missing_in_batch_t3:
    df_batch_t3_expanded[col] = np.nan
df_batch_t3_aligned = df_batch_t3_expanded[merged_updated_cols].copy()
print("Columnas de df_batch_t3_expanded alineadas con df_merged_updated.")

# 5. Concatenar df_merged_updated y df_batch_t3_aligned para crear df_merged_updated_final
df_merged_updated_final = pd.concat([df_merged_updated, df_batch_t3_aligned], ignore_index=True)
print(f"DataFrames concatenados. Dimensiones iniciales: {df_merged_updated_final.shape}")

# 6. Eliminar duplicados si los hubiera y ordenar
df_merged_updated_final.drop_duplicates(inplace=True)
df_merged_updated_final.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
df_merged_updated_final.reset_index(drop=True, inplace=True)
print(f"Duplicados eliminados y DataFrame ordenado. Dimensiones finales: {df_merged_updated_final.shape}")

# 7. Actualizar el DataFrame df_merged_updated con el contenido de df_merged_updated_final
df_merged_updated = df_merged_updated_final.copy()

# 8. Imprimir las nuevas dimensiones y el rango de fechas de df_merged_updated para verificar
print(f"\nNuevas dimensiones de df_merged_updated: {df_merged_updated.shape}")
print(f"Rango de fechas de df_merged_updated: {df_merged_updated['purchase_date'].min().strftime('%Y-%m-%d')} a {df_merged_updated['purchase_date'].max().strftime('%Y-%m-%d')}")

print("--- Integración de batch_t3 completada ---")
````

````python
import pandas as pd
from itertools import product
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from lightgbm import LGBMClassifier
import numpy as np
import warnings
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import optuna
import joblib

print("--- Initial Data Loading and Environment Setup ---")

# --- 1. Load Initial Datasets ---

try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    df_clientes = pd.read_parquet('clientes.parquet')
    df_productos = pd.read_parquet('productos.parquet')
    print("Datasets cargados exitosamente.")
except FileNotFoundError:
    print("Error: Uno o más archivos Parquet no se encontraron. Asegúrese de que estén en el directorio correcto.")
    exit() # Exit if crucial files are not found

# --- 2. Preprocess df_transacciones and Create df_merged ---
df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date

# Group transactions by customer, product, and day to sum 'items_neto'
df_trans_agg = df_transacciones.groupby(
    ['customer_id', 'product_id', 'purchase_date']
).agg(
    items_neto=('items', 'sum')
).reset_index()

# Filter for effective purchases (items_neto > 0)
df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)

df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])
df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week

df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')
df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')
df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)
print(f"Initial df_merged created. Dimensions: {df_merged.shape}")

# --- 3. Define Helper Functions for Feature Engineering ---
def year_week_to_date(yw_str):
    try:
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        return pd.NaT

def get_lag_week(target_week_str, lag):
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    if not target_weeks:
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])
    df_final = pd.merge(df_base, target_transactions, left_on=['customer_id', 'product_id', 'target_week'], right_on=['customer_id', 'product_id', 'year_week'], how='left')
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])
    df_final['target'] = df_final['target'].fillna(0).astype(int)
    return df_final

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    df_with_lags = base_df.copy()
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()
    for lag in lag_weeks:
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())
        df_with_lags = pd.merge(df_with_lags, hist_lag[merge_cols], on=['customer_id', 'product_id', 'lookup_week'], how='left')
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])
    return df_with_lags

def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])
    for window in windows:
        hist_data_indexed = hist_data.set_index('date')
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)
        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                rolling_col = rolling_col.reset_index()
                rolling_col = rolling_col.rename(columns={feature_col: col_name})
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(), on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})
                df_with_rolling = pd.merge(df_with_rolling, rolling_col[['customer_id', 'product_id', 'target_week', col_name]], on=['customer_id', 'product_id', 'target_week'], how='left')
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)
    return df_with_rolling

def add_temporal_features(df):
    df_temp = df.copy()
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    df_temp['month'] = df_temp['date'].dt.month.astype(int)
    df_temp = df_temp.drop(columns=['date'])
    return df_temp
print("Helper functions defined.")

# --- 4. Define Global Feature Lists and Selectors ---
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]
feature_columns = numeric_features + categorical_features
print("Global feature lists and selectors set up.")
````

````python
print("--- Integrando compras efectivas de batch_t3 en el historial consolidado ---")

# Ensure df_clientes and df_productos are available (re-load if necessary)
# Assuming df_clientes and df_productos are already loaded in the environment from previous cells

# Initialize df_merged_updated if it's not defined (this handles the first batch integration)
if 'df_merged_updated' not in globals():
    df_merged_updated = df_merged.copy()
    print("df_merged_updated inicializado con df_merged (asumiendo primera integración de batch). Dims:", df_merged_updated.shape)

# --- Steps to integrate batch_t1 (re-executed for kernel consistency) ---
try:
    df_batch_t1_raw = pd.read_parquet('batch_t1.parquet')
    df_batch_t1_raw['purchase_date'] = pd.to_datetime(df_batch_t1_raw['purchase_date']).dt.date
    df_batch_t1_agg = df_batch_t1_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t1_effective_purchases = df_batch_t1_agg[df_batch_t1_agg['items_neto'] > 0].copy()
    df_batch_t1_effective_purchases_renamed = df_batch_t1_effective_purchases.rename(columns={'items_neto': 'items'})

    df_batch_t1_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t1_effective_purchases_renamed['purchase_date'])
    df_batch_t1_effective_purchases_renamed['year'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t1_effective_purchases_renamed['week'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t1_expanded = pd.merge(df_batch_t1_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t1_expanded = pd.merge(df_batch_t1_expanded, df_productos, on='product_id', how='left')
    df_batch_t1_expanded['year_week'] = df_batch_t1_expanded['year'].astype(str) + '-W' + df_batch_t1_expanded['week'].astype(str).str.zfill(2)

    columns_to_align_t1 = df_merged_updated.columns.tolist()
    missing_in_batch_t1_expanded = set(columns_to_align_t1) - set(df_batch_t1_expanded.columns)
    for col in missing_in_batch_t1_expanded:
        df_batch_t1_expanded[col] = np.nan
    df_batch_t1_aligned = df_batch_t1_expanded[columns_to_align_t1].copy()

    df_merged_updated = pd.concat([df_merged_updated, df_batch_t1_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t1 integrated into df_merged_updated. Dims:", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t1 (re-execution): {e}")

# --- Steps to integrate batch_t2 (re-executed for kernel consistency) ---
try:
    df_batch_t2_raw = pd.read_parquet('batch_t2.parquet')
    df_batch_t2_raw['purchase_date'] = pd.to_datetime(df_batch_t2_raw['purchase_date']).dt.date
    df_batch_t2_agg = df_batch_t2_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t2_effective_purchases = df_batch_t2_agg[df_batch_t2_agg['items_neto'] > 0].copy()
    df_batch_t2_effective_purchases_renamed = df_batch_t2_effective_purchases.rename(columns={'items_neto': 'items'})

    df_batch_t2_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t2_effective_purchases_renamed['purchase_date'])
    df_batch_t2_effective_purchases_renamed['year'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t2_effective_purchases_renamed['week'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t2_expanded = pd.merge(df_batch_t2_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t2_expanded = pd.merge(df_batch_t2_expanded, df_productos, on='product_id', how='left')
    df_batch_t2_expanded['year_week'] = df_batch_t2_expanded['year'].astype(str) + '-W' + df_batch_t2_expanded['week'].astype(str).str.zfill(2)

    columns_to_align_t2 = df_merged_updated.columns.tolist()
    missing_in_batch_t2_expanded = set(columns_to_align_t2) - set(df_batch_t2_expanded.columns)
    for col in missing_in_batch_t2_expanded:
        df_batch_t2_expanded[col] = np.nan
    df_batch_t2_aligned = df_batch_t2_expanded[columns_to_align_t2].copy()

    df_merged_updated = pd.concat([df_merged_updated, df_batch_t2_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t2 integrated into df_merged_updated. Dims:", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t2 (re-execution): {e}")


# --- Current subtask: Integrate batch_t3 ---
# 1. Renombrar 'items_neto' a 'items' en df_batch_t3_effective_purchases
df_batch_t3_effective_purchases_renamed = df_batch_t3_effective_purchases.rename(columns={'items_neto': 'items'})
print("Columna 'items_neto' renombrada a 'items' en df_batch_t3_effective_purchases.")

# 2. Realizar un merge del DataFrame df_batch_t3_effective_purchases_renamed con df_clientes y df_productos
df_batch_t3_expanded = pd.merge(
    df_batch_t3_effective_purchases_renamed,
    df_clientes,
    on='customer_id',
    how='left'
)
df_batch_t3_expanded = pd.merge(
    df_batch_t3_expanded,
    df_productos,
    on='product_id',
    how='left'
)
print("df_batch_t3 expandido con features de clientes y productos.")

# 3. Crear la columna 'year_week' en df_batch_t3_expanded
df_batch_t3_expanded['year_week'] = df_batch_t3_expanded['year'].astype(str) + '-W' + df_batch_t3_expanded['week'].astype(str).str.zfill(2)

# 4. Filtrar las columnas de df_batch_t3_expanded para que coincidan con las columnas de df_merged_updated y reordenarlas
merged_updated_cols = df_merged_updated.columns.tolist()

missing_in_batch_t3 = set(merged_updated_cols) - set(df_batch_t3_expanded.columns)
for col in missing_in_batch_t3:
    df_batch_t3_expanded[col] = np.nan
df_batch_t3_aligned = df_batch_t3_expanded[merged_updated_cols].copy()
print("Columnas de df_batch_t3_expanded alineadas con df_merged_updated.")

# 5. Concatenar df_merged_updated y df_batch_t3_aligned para crear df_merged_updated_final
df_merged_updated_final = pd.concat([df_merged_updated, df_batch_t3_aligned], ignore_index=True)
print(f"DataFrames concatenados. Dimensiones iniciales: {df_merged_updated_final.shape}")

# 6. Eliminar duplicados si los hubiera y ordenar
df_merged_updated_final.drop_duplicates(inplace=True)
df_merged_updated_final.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
df_merged_updated_final.reset_index(drop=True, inplace=True)
print(f"Duplicados eliminados y DataFrame ordenado. Dimensiones finales: {df_merged_updated_final.shape}")

# 7. Actualizar el DataFrame df_merged_updated con el contenido de df_merged_updated_final
df_merged_updated = df_merged_updated_final.copy()

# 8. Imprimir las nuevas dimensiones y el rango de fechas de df_merged_updated para verificar
print(f"\nNuevas dimensiones de df_merged_updated: {df_merged_updated.shape}")
print(f"Rango de fechas de df_merged_updated: {df_merged_updated['purchase_date'].min().strftime('%Y-%m-%d')} a {df_merged_updated['purchase_date'].max().strftime('%Y-%m-%d')}")

print("--- Integración de batch_t3 completada ---")
````

````python
import pandas as pd
from itertools import product
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from lightgbm import LGBMClassifier
import numpy as np
import warnings
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import optuna
import joblib

print("--- Initial Data Loading and Environment Setup ---")

# --- 1. Load Initial Datasets ---

try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    df_clientes = pd.read_parquet('clientes.parquet')
    df_productos = pd.read_parquet('productos.parquet')
    print("Datasets cargados exitosamente.")
except FileNotFoundError:
    print("Error: Uno o más archivos Parquet no se encontraron. Asegúrese de que estén en el directorio correcto.")
    exit() # Exit if crucial files are not found

# --- 2. Preprocess df_transacciones and Create df_merged ---
df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date

# Group transactions by customer, product, and day to sum 'items_neto'
df_trans_agg = df_transacciones.groupby(
    ['customer_id', 'product_id', 'purchase_date']
).agg(
    items_neto=('items', 'sum')
).reset_index()

# Filter for effective purchases (items_neto > 0)
df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)

df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])
df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week

df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')
df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')
df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)
print(f"Initial df_merged created. Dimensions: {df_merged.shape}")

# --- 3. Define Helper Functions for Feature Engineering ---
def year_week_to_date(yw_str):
    try:
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        return pd.NaT

def get_lag_week(target_week_str, lag):
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    if not target_weeks:
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])
    df_final = pd.merge(df_base, target_transactions, left_on=['customer_id', 'product_id', 'target_week'], right_on=['customer_id', 'product_id', 'year_week'], how='left')
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])
    df_final['target'] = df_final['target'].fillna(0).astype(int)
    return df_final

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    df_with_lags = base_df.copy()
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()
    for lag in lag_weeks:
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())
        df_with_lags = pd.merge(df_with_lags, hist_lag[merge_cols], on=['customer_id', 'product_id', 'lookup_week'], how='left')
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])
    return df_with_lags

def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])
    for window in windows:
        hist_data_indexed = hist_data.set_index('date')
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)
        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                rolling_col = rolling_col.reset_index()
                rolling_col = rolling_col.rename(columns={feature_col: col_name})
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(), on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})
                df_with_rolling = pd.merge(df_with_rolling, rolling_col[['customer_id', 'product_id', 'target_week', col_name]], on=['customer_id', 'product_id', 'target_week'], how='left')
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)
    return df_with_rolling

def add_temporal_features(df):
    df_temp = df.copy()
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    df_temp['month'] = df_temp['date'].dt.month.astype(int)
    df_temp = df_temp.drop(columns=['date'])
    return df_temp
print("Helper functions defined.")

# --- 4. Define Global Feature Lists and Selectors ---
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]
feature_columns = numeric_features + categorical_features
print("Global feature lists and selectors set up.")
````

````python
import pandas as pd
from itertools import product
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from lightgbm import LGBMClassifier
import numpy as np
import warnings
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import optuna
import joblib

print("--- Re-ejecutando pasos previos para asegurar el estado del kernel y generar predicciones para 2025-W03 ---")
print("ATENCIÓN: Asegúrese de que 'transacciones.parquet', 'clientes.parquet', 'productos.parquet', 'batch_t1.parquet', 'batch_t2.parquet', 'batch_t3.parquet' estén en el directorio de trabajo actual.")

# --- 1. Initial Data Loading & Preprocessing (up to 2024-W52) ---
try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    df_clientes = pd.read_parquet('clientes.parquet')
    df_productos = pd.read_parquet('productos.parquet')

    df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date
    df_trans_agg = df_transacciones.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
    df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)
    df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])
    df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
    df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week
    df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')
    df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')
    df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)
    print("Initial df_merged (up to 2024-W52) recreated. Dims: ", df_merged.shape)
except Exception as e:
    print(f"Error recreating initial df_merged: {e}")

# --- 2. Helper Functions for Feature Engineering ---
def year_week_to_date(yw_str):
    try:
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        return pd.NaT

def get_lag_week(target_week_str, lag):
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    if not target_weeks:
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])
    df_final = pd.merge(df_base, target_transactions, left_on=['customer_id', 'product_id', 'target_week'], right_on=['customer_id', 'product_id', 'year_week'], how='left')
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])
    df_final['target'] = df_final['target'].fillna(0).astype(int)
    return df_final

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    df_with_lags = base_df.copy()
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()
    for lag in lag_weeks:
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())
        df_with_lags = pd.merge(df_with_lags, hist_lag[merge_cols], on=['customer_id', 'product_id', 'lookup_week'], how='left')
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])
    return df_with_lags

def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])
    for window in windows:
        hist_data_indexed = hist_data.set_index('date')
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)
        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                rolling_col = rolling_col.reset_index()
                rolling_col = rolling_col.rename(columns={feature_col: col_name})
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(), on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})
                df_with_rolling = pd.merge(df_with_rolling, rolling_col[['customer_id', 'product_id', 'target_week', col_name]], on=['customer_id', 'product_id', 'target_week'], how='left')
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)
    return df_with_rolling

def add_temporal_features(df):
    df_temp = df.copy()
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    df_temp['month'] = df_temp['date'].dt.month.astype(int)
    df_temp = df_temp.drop(columns=['date'])
    return df_temp
print("Helper functions re-defined.")

# --- 3. Define Common Feature Engineering Variables ---
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]
feature_columns = numeric_features + categorical_features
print("Feature engineering variables and lists re-defined.")

# --- 4. Process and Integrate batch_t1 (up to 2025-W01) ---
try:
    df_batch_t1_raw = pd.read_parquet('batch_t1.parquet')
    df_batch_t1_raw['purchase_date'] = pd.to_datetime(df_batch_t1_raw['purchase_date']).dt.date
    df_batch_t1_agg = df_batch_t1_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t1_effective_purchases = df_batch_t1_agg[df_batch_t1_agg['items_neto'] > 0].copy()
    df_batch_t1_effective_purchases_renamed = df_batch_t1_effective_purchases.rename(columns={'items_neto': 'items'})

    # Fix: Ensure purchase_date is datetime and extract year/week for batch_t1
    df_batch_t1_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t1_effective_purchases_renamed['purchase_date'])
    df_batch_t1_effective_purchases_renamed['year'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t1_effective_purchases_renamed['week'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t1_expanded = pd.merge(df_batch_t1_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t1_expanded = pd.merge(df_batch_t1_expanded, df_productos, on='product_id', how='left')
    df_batch_t1_expanded['year_week'] = df_batch_t1_expanded['year'].astype(str) + '-W' + df_batch_t1_expanded['week'].astype(str).str.zfill(2)

    # Initialize df_merged_updated with df_merged plus batch_t1. For subsequent batches, this will be updated.
    columns_to_align = df_merged.columns.tolist()
    missing_in_batch_t1 = set(columns_to_align) - set(df_batch_t1_expanded.columns)
    for col in missing_in_batch_t1:
        df_batch_t1_expanded[col] = np.nan # Use np.nan for missing columns to maintain dtype compatibility
    df_batch_t1_aligned = df_batch_t1_expanded[columns_to_align].copy()

    df_merged_updated = pd.concat([df_merged, df_batch_t1_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t1 integrated into df_merged_updated. Dims: ", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t1: {e}")

# --- 5. Redefine Data Splits after batch_t1 ---
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_test = 4
num_weeks_val = 4
idx_val_start_updated = len(weeks_total_updated) - num_weeks_test - num_weeks_val
idx_test_start_updated = len(weeks_total_updated) - num_weeks_test
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:] # Contains 2025-W01 for evaluation

df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()
print("Data splits re-defined after batch_t1.")

# --- 6. Regenerate Feature Engineering after batch_t1 ---
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
df_train_hist_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)

df_base_val = create_target_base(df_val, weeks_val)
df_base_test = create_target_base(df_test, weeks_test)

df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

X_val = df_feat_val[feature_columns].copy() # Global X_val
y_val = df_feat_val['target'].copy() # Global y_val
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()
print("Feature engineering re-generated after batch_t1.")

# --- 7. Re-run Optuna Optimization and Model Training (first run) ---
warnings.filterwarnings('ignore', category=optuna.exceptions.ExperimentalWarning)
warnings.filterwarnings('ignore', category=UserWarning)

def objective(trial):
    num_imputer_strategy = trial.suggest_categorical('num_imputer_strategy', ['mean', 'median'])
    numeric_imputer = SimpleImputer(strategy=num_imputer_strategy)
    scaler_choice = trial.suggest_categorical('scaler', ['StandardScaler', 'MinMaxScaler', 'RobustScaler', 'None'])
    if scaler_choice == 'StandardScaler': scaler = StandardScaler()
    elif scaler_choice == 'MinMaxScaler': scaler = MinMaxScaler()
    elif scaler_choice == 'RobustScaler': scaler = RobustScaler()
    else: scaler = 'passthrough'
    numeric_transformer = make_pipeline(numeric_imputer, scaler) if scaler != 'passthrough' else make_pipeline(numeric_imputer)
    categorical_transformer = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    preprocessor = ColumnTransformer(transformers=[('num', numeric_transformer, numeric_features), ('cat', categorical_transformer, categorical_features)], remainder='drop')
    lgbm_params = {
        'objective': 'binary', 'metric': 'f1', 'random_state': 42, 'n_jobs': -1,
        'n_estimators': trial.suggest_int('n_estimators', 200, 1500, step=100),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 20, 100),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-3, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'is_unbalance': trial.suggest_categorical('is_unbalance', [True, False])
    }
    model = LGBMClassifier(**lgbm_params)
    pipeline = Pipeline(steps=[('preprocessor', preprocessor), ('classifier', model)])
    cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    try: scores = cross_val_score(pipeline, X_val, y_val, scoring='f1', cv=cv, n_jobs=-1); f1_mean = np.mean(scores); return f1_mean
    except Exception as e: return 0.0

study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=15, timeout=1800)
best_params = study.best_trial.params

num_imputer_final = SimpleImputer(strategy=best_params['num_imputer_strategy'])
scaler_choice_final = best_params['scaler']
if scaler_choice_final == 'StandardScaler': scaler_final = StandardScaler()
elif scaler_choice_final == 'MinMaxScaler': scaler_final = MinMaxScaler()
elif scaler_choice_final == 'RobustScaler': scaler_final = RobustScaler()
else: scaler_final = 'passthrough'
numeric_transformer_final = make_pipeline(num_imputer_final, scaler_final) if scaler_final != 'passthrough' else make_pipeline(num_imputer_final)
categorical_transformer_final = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
preprocessor_final = ColumnTransformer(transformers=[('num', numeric_transformer_final, numeric_features), ('cat', categorical_transformer_final, categorical_features)], remainder='drop')
lgbm_best_params = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_final = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params)
final_pipeline = Pipeline(steps=[('preprocessor', preprocessor_final), ('classifier', lgbm_final)])
final_pipeline.fit(X_val, y_val)

y_pred_proba_val = final_pipeline.predict_proba(X_val)[:, 1]
optimal_threshold = 0
best_f1_score = 0
for threshold in np.arange(0.0, 1.01, 0.01):
    y_pred_binary = (y_pred_proba_val >= threshold).astype(int)
    f1 = f1_score(y_val, y_pred_binary, pos_label=1, zero_division=0)
    if f1 > best_f1_score:
        best_f1_score = f1
        optimal_threshold = threshold
print(f"Model re-trained (after batch_t1) and optimal_threshold found: {optimal_threshold:.2f}")

# --- 6. Process and Integrate batch_t2 (up to 2025-W02) ---
try:
    df_batch_t2_raw = pd.read_parquet('batch_t2.parquet')
    df_batch_t2_raw['purchase_date'] = pd.to_datetime(df_batch_t2_raw['purchase_date']).dt.date
    df_batch_t2_agg = df_batch_t2_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t2_effective_purchases = df_batch_t2_agg[df_batch_t2_agg['items_neto'] > 0].copy()
    df_batch_t2_effective_purchases_renamed = df_batch_t2_effective_purchases.rename(columns={'items_neto': 'items'})

    # Fix: Ensure purchase_date is datetime and extract year/week for batch_t2
    df_batch_t2_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t2_effective_purchases_renamed['purchase_date'])
    df_batch_t2_effective_purchases_renamed['year'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t2_effective_purchases_renamed['week'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t2_expanded = pd.merge(df_batch_t2_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t2_expanded = pd.merge(df_batch_t2_expanded, df_productos, on='product_id', how='left')
    df_batch_t2_expanded['year_week'] = df_batch_t2_expanded['year'].astype(str) + '-W' + df_batch_t2_expanded['week'].astype(str).str.zfill(2)

    # Align columns before concatenation
    columns_to_align = df_merged_updated.columns.tolist()
    missing_in_batch_t2 = set(columns_to_align) - set(df_batch_t2_expanded.columns)
    for col in missing_in_batch_t2:
        df_batch_t2_expanded[col] = np.nan # Use np.nan for missing columns to maintain dtype compatibility
    df_batch_t2_aligned = df_batch_t2_expanded[columns_to_align].copy()

    df_merged_updated_final = pd.concat([df_merged_updated, df_batch_t2_aligned], ignore_index=True)
    df_merged_updated_final.drop_duplicates(inplace=True)
    df_merged_updated_final.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated_final.reset_index(drop=True, inplace=True)
    df_merged_updated = df_merged_updated_final.copy()
    print("batch_t2 integrated into df_merged_updated. Dims: ", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t2: {e}")

# --- 7. Redefine Data Splits after batch_t2 ---
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_test = 4
num_weeks_val = 4
idx_val_start_updated = len(weeks_total_updated) - num_weeks_test - num_weeks_val
idx_test_start_updated = len(weeks_total_updated) - num_weeks_test
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:] # Contains 2025-W02 for evaluation

df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()
print("Data splits re-defined after batch_t2.")

# --- 8. Regenerate Feature Engineering after batch_t2 ---
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
df_train_hist_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)

df_base_val = create_target_base(df_val, weeks_val)
df_base_test = create_target_base(df_test, weeks_test)

df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()
print("Feature engineering re-generated after batch_t2.")

# --- 9. Re-train Model after batch_t2 ---
lgbm_best_params_re_fit = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_re_fit = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params_re_fit)

num_imputer_re_fit = SimpleImputer(strategy=best_params['num_imputer_strategy'])
scaler_choice_re_fit = best_params['scaler']
if scaler_choice_re_fit == 'StandardScaler': scaler_re_fit = StandardScaler()
elif scaler_choice_re_fit == 'MinMaxScaler': scaler_re_fit = MinMaxScaler()
elif scaler_choice_re_fit == 'RobustScaler': scaler_re_fit = RobustScaler()
else: scaler_re_fit = 'passthrough'
numeric_transformer_re_fit = make_pipeline(num_imputer_re_fit, scaler_re_fit) if scaler_re_fit != 'passthrough' else make_pipeline(num_imputer_re_fit)
categorical_transformer_re_fit = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
preprocessor_final = ColumnTransformer(transformers=[('num', numeric_transformer_re_fit, numeric_features), ('cat', categorical_transformer_re_fit, categorical_features)], remainder='drop')

final_pipeline = Pipeline(steps=[('preprocessor', preprocessor_final), ('classifier', lgbm_re_fit)])

final_pipeline.fit(X_val, y_val)
print("Model re-trained with data including batch_t2.")

# Re-calculate optimal_threshold after retraining with batch_t2 data
y_pred_proba_val_current = final_pipeline.predict_proba(X_val)[:, 1]
optimal_threshold = 0
best_f1_score_current = 0
for threshold in np.arange(0.0, 1.01, 0.01):
    y_pred_binary_current = (y_pred_proba_val_current >= threshold).astype(int)
    f1_current = f1_score(y_val, y_pred_binary_current, pos_label=1, zero_division=0)
    if f1_current > best_f1_score_current:
        best_f1_score_current = f1_current
        optimal_threshold = threshold
print(f"Optimal threshold recalculated after re-training: {optimal_threshold:.2f}")

# --- 10. Prepare and Generate Predictions for 2025-W03 ---
new_target_week = '2025-W03'

current_historical_data = df_merged_updated.copy()
current_historical_data['purchased'] = 1 # Add 'purchased' column for feature generation

active_customers_for_new_week = df_clientes['customer_id'].unique()
active_products_for_new_week = df_productos['product_id'].unique()
df_new_predictions_base = pd.DataFrame(list(product(active_customers_for_new_week, active_products_for_new_week, [new_target_week])), columns=['customer_id', 'product_id', 'target_week'])
print(f"Tabla base para {new_target_week} creada. Dimensions: {df_new_predictions_base.shape}")

df_new_features = add_lag_features(
    base_df=df_new_predictions_base,
    transactions_hist_df=current_historical_data,
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

df_new_features = add_rolling_features(
    base_df=df_new_features,
    transactions_hist_agg_df=current_historical_data.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1),
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)
print(f"Features de Lag y Rolling añadidas. Dimensions: {df_new_features.shape}")

df_new_features = pd.merge(df_new_features, productos_sel, on='product_id', how='left')
df_new_features = pd.merge(df_new_features, clientes_sel, on='customer_id', how='left')

df_new_features = add_temporal_features(df_new_features)
print(f"Features estáticas y temporales añadidas. Dimensions: {df_new_features.shape}")

X_new_predict = df_new_features[feature_columns].copy()
print(f"X_new_predict listo para la inferencia. Dimensions: {X_new_predict.shape}")

new_probabilities = final_pipeline.predict_proba(X_new_predict)
predicted_new_purchase_proba = new_probabilities[:, 1]

new_predictions_df = df_new_features[['customer_id', 'product_id', 'target_week']].copy()
new_predictions_df['predicted_proba'] = predicted_new_purchase_proba
new_predictions_df['predicted_purchase_binary'] = (new_predictions_df['predicted_proba'] >= optimal_threshold).astype(int)

N_top = 20
print(f"\nTop {N_top} predicciones de compra para la semana '{new_target_week}' (con umbral {optimal_threshold:.2f}):")
print(new_predictions_df.sort_values(by='predicted_proba', ascending=False).head(N_top).to_string())

# --- 11. Save Predictions for 2025-W03 ---
output_filename_w03 = 'purchase_predictions_2025_W03.csv'
try:
    new_predictions_df.to_csv(output_filename_w03, index=False)
    print(f"El DataFrame de predicciones para 2025-W03 ha sido guardado exitosamente como '{output_filename_w03}'.")
except Exception as e:
    print(f"Error al guardar el archivo de predicciones para 2025-W03: {e}")
````

````python
print("--- Integrando compras efectivas de batch_t3 en el historial consolidado ---")

# Ensure df_clientes and df_productos are available (re-load if necessary)
# Assuming df_clientes and df_productos are already loaded in the environment from previous cells

# Initialize df_merged_updated if it's not defined (this handles the first batch integration)
if 'df_merged_updated' not in globals():
    df_merged_updated = df_merged.copy()
    print("df_merged_updated inicializado con df_merged (asumiendo primera integración de batch). Dims:", df_merged_updated.shape)

# --- Steps to integrate batch_t1 (re-executed for kernel consistency) ---
try:
    df_batch_t1_raw = pd.read_parquet('batch_t1.parquet')
    df_batch_t1_raw['purchase_date'] = pd.to_datetime(df_batch_t1_raw['purchase_date']).dt.date
    df_batch_t1_agg = df_batch_t1_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t1_effective_purchases = df_batch_t1_agg[df_batch_t1_agg['items_neto'] > 0].copy()
    df_batch_t1_effective_purchases_renamed = df_batch_t1_effective_purchases.rename(columns={'items_neto': 'items'})

    df_batch_t1_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t1_effective_purchases_renamed['purchase_date'])
    df_batch_t1_effective_purchases_renamed['year'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t1_effective_purchases_renamed['week'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t1_expanded = pd.merge(df_batch_t1_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t1_expanded = pd.merge(df_batch_t1_expanded, df_productos, on='product_id', how='left')
    df_batch_t1_expanded['year_week'] = df_batch_t1_expanded['year'].astype(str) + '-W' + df_batch_t1_expanded['week'].astype(str).str.zfill(2)

    columns_to_align_t1 = df_merged_updated.columns.tolist()
    missing_in_batch_t1_expanded = set(columns_to_align_t1) - set(df_batch_t1_expanded.columns)
    for col in missing_in_batch_t1_expanded:
        df_batch_t1_expanded[col] = np.nan
    df_batch_t1_aligned = df_batch_t1_expanded[columns_to_align_t1].copy()

    df_merged_updated = pd.concat([df_merged_updated, df_batch_t1_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t1 integrated into df_merged_updated. Dims:", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t1 (re-execution): {e}")

# --- Steps to integrate batch_t2 (re-executed for kernel consistency) ---
try:
    df_batch_t2_raw = pd.read_parquet('batch_t2.parquet')
    df_batch_t2_raw['purchase_date'] = pd.to_datetime(df_batch_t2_raw['purchase_date']).dt.date
    df_batch_t2_agg = df_batch_t2_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t2_effective_purchases = df_batch_t2_agg[df_batch_t2_agg['items_neto'] > 0].copy()
    df_batch_t2_effective_purchases_renamed = df_batch_t2_effective_purchases.rename(columns={'items_neto': 'items'})

    df_batch_t2_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t2_effective_purchases_renamed['purchase_date'])
    df_batch_t2_effective_purchases_renamed['year'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t2_effective_purchases_renamed['week'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t2_expanded = pd.merge(df_batch_t2_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t2_expanded = pd.merge(df_batch_t2_expanded, df_productos, on='product_id', how='left')
    df_batch_t2_expanded['year_week'] = df_batch_t2_expanded['year'].astype(str) + '-W' + df_batch_t2_expanded['week'].astype(str).str.zfill(2)

    columns_to_align_t2 = df_merged_updated.columns.tolist()
    missing_in_batch_t2_expanded = set(columns_to_align_t2) - set(df_batch_t2_expanded.columns)
    for col in missing_in_batch_t2_expanded:
        df_batch_t2_expanded[col] = np.nan
    df_batch_t2_aligned = df_batch_t2_expanded[columns_to_align_t2].copy()

    df_merged_updated = pd.concat([df_merged_updated, df_batch_t2_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t2 integrated into df_merged_updated. Dims:", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t2 (re-execution): {e}")


# --- Current subtask: Integrate batch_t3 ---
# 1. Renombrar 'items_neto' a 'items' en df_batch_t3_effective_purchases
df_batch_t3_effective_purchases_renamed = df_batch_t3_effective_purchases.rename(columns={'items_neto': 'items'})
print("Columna 'items_neto' renombrada a 'items' en df_batch_t3_effective_purchases.")

# 2. Realizar un merge del DataFrame df_batch_t3_effective_purchases_renamed con df_clientes y df_productos
df_batch_t3_expanded = pd.merge(
    df_batch_t3_effective_purchases_renamed,
    df_clientes,
    on='customer_id',
    how='left'
)
df_batch_t3_expanded = pd.merge(
    df_batch_t3_expanded,
    df_productos,
    on='product_id',
    how='left'
)
print("df_batch_t3 expandido con features de clientes y productos.")

# 3. Crear la columna 'year_week' en df_batch_t3_expanded
df_batch_t3_expanded['year_week'] = df_batch_t3_expanded['year'].astype(str) + '-W' + df_batch_t3_expanded['week'].astype(str).str.zfill(2)

# 4. Filtrar las columnas de df_batch_t3_expanded para que coincidan con las columnas de df_merged_updated y reordenarlas
merged_updated_cols = df_merged_updated.columns.tolist()

missing_in_batch_t3 = set(merged_updated_cols) - set(df_batch_t3_expanded.columns)
for col in missing_in_batch_t3:
    df_batch_t3_expanded[col] = np.nan
df_batch_t3_aligned = df_batch_t3_expanded[merged_updated_cols].copy()
print("Columnas de df_batch_t3_expanded alineadas con df_merged_updated.")

# 5. Concatenar df_merged_updated y df_batch_t3_aligned para crear df_merged_updated_final
df_merged_updated_final = pd.concat([df_merged_updated, df_batch_t3_aligned], ignore_index=True)
print(f"DataFrames concatenados. Dimensiones iniciales: {df_merged_updated_final.shape}")

# 6. Eliminar duplicados si los hubiera y ordenar
df_merged_updated_final.drop_duplicates(inplace=True)
df_merged_updated_final.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
df_merged_updated_final.reset_index(drop=True, inplace=True)
print(f"Duplicados eliminados y DataFrame ordenado. Dimensiones finales: {df_merged_updated_final.shape}")

# 7. Actualizar el DataFrame df_merged_updated con el contenido de df_merged_updated_final
df_merged_updated = df_merged_updated_final.copy()

# 8. Imprimir las nuevas dimensiones y el rango de fechas de df_merged_updated para verificar
print(f"\nNuevas dimensiones de df_merged_updated: {df_merged_updated.shape}")
print(f"Rango de fechas de df_merged_updated: {df_merged_updated['purchase_date'].min().strftime('%Y-%m-%d')} a {df_merged_updated['purchase_date'].max().strftime('%Y-%m-%d')}")

print("--- Integración de batch_t3 completada ---")
````

````python
import pandas as pd
from itertools import product
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from lightgbm import LGBMClassifier
import numpy as np
import warnings
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import optuna
import joblib

print("--- Initial Data Loading and Environment Setup ---")

# --- 1. Load Initial Datasets ---

try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    df_clientes = pd.read_parquet('clientes.parquet')
    df_productos = pd.read_parquet('productos.parquet')
    print("Datasets cargados exitosamente.")
except FileNotFoundError:
    print("Error: Uno o más archivos Parquet no se encontraron. Asegúrese de que estén en el directorio correcto.")
    exit() # Exit if crucial files are not found

# --- 2. Preprocess df_transacciones and Create df_merged ---
df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date

# Group transactions by customer, product, and day to sum 'items_neto'
df_trans_agg = df_transacciones.groupby(
    ['customer_id', 'product_id', 'purchase_date']
).agg(
    items_neto=('items', 'sum')
).reset_index()

# Filter for effective purchases (items_neto > 0)
df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)

df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])
df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week

df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')
df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')
df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)
print(f"Initial df_merged created. Dimensions: {df_merged.shape}")

# --- 3. Define Helper Functions for Feature Engineering ---
def year_week_to_date(yw_str):
    try:
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        return pd.NaT

def get_lag_week(target_week_str, lag):
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    if not target_weeks:
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])
    df_final = pd.merge(df_base, target_transactions, left_on=['customer_id', 'product_id', 'target_week'], right_on=['customer_id', 'product_id', 'year_week'], how='left')
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])
    df_final['target'] = df_final['target'].fillna(0).astype(int)
    return df_final

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    df_with_lags = base_df.copy()
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()
    for lag in lag_weeks:
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())
        df_with_lags = pd.merge(df_with_lags, hist_lag[merge_cols], on=['customer_id', 'product_id', 'lookup_week'], how='left')
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])
    return df_with_lags

def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])
    for window in windows:
        hist_data_indexed = hist_data.set_index('date')
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)
        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                rolling_col = rolling_col.reset_index()
                rolling_col = rolling_col.rename(columns={feature_col: col_name})
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(), on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})
                df_with_rolling = pd.merge(df_with_rolling, rolling_col[['customer_id', 'product_id', 'target_week', col_name]], on=['customer_id', 'product_id', 'target_week'], how='left')
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)
    return df_with_rolling

def add_temporal_features(df):
    df_temp = df.copy()
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    df_temp['month'] = df_temp['date'].dt.month.astype(int)
    df_temp = df_temp.drop(columns=['date'])
    return df_temp
print("Helper functions defined.")

# --- 4. Define Global Feature Lists and Selectors ---
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]
feature_columns = numeric_features + categorical_features
print("Global feature lists and selectors set up.")
````

````python
import pandas as pd
from itertools import product
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from lightgbm import LGBMClassifier
import numpy as np
import warnings
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import optuna
import joblib

print("--- Re-ejecutando pasos previos para asegurar el estado del kernel y generar predicciones para 2025-W03 ---")
print("ATENCIÓN: Asegúrese de que 'transacciones.parquet', 'clientes.parquet', 'productos.parquet', 'batch_t1.parquet', 'batch_t2.parquet', 'batch_t3.parquet' estén en el directorio de trabajo actual.")

# --- 1. Initial Data Loading & Preprocessing (up to 2024-W52) ---
try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    df_clientes = pd.read_parquet('clientes.parquet')
    df_productos = pd.read_parquet('productos.parquet')

    df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date
    df_trans_agg = df_transacciones.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
    df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)
    df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])
    df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
    df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week
    df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')
    df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')
    df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)
    print("Initial df_merged (up to 2024-W52) recreated. Dims: ", df_merged.shape)
except Exception as e:
    print(f"Error recreating initial df_merged: {e}")

# --- 2. Helper Functions for Feature Engineering ---
def year_week_to_date(yw_str):
    try:
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        return pd.NaT

def get_lag_week(target_week_str, lag):
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    if not target_weeks:
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])
    df_final = pd.merge(df_base, target_transactions, left_on=['customer_id', 'product_id', 'target_week'], right_on=['customer_id', 'product_id', 'year_week'], how='left')
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])
    df_final['target'] = df_final['target'].fillna(0).astype(int)
    return df_final

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    df_with_lags = base_df.copy()
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()
    for lag in lag_weeks:
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())
        df_with_lags = pd.merge(df_with_lags, hist_lag[merge_cols], on=['customer_id', 'product_id', 'lookup_week'], how='left')
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])
    return df_with_lags

def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])
    for window in windows:
        hist_data_indexed = hist_data.set_index('date')
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)
        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                rolling_col = rolling_col.reset_index()
                rolling_col = rolling_col.rename(columns={feature_col: col_name})
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(), on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})
                df_with_rolling = pd.merge(df_with_rolling, rolling_col[['customer_id', 'product_id', 'target_week', col_name]], on=['customer_id', 'product_id', 'target_week'], how='left')
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)
    return df_with_rolling

def add_temporal_features(df):
    df_temp = df.copy()
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    df_temp['month'] = df_temp['date'].dt.month.astype(int)
    df_temp = df_temp.drop(columns=['date'])
    return df_temp
print("Helper functions re-defined.")

# --- 3. Define Common Feature Engineering Variables ---
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]
feature_columns = numeric_features + categorical_features
print("Feature engineering variables and lists re-defined.")

# --- 4. Process and Integrate batch_t1 (up to 2025-W01) ---
try:
    df_batch_t1_raw = pd.read_parquet('batch_t1.parquet')
    df_batch_t1_raw['purchase_date'] = pd.to_datetime(df_batch_t1_raw['purchase_date']).dt.date
    df_batch_t1_agg = df_batch_t1_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t1_effective_purchases = df_batch_t1_agg[df_batch_t1_agg['items_neto'] > 0].copy()
    df_batch_t1_effective_purchases_renamed = df_batch_t1_effective_purchases.rename(columns={'items_neto': 'items'})

    # Fix: Ensure purchase_date is datetime and extract year/week for batch_t1
    df_batch_t1_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t1_effective_purchases_renamed['purchase_date'])
    df_batch_t1_effective_purchases_renamed['year'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t1_effective_purchases_renamed['week'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t1_expanded = pd.merge(df_batch_t1_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t1_expanded = pd.merge(df_batch_t1_expanded, df_productos, on='product_id', how='left')
    df_batch_t1_expanded['year_week'] = df_batch_t1_expanded['year'].astype(str) + '-W' + df_batch_t1_expanded['week'].astype(str).str.zfill(2)

    # Initialize df_merged_updated with df_merged plus batch_t1. For subsequent batches, this will be updated.
    columns_to_align = df_merged.columns.tolist()
    missing_in_batch_t1 = set(columns_to_align) - set(df_batch_t1_expanded.columns)
    for col in missing_in_batch_t1:
        df_batch_t1_expanded[col] = np.nan # Use np.nan for missing columns to maintain dtype compatibility
    df_batch_t1_aligned = df_batch_t1_expanded[columns_to_align].copy()

    df_merged_updated = pd.concat([df_merged, df_batch_t1_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t1 integrated into df_merged_updated. Dims: ", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t1: {e}")

# --- 5. Redefine Data Splits after batch_t1 ---
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_test = 4
num_weeks_val = 4
idx_val_start_updated = len(weeks_total_updated) - num_weeks_test - num_weeks_val
idx_test_start_updated = len(weeks_total_updated) - num_weeks_test
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:] # Contains 2025-W01 for evaluation

df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()
print("Data splits re-defined after batch_t1.")

# --- 6. Regenerate Feature Engineering after batch_t1 ---
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
df_train_hist_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)

df_base_val = create_target_base(df_val, weeks_val)
df_base_test = create_target_base(df_test, weeks_test)

df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

X_val = df_feat_val[feature_columns].copy() # Global X_val
y_val = df_feat_val['target'].copy() # Global y_val
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()
print("Feature engineering re-generated after batch_t1.")

# --- 7. Re-run Optuna Optimization and Model Training (first run) ---
warnings.filterwarnings('ignore', category=optuna.exceptions.ExperimentalWarning)
warnings.filterwarnings('ignore', category=UserWarning)

def objective(trial):
    num_imputer_strategy = trial.suggest_categorical('num_imputer_strategy', ['mean', 'median'])
    numeric_imputer = SimpleImputer(strategy=num_imputer_strategy)
    scaler_choice = trial.suggest_categorical('scaler', ['StandardScaler', 'MinMaxScaler', 'RobustScaler', 'None'])
    if scaler_choice == 'StandardScaler': scaler = StandardScaler()
    elif scaler_choice == 'MinMaxScaler': scaler = MinMaxScaler()
    elif scaler_choice == 'RobustScaler': scaler = RobustScaler()
    else: scaler = 'passthrough'
    numeric_transformer = make_pipeline(numeric_imputer, scaler) if scaler != 'passthrough' else make_pipeline(numeric_imputer)
    categorical_transformer = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    preprocessor = ColumnTransformer(transformers=[('num', numeric_transformer, numeric_features), ('cat', categorical_transformer, categorical_features)], remainder='drop')
    lgbm_params = {
        'objective': 'binary', 'metric': 'f1', 'random_state': 42, 'n_jobs': -1,
        'n_estimators': trial.suggest_int('n_estimators', 200, 1500, step=100),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 20, 100),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-3, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'is_unbalance': trial.suggest_categorical('is_unbalance', [True, False])
    }
    model = LGBMClassifier(**lgbm_params)
    pipeline = Pipeline(steps=[('preprocessor', preprocessor), ('classifier', model)])
    cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    try: scores = cross_val_score(pipeline, X_val, y_val, scoring='f1', cv=cv, n_jobs=-1); f1_mean = np.mean(scores); return f1_mean
    except Exception as e: return 0.0

study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=15, timeout=1800)
best_params = study.best_trial.params

num_imputer_final = SimpleImputer(strategy=best_params['num_imputer_strategy'])
scaler_choice_final = best_params['scaler']
if scaler_choice_final == 'StandardScaler': scaler_final = StandardScaler()
elif scaler_choice_final == 'MinMaxScaler': scaler_final = MinMaxScaler()
elif scaler_choice_final == 'RobustScaler': scaler_final = RobustScaler()
else: scaler_final = 'passthrough'
numeric_transformer_final = make_pipeline(num_imputer_final, scaler_final) if scaler_final != 'passthrough' else make_pipeline(num_imputer_final)
categorical_transformer_final = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
preprocessor_final = ColumnTransformer(transformers=[('num', numeric_transformer_final, numeric_features), ('cat', categorical_transformer_final, categorical_features)], remainder='drop')
lgbm_best_params = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_final = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params)
final_pipeline = Pipeline(steps=[('preprocessor', preprocessor_final), ('classifier', lgbm_final)])
final_pipeline.fit(X_val, y_val)

y_pred_proba_val = final_pipeline.predict_proba(X_val)[:, 1]
optimal_threshold = 0
best_f1_score = 0
for threshold in np.arange(0.0, 1.01, 0.01):
    y_pred_binary = (y_pred_proba_val >= threshold).astype(int)
    f1 = f1_score(y_val, y_pred_binary, pos_label=1, zero_division=0)
    if f1 > best_f1_score:
        best_f1_score = f1
        optimal_threshold = threshold
print(f"Model re-trained (after batch_t1) and optimal_threshold found: {optimal_threshold:.2f}")

# --- 6. Process and Integrate batch_t2 (up to 2025-W02) ---
try:
    df_batch_t2_raw = pd.read_parquet('batch_t2.parquet')
    df_batch_t2_raw['purchase_date'] = pd.to_datetime(df_batch_t2_raw['purchase_date']).dt.date
    df_batch_t2_agg = df_batch_t2_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t2_effective_purchases = df_batch_t2_agg[df_batch_t2_agg['items_neto'] > 0].copy()
    df_batch_t2_effective_purchases_renamed = df_batch_t2_effective_purchases.rename(columns={'items_neto': 'items'})

    # Fix: Ensure purchase_date is datetime and extract year/week for batch_t2
    df_batch_t2_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t2_effective_purchases_renamed['purchase_date'])
    df_batch_t2_effective_purchases_renamed['year'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t2_effective_purchases_renamed['week'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t2_expanded = pd.merge(df_batch_t2_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t2_expanded = pd.merge(df_batch_t2_expanded, df_productos, on='product_id', how='left')
    df_batch_t2_expanded['year_week'] = df_batch_t2_expanded['year'].astype(str) + '-W' + df_batch_t2_expanded['week'].astype(str).str.zfill(2)

    # Align columns before concatenation
    columns_to_align = df_merged_updated.columns.tolist()
    missing_in_batch_t2 = set(columns_to_align) - set(df_batch_t2_expanded.columns)
    for col in missing_in_batch_t2:
        df_batch_t2_expanded[col] = np.nan # Use np.nan for missing columns to maintain dtype compatibility
    df_batch_t2_aligned = df_batch_t2_expanded[columns_to_align].copy()

    df_merged_updated_final = pd.concat([df_merged_updated, df_batch_t2_aligned], ignore_index=True)
    df_merged_updated_final.drop_duplicates(inplace=True)
    df_merged_updated_final.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated_final.reset_index(drop=True, inplace=True)
    df_merged_updated = df_merged_updated_final.copy()
    print("batch_t2 integrated into df_merged_updated. Dims: ", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t2: {e}")

# --- 7. Redefine Data Splits after batch_t2 ---
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_test = 4
num_weeks_val = 4
idx_val_start_updated = len(weeks_total_updated) - num_weeks_test - num_weeks_val
idx_test_start_updated = len(weeks_total_updated) - num_weeks_test
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:] # Contains 2025-W02 for evaluation

df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()
print("Data splits re-defined after batch_t2.")

# --- 8. Regenerate Feature Engineering after batch_t2 ---
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
df_train_hist_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)

df_base_val = create_target_base(df_val, weeks_val)
df_base_test = create_target_base(df_test, weeks_test)

df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()
print("Feature engineering re-generated after batch_t2.")

# --- 9. Re-train Model after batch_t2 ---
lgbm_best_params_re_fit = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_re_fit = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params_re_fit)

num_imputer_re_fit = SimpleImputer(strategy=best_params['num_imputer_strategy'])
scaler_choice_re_fit = best_params['scaler']
if scaler_choice_re_fit == 'StandardScaler': scaler_re_fit = StandardScaler()
elif scaler_choice_re_fit == 'MinMaxScaler': scaler_re_fit = MinMaxScaler()
elif scaler_choice_re_fit == 'RobustScaler': scaler_re_fit = RobustScaler()
else: scaler_re_fit = 'passthrough'
numeric_transformer_re_fit = make_pipeline(num_imputer_re_fit, scaler_re_fit) if scaler_re_fit != 'passthrough' else make_pipeline(num_imputer_re_fit)
categorical_transformer_re_fit = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
preprocessor_final = ColumnTransformer(transformers=[('num', numeric_transformer_re_fit, numeric_features), ('cat', categorical_transformer_re_fit, categorical_features)], remainder='drop')

final_pipeline = Pipeline(steps=[('preprocessor', preprocessor_final), ('classifier', lgbm_re_fit)])

final_pipeline.fit(X_val, y_val)
print("Model re-trained with data including batch_t2.")

# Re-calculate optimal_threshold after retraining with batch_t2 data
y_pred_proba_val_current = final_pipeline.predict_proba(X_val)[:, 1]
optimal_threshold = 0
best_f1_score_current = 0
for threshold in np.arange(0.0, 1.01, 0.01):
    y_pred_binary_current = (y_pred_proba_val_current >= threshold).astype(int)
    f1_current = f1_score(y_val, y_pred_binary_current, pos_label=1, zero_division=0)
    if f1_current > best_f1_score_current:
        best_f1_score_current = f1_current
        optimal_threshold = threshold
print(f"Optimal threshold recalculated after re-training: {optimal_threshold:.2f}")

# --- 10. Prepare and Generate Predictions for 2025-W03 ---
new_target_week = '2025-W03'

current_historical_data = df_merged_updated.copy()
current_historical_data['purchased'] = 1 # Add 'purchased' column for feature generation

active_customers_for_new_week = df_clientes['customer_id'].unique()
active_products_for_new_week = df_productos['product_id'].unique()
df_new_predictions_base = pd.DataFrame(list(product(active_customers_for_new_week, active_products_for_new_week, [new_target_week])), columns=['customer_id', 'product_id', 'target_week'])
print(f"Tabla base para {new_target_week} creada. Dimensions: {df_new_predictions_base.shape}")

df_new_features = add_lag_features(
    base_df=df_new_predictions_base,
    transactions_hist_df=current_historical_data,
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

df_new_features = add_rolling_features(
    base_df=df_new_features,
    transactions_hist_agg_df=current_historical_data.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1),
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)
print(f"Features de Lag y Rolling añadidas. Dimensions: {df_new_features.shape}")

df_new_features = pd.merge(df_new_features, productos_sel, on='product_id', how='left')
df_new_features = pd.merge(df_new_features, clientes_sel, on='customer_id', how='left')

df_new_features = add_temporal_features(df_new_features)
print(f"Features estáticas y temporales añadidas. Dimensions: {df_new_features.shape}")

X_new_predict = df_new_features[feature_columns].copy()
print(f"X_new_predict listo para la inferencia. Dimensions: {X_new_predict.shape}")

new_probabilities = final_pipeline.predict_proba(X_new_predict)
predicted_new_purchase_proba = new_probabilities[:, 1]

new_predictions_df = df_new_features[['customer_id', 'product_id', 'target_week']].copy()
new_predictions_df['predicted_proba'] = predicted_new_purchase_proba
new_predictions_df['predicted_purchase_binary'] = (new_predictions_df['predicted_proba'] >= optimal_threshold).astype(int)

N_top = 20
print(f"\nTop {N_top} predicciones de compra para la semana '{new_target_week}' (con umbral {optimal_threshold:.2f}):")
print(new_predictions_df.sort_values(by='predicted_proba', ascending=False).head(N_top).to_string())

# --- 11. Save Predictions for 2025-W03 ---
output_filename_w03 = 'purchase_predictions_2025_W03.csv'
try:
    new_predictions_df.to_csv(output_filename_w03, index=False)
    print(f"El DataFrame de predicciones para 2025-W03 ha sido guardado exitosamente como '{output_filename_w03}'.")
except Exception as e:
    print(f"Error al guardar el archivo de predicciones para 2025-W03: {e}")
````

````python
print("--- Redefiniendo particiones de datos con historial actualizado ---")

# 1. Obtener las semanas totales únicas de `df_merged_updated` y ordénelas.
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_total_updated = len(weeks_total_updated)

# 2. Mantener las definiciones de num_weeks_test y num_weeks_val como 4 semanas cada una.
num_weeks_test = 4
num_weeks_val = 4

# 3. Calcular los índices de inicio para los conjuntos de validación y prueba
idx_val_start_updated = num_weeks_total_updated - num_weeks_test - num_weeks_val
idx_test_start_updated = num_weeks_total_updated - num_weeks_test

# 4. Crear las listas de semanas para entrenamiento, validación y prueba
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:]

print(f"Semanas para Entrenamiento actualizadas ({len(weeks_train)}): {weeks_train[0]} a {weeks_train[-1]}")
print(f"Semanas para Validación actualizadas ({len(weeks_val)}): {weeks_val[0]} a {weeks_val[-1]}")
print(f"Semanas para Prueba actualizadas ({len(weeks_test)}): {weeks_test[0]} a {weeks_test[-1]}")

# 5. Generar los nuevos DataFrames df_train, df_val y df_test
df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()

# 6. Imprimir dimensiones y rangos de fechas para verificación
print(f"\nDimensiones df_train actualizado: {df_train.shape}")
print(f"Fechas en df_train actualizado: {df_train['purchase_date'].min().strftime('%Y-%m-%d')} a {df_train['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_val actualizado: {df_val.shape}")
print(f"Fechas en df_val actualizado: {df_val['purchase_date'].min().strftime('%Y-%m-%d')} a {df_val['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_test actualizado: {df_test.shape}")
print(f"Fechas en df_test actualizado: {df_test['purchase_date'].min().strftime('%Y-%m-%d')} a {df_test['purchase_date'].max().strftime('%Y-%m-%d')}")

print("--- Redefinición de particiones completada ---")
````

````python
print("--- Iniciando regeneración de Feature Engineering con historial actualizado ---")

# 1. Recrear el df_train_val DataFrame con los datos actualizados
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
print(f"df_train_val recreado. Dimensiones: {df_train_val.shape}")

# 2. Recrear df_train_hist_agg (historial para el conjunto de validación)
df_train_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_hist_agg = df_train_agg.copy()
print(f"df_train_hist_agg recreado. Dimensiones: {df_train_hist_agg.shape}")

# 3. Recrear df_train_val_hist_agg (historial para el conjunto de prueba)
df_train_val_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val_agg.copy()
print(f"df_train_val_hist_agg recreado. Dimensiones: {df_train_val_hist_agg.shape}")

# 4. Recrear df_base_val y df_base_test
# Se asume que `create_target_base` está definido en el contexto
# `target_weeks_val` y `target_weeks_test` se deben haber actualizado previamente en el paso de Holdout
print("\n--- Recreando bases para validación y prueba ---")
df_base_val = create_target_base(df_val, weeks_val) # Usar df_val y las semanas actualizadas
df_base_test = create_target_base(df_test, weeks_test) # Usar df_test y las semanas actualizadas

print("\n--- Aplicando Features de Lag ---")
# 5. Aplicar add_lag_features a df_base_val
df_feat_val = add_lag_features(
    base_df=df_base_val,
    transactions_hist_df=df_train_hist_agg, # Historial para validar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

# 6. Aplicar add_lag_features a df_base_test
df_feat_test = add_lag_features(
    base_df=df_base_test,
    transactions_hist_df=df_train_val_hist_agg, # Historial para probar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

print("\n--- Aplicando Features de Ventana Móvil ---")
# 7. Aplicar add_rolling_features a df_feat_val
df_feat_val = add_rolling_features(
    base_df=df_feat_val,
    transactions_hist_agg_df=df_train_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

# 8. Aplicar add_rolling_features a df_feat_test
df_feat_test = add_rolling_features(
    base_df=df_feat_test,
    transactions_hist_agg_df=df_train_val_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

print("\n--- Añadiendo Features Estáticas y Temporales ---")
# 9. Mergear features de producto con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')

# 10. Mergear features de cliente con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

# 11. Aplicar add_temporal_features a df_feat_val y df_feat_test
df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

print("--- Verificación de Features Regeneradas ---")
print(f"\nDimensiones de df_feat_val: {df_feat_val.shape}")
print("Primeras 5 filas de df_feat_val:")
print(df_feat_val.head().to_string())
print("\nNaN check en df_feat_val:")
nan_check_val = df_feat_val.isnull().sum()
print(nan_check_val[nan_check_val > 0].to_string())

print(f"\nDimensiones de df_feat_test: {df_feat_test.shape}")
print("Primeras 5 filas de df_feat_test:")
print(df_feat_test.head().to_string())
print("\nNaN check en df_feat_test:")
nan_check_test = df_feat_test.isnull().sum()
print(nan_check_test[nan_check_test > 0].to_string())

# 12. Actualizar X_val, y_val, X_test, y_test
X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()

print("\n--- Feature Engineering regenerado y conjuntos X, y actualizados ---")
````

````python
print("--- Reentrenando el Modelo LightGBM Optimizado ---")

# Obtener los mejores parámetros de la optimización anterior
best_params = study.best_trial.params

# 1. Recrear el preprocesador con los mejores parámetros
num_imputer_strategy_final = best_params['num_imputer_strategy']
numeric_imputer_re_fit = SimpleImputer(strategy=num_imputer_strategy_final)

scaler_choice_re_fit = best_params['scaler']
if scaler_choice_re_fit == 'StandardScaler':
    scaler_re_fit = StandardScaler()
elif scaler_choice_re_fit == 'MinMaxScaler':
    scaler_re_fit = MinMaxScaler()
elif scaler_choice_re_fit == 'RobustScaler':
    scaler_re_fit = RobustScaler()
else:
    scaler_re_fit = 'passthrough'

numeric_transformer_re_fit = make_pipeline(numeric_imputer_re_fit, scaler_re_fit) if scaler_re_fit != 'passthrough' else make_pipeline(numeric_imputer_re_fit)
categorical_transformer_re_fit = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))

preprocessor_re_fit = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_re_fit, numeric_features),
        ('cat', categorical_transformer_re_fit, categorical_features)
    ],
    remainder='drop'
)
print("Preprocesador recreado con los mejores parámetros.")

# 2. Recrear el clasificador LightGBM con los mejores parámetros
lgbm_best_params_re_fit = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_re_fit = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params_re_fit)
print("Clasificador LightGBM recreado con los mejores parámetros.")

# 3. Crear el final_pipeline recreado
final_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_re_fit),
    ('classifier', lgbm_re_fit)
])
print("Pipeline final recreado.")

# 4. Entrenar el final_pipeline con los datos de entrenamiento actualizados (X_val y y_val)
print("Entrenando el pipeline final con X_val (historial actualizado) y y_val...")
final_pipeline.fit(X_val, y_val)
print("Reentrenamiento completado con éxito.")

# Opcional: Evaluar rápidamente el modelo reentrenado en X_val
from sklearn.metrics import classification_report
y_pred_val_re_fit = final_pipeline.predict(X_val)
print("\nClassification Report para X_val (después del reentrenamiento):")
print(classification_report(y_val, y_pred_val_re_fit, zero_division=0))
````

````python
print("--- Preparando datos para predicción (2025-W04) ---")

# 1. Definir la nueva semana objetivo
new_target_week = '2025-W04'

# 2. Asegurarse de que `current_historical_data` contenga `df_merged_updated` (que incluye el historial hasta 2025-W03) y añadir una columna 'purchased' con valor 1.
current_historical_data = df_merged_updated.copy()
current_historical_data['purchased'] = 1

# 3. Crear un DataFrame base (`df_new_predictions_base`) con todas las combinaciones ùnicas de `customer_id` y `product_id`
active_customers_for_new_week = df_clientes['customer_id'].unique()
active_products_for_new_week = df_productos['product_id'].unique()
df_new_predictions_base = pd.DataFrame(list(product(active_customers_for_new_week, active_products_for_new_week, [new_target_week])), columns=['customer_id', 'product_id', 'target_week'])
print(f"Tabla base para {new_target_week} creada. Dimensiones: {df_new_predictions_base.shape}")

# 4. Aplicar la función `add_lag_features`
df_new_features = add_lag_features(
    base_df=df_new_predictions_base,
    transactions_hist_df=current_historical_data,
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

# 5. Aplicar la función `add_rolling_features`
df_new_features = add_rolling_features(
    base_df=df_new_features,
    transactions_hist_agg_df=current_historical_data.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1),
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)
print(f"Features de Lag y Rolling añadidas. Dimensiones: {df_new_features.shape}")

# 6. Añadir las características estáticas de producto y cliente
df_new_features = pd.merge(df_new_features, productos_sel, on='product_id', how='left')
df_new_features = pd.merge(df_new_features, clientes_sel, on='customer_id', how='left')

# 7. Aplicar la función `add_temporal_features`
df_new_features = add_temporal_features(df_new_features)
print(f"Features estáticas y temporales añadidas. Dimensiones: {df_new_features.shape}")

# 8. Seleccionar solo las `feature_columns`
X_new_predict = df_new_features[feature_columns].copy()
print(f"X_new_predict listo para la inferencia. Dimensiones: {X_new_predict.shape}")

# 9. Realizar las predicciones de probabilidad con el modelo cargado (final_pipeline)
new_probabilities = final_pipeline.predict_proba(X_new_predict)
predicted_new_purchase_proba = new_probabilities[:, 1]

# 10. Aplicar el umbral óptimo para obtener predicciones binarias
new_predictions_df = df_new_features[['customer_id', 'product_id', 'target_week']].copy()
new_predictions_df['predicted_proba'] = predicted_new_purchase_proba
new_predictions_df['predicted_purchase_binary'] = (new_predictions_df['predicted_proba'] >= optimal_threshold).astype(int)

# 11. Formatear y mostrar las top N predicciones binarias y sus probabilidades
N_top = 20
print(f"\nTop {N_top} predicciones de compra para la semana '{new_target_week}' (con umbral {optimal_threshold:.2f}):")
print(new_predictions_df.sort_values(by='predicted_proba', ascending=False).head(N_top).to_string())
````

````python
import pandas as pd
from itertools import product
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from lightgbm import LGBMClassifier
import numpy as np
import warnings
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import optuna
import joblib

print("--- Initial Data Loading and Environment Setup ---")

# --- 1. Load Initial Datasets ---

try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    df_clientes = pd.read_parquet('clientes.parquet')
    df_productos = pd.read_parquet('productos.parquet')
    print("Datasets cargados exitosamente.")
except FileNotFoundError:
    print("Error: Uno o más archivos Parquet no se encontraron. Asegúrese de que estén en el directorio correcto.")
    exit() # Exit if crucial files are not found

# --- 2. Preprocess df_transacciones and Create df_merged ---
df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date

# Group transactions by customer, product, and day to sum 'items_neto'
df_trans_agg = df_transacciones.groupby(
    ['customer_id', 'product_id', 'purchase_date']
).agg(
    items_neto=('items', 'sum')
).reset_index()

# Filter for effective purchases (items_neto > 0)
df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)

df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])
df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week

df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')
df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')
df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)
print(f"Initial df_merged created. Dimensions: {df_merged.shape}")

# --- 3. Define Helper Functions for Feature Engineering ---
def year_week_to_date(yw_str):
    try:
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        return pd.NaT

def get_lag_week(target_week_str, lag):
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    if not target_weeks:
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])
    df_final = pd.merge(df_base, target_transactions, left_on=['customer_id', 'product_id', 'target_week'], right_on=['customer_id', 'product_id', 'year_week'], how='left')
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])
    df_final['target'] = df_final['target'].fillna(0).astype(int)
    return df_final

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    df_with_lags = base_df.copy()
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()
    for lag in lag_weeks:
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())
        df_with_lags = pd.merge(df_with_lags, hist_lag[merge_cols], on=['customer_id', 'product_id', 'lookup_week'], how='left')
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])
    return df_with_lags

def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])
    for window in windows:
        hist_data_indexed = hist_data.set_index('date')
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)
        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                rolling_col = rolling_col.reset_index()
                rolling_col = rolling_col.rename(columns={feature_col: col_name})
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(), on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})
                df_with_rolling = pd.merge(df_with_rolling, rolling_col[['customer_id', 'product_id', 'target_week', col_name]], on=['customer_id', 'product_id', 'target_week'], how='left')
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)
    return df_with_rolling

def add_temporal_features(df):
    df_temp = df.copy()
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    df_temp['month'] = df_temp['date'].dt.month.astype(int)
    df_temp = df_temp.drop(columns=['date'])
    return df_temp
print("Helper functions defined.")

# --- 4. Define Global Feature Lists and Selectors ---
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]
feature_columns = numeric_features + categorical_features
print("Global feature lists and selectors set up.")
````

````python
import pandas as pd
from itertools import product
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from lightgbm import LGBMClassifier
import numpy as np
import warnings
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import optuna
import joblib

print("--- Re-ejecutando pasos previos para asegurar el estado del kernel y generar predicciones para 2025-W03 ---")
print("ATENCIÓN: Asegúrese de que 'transacciones.parquet', 'clientes.parquet', 'productos.parquet', 'batch_t1.parquet', 'batch_t2.parquet', 'batch_t3.parquet' estén en el directorio de trabajo actual.")

# --- 1. Initial Data Loading & Preprocessing (up to 2024-W52) ---
try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    df_clientes = pd.read_parquet('clientes.parquet')
    df_productos = pd.read_parquet('productos.parquet')

    df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date
    df_trans_agg = df_transacciones.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
    df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)
    df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])
    df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
    df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week
    df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')
    df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')
    df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)
    print("Initial df_merged (up to 2024-W52) recreated. Dims: ", df_merged.shape)
except Exception as e:
    print(f"Error recreating initial df_merged: {e}")

# --- 2. Helper Functions for Feature Engineering ---
def year_week_to_date(yw_str):
    try:
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        return pd.NaT

def get_lag_week(target_week_str, lag):
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    if not target_weeks:
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])
    df_final = pd.merge(df_base, target_transactions, left_on=['customer_id', 'product_id', 'target_week'], right_on=['customer_id', 'product_id', 'year_week'], how='left')
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])
    df_final['target'] = df_final['target'].fillna(0).astype(int)
    return df_final

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    df_with_lags = base_df.copy()
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()
    for lag in lag_weeks:
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())
        df_with_lags = pd.merge(df_with_lags, hist_lag[merge_cols], on=['customer_id', 'product_id', 'lookup_week'], how='left')
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])
    return df_with_lags

def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])
    for window in windows:
        hist_data_indexed = hist_data.set_index('date')
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)
        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                rolling_col = rolling_col.reset_index()
                rolling_col = rolling_col.rename(columns={feature_col: col_name})
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(), on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})
                df_with_rolling = pd.merge(df_with_rolling, rolling_col[['customer_id', 'product_id', 'target_week', col_name]], on=['customer_id', 'product_id', 'target_week'], how='left')
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)
    return df_with_rolling

def add_temporal_features(df):
    df_temp = df.copy()
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    df_temp['month'] = df_temp['date'].dt.month.astype(int)
    df_temp = df_temp.drop(columns=['date'])
    return df_temp
print("Helper functions re-defined.")

# --- 3. Define Common Feature Engineering Variables ---
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]
feature_columns = numeric_features + categorical_features
print("Feature engineering variables and lists re-defined.")

# --- 4. Process and Integrate batch_t1 (up to 2025-W01) ---
try:
    df_batch_t1_raw = pd.read_parquet('batch_t1.parquet')
    df_batch_t1_raw['purchase_date'] = pd.to_datetime(df_batch_t1_raw['purchase_date']).dt.date
    df_batch_t1_agg = df_batch_t1_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t1_effective_purchases = df_batch_t1_agg[df_batch_t1_agg['items_neto'] > 0].copy()
    df_batch_t1_effective_purchases_renamed = df_batch_t1_effective_purchases.rename(columns={'items_neto': 'items'})

    # Fix: Ensure purchase_date is datetime and extract year/week for batch_t1
    df_batch_t1_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t1_effective_purchases_renamed['purchase_date'])
    df_batch_t1_effective_purchases_renamed['year'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t1_effective_purchases_renamed['week'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t1_expanded = pd.merge(df_batch_t1_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t1_expanded = pd.merge(df_batch_t1_expanded, df_productos, on='product_id', how='left')
    df_batch_t1_expanded['year_week'] = df_batch_t1_expanded['year'].astype(str) + '-W' + df_batch_t1_expanded['week'].astype(str).str.zfill(2)

    # Initialize df_merged_updated with df_merged plus batch_t1. For subsequent batches, this will be updated.
    columns_to_align = df_merged.columns.tolist()
    missing_in_batch_t1 = set(columns_to_align) - set(df_batch_t1_expanded.columns)
    for col in missing_in_batch_t1:
        df_batch_t1_expanded[col] = np.nan # Use np.nan for missing columns to maintain dtype compatibility
    df_batch_t1_aligned = df_batch_t1_expanded[columns_to_align].copy()

    df_merged_updated = pd.concat([df_merged, df_batch_t1_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t1 integrated into df_merged_updated. Dims: ", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t1: {e}")

# --- 5. Redefine Data Splits after batch_t1 ---
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_test = 4
num_weeks_val = 4
idx_val_start_updated = len(weeks_total_updated) - num_weeks_test - num_weeks_val
idx_test_start_updated = len(weeks_total_updated) - num_weeks_test
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:] # Contains 2025-W01 for evaluation

df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()
print("Data splits re-defined after batch_t1.")

# --- 6. Regenerate Feature Engineering after batch_t1 ---
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
df_train_hist_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)

df_base_val = create_target_base(df_val, weeks_val)
df_base_test = create_target_base(df_test, weeks_test)

df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

X_val = df_feat_val[feature_columns].copy() # Global X_val
y_val = df_feat_val['target'].copy() # Global y_val
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()
print("Feature engineering re-generated after batch_t1.")

# --- 7. Re-run Optuna Optimization and Model Training (first run) ---
warnings.filterwarnings('ignore', category=optuna.exceptions.ExperimentalWarning)
warnings.filterwarnings('ignore', category=UserWarning)

def objective(trial):
    num_imputer_strategy = trial.suggest_categorical('num_imputer_strategy', ['mean', 'median'])
    numeric_imputer = SimpleImputer(strategy=num_imputer_strategy)
    scaler_choice = trial.suggest_categorical('scaler', ['StandardScaler', 'MinMaxScaler', 'RobustScaler', 'None'])
    if scaler_choice == 'StandardScaler': scaler = StandardScaler()
    elif scaler_choice == 'MinMaxScaler': scaler = MinMaxScaler()
    elif scaler_choice == 'RobustScaler': scaler = RobustScaler()
    else: scaler = 'passthrough'
    numeric_transformer = make_pipeline(numeric_imputer, scaler) if scaler != 'passthrough' else make_pipeline(numeric_imputer)
    categorical_transformer = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    preprocessor = ColumnTransformer(transformers=[('num', numeric_transformer, numeric_features), ('cat', categorical_transformer, categorical_features)], remainder='drop')
    lgbm_params = {
        'objective': 'binary', 'metric': 'f1', 'random_state': 42, 'n_jobs': -1,
        'n_estimators': trial.suggest_int('n_estimators', 200, 1500, step=100),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 20, 100),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-3, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'is_unbalance': trial.suggest_categorical('is_unbalance', [True, False])
    }
    model = LGBMClassifier(**lgbm_params)
    pipeline = Pipeline(steps=[('preprocessor', preprocessor), ('classifier', model)])
    cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    try: scores = cross_val_score(pipeline, X_val, y_val, scoring='f1', cv=cv, n_jobs=-1); f1_mean = np.mean(scores); return f1_mean
    except Exception as e: return 0.0

study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=15, timeout=1800)
best_params = study.best_trial.params

num_imputer_final = SimpleImputer(strategy=best_params['num_imputer_strategy'])
scaler_choice_final = best_params['scaler']
if scaler_choice_final == 'StandardScaler': scaler_final = StandardScaler()
elif scaler_choice_final == 'MinMaxScaler': scaler_final = MinMaxScaler()
elif scaler_choice_final == 'RobustScaler': scaler_final = RobustScaler()
else: scaler_final = 'passthrough'
numeric_transformer_final = make_pipeline(num_imputer_final, scaler_final) if scaler_final != 'passthrough' else make_pipeline(num_imputer_final)
categorical_transformer_final = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
preprocessor_final = ColumnTransformer(transformers=[('num', numeric_transformer_final, numeric_features), ('cat', categorical_transformer_final, categorical_features)], remainder='drop')
lgbm_best_params = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_final = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params)
final_pipeline = Pipeline(steps=[('preprocessor', preprocessor_final), ('classifier', lgbm_final)])
final_pipeline.fit(X_val, y_val)

y_pred_proba_val = final_pipeline.predict_proba(X_val)[:, 1]
optimal_threshold = 0
best_f1_score = 0
for threshold in np.arange(0.0, 1.01, 0.01):
    y_pred_binary = (y_pred_proba_val >= threshold).astype(int)
    f1 = f1_score(y_val, y_pred_binary, pos_label=1, zero_division=0)
    if f1 > best_f1_score:
        best_f1_score = f1
        optimal_threshold = threshold
print(f"Model re-trained (after batch_t1) and optimal_threshold found: {optimal_threshold:.2f}")

# --- 6. Process and Integrate batch_t2 (up to 2025-W02) ---
try:
    df_batch_t2_raw = pd.read_parquet('batch_t2.parquet')
    df_batch_t2_raw['purchase_date'] = pd.to_datetime(df_batch_t2_raw['purchase_date']).dt.date
    df_batch_t2_agg = df_batch_t2_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t2_effective_purchases = df_batch_t2_agg[df_batch_t2_agg['items_neto'] > 0].copy()
    df_batch_t2_effective_purchases_renamed = df_batch_t2_effective_purchases.rename(columns={'items_neto': 'items'})

    # Fix: Ensure purchase_date is datetime and extract year/week for batch_t2
    df_batch_t2_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t2_effective_purchases_renamed['purchase_date'])
    df_batch_t2_effective_purchases_renamed['year'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t2_effective_purchases_renamed['week'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t2_expanded = pd.merge(df_batch_t2_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t2_expanded = pd.merge(df_batch_t2_expanded, df_productos, on='product_id', how='left')
    df_batch_t2_expanded['year_week'] = df_batch_t2_expanded['year'].astype(str) + '-W' + df_batch_t2_expanded['week'].astype(str).str.zfill(2)

    # Align columns before concatenation
    columns_to_align = df_merged_updated.columns.tolist()
    missing_in_batch_t2 = set(columns_to_align) - set(df_batch_t2_expanded.columns)
    for col in missing_in_batch_t2:
        df_batch_t2_expanded[col] = np.nan # Use np.nan for missing columns to maintain dtype compatibility
    df_batch_t2_aligned = df_batch_t2_expanded[columns_to_align].copy()

    df_merged_updated_final = pd.concat([df_merged_updated, df_batch_t2_aligned], ignore_index=True)
    df_merged_updated_final.drop_duplicates(inplace=True)
    df_merged_updated_final.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated_final.reset_index(drop=True, inplace=True)
    df_merged_updated = df_merged_updated_final.copy()
    print("batch_t2 integrated into df_merged_updated. Dims: ", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t2: {e}")

# --- 7. Redefine Data Splits after batch_t2 ---
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_test = 4
num_weeks_val = 4
idx_val_start_updated = len(weeks_total_updated) - num_weeks_test - num_weeks_val
idx_test_start_updated = len(weeks_total_updated) - num_weeks_test
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:] # Contains 2025-W02 for evaluation

df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()
print("Data splits re-defined after batch_t2.")

# --- 8. Regenerate Feature Engineering after batch_t2 ---
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
df_train_hist_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)

df_base_val = create_target_base(df_val, weeks_val)
df_base_test = create_target_base(df_test, weeks_test)

df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()
print("Feature engineering re-generated after batch_t2.")

# --- 9. Re-train Model after batch_t2 ---
lgbm_best_params_re_fit = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_re_fit = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params_re_fit)

num_imputer_re_fit = SimpleImputer(strategy=best_params['num_imputer_strategy'])
scaler_choice_re_fit = best_params['scaler']
if scaler_choice_re_fit == 'StandardScaler': scaler_re_fit = StandardScaler()
elif scaler_choice_re_fit == 'MinMaxScaler': scaler_re_fit = MinMaxScaler()
elif scaler_choice_re_fit == 'RobustScaler': scaler_re_fit = RobustScaler()
else: scaler_re_fit = 'passthrough'
numeric_transformer_re_fit = make_pipeline(num_imputer_re_fit, scaler_re_fit) if scaler_re_fit != 'passthrough' else make_pipeline(num_imputer_re_fit)
categorical_transformer_re_fit = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
preprocessor_final = ColumnTransformer(transformers=[('num', numeric_transformer_re_fit, numeric_features), ('cat', categorical_transformer_re_fit, categorical_features)], remainder='drop')

final_pipeline = Pipeline(steps=[('preprocessor', preprocessor_final), ('classifier', lgbm_re_fit)])

final_pipeline.fit(X_val, y_val)
print("Model re-trained with data including batch_t2.")

# Re-calculate optimal_threshold after retraining with batch_t2 data
y_pred_proba_val_current = final_pipeline.predict_proba(X_val)[:, 1]
optimal_threshold = 0
best_f1_score_current = 0
for threshold in np.arange(0.0, 1.01, 0.01):
    y_pred_binary_current = (y_pred_proba_val_current >= threshold).astype(int)
    f1_current = f1_score(y_val, y_pred_binary_current, pos_label=1, zero_division=0)
    if f1_current > best_f1_score_current:
        best_f1_score_current = f1_current
        optimal_threshold = threshold
print(f"Optimal threshold recalculated after re-training: {optimal_threshold:.2f}")

# --- 10. Prepare and Generate Predictions for 2025-W03 ---
new_target_week = '2025-W03'

current_historical_data = df_merged_updated.copy()
current_historical_data['purchased'] = 1 # Add 'purchased' column for feature generation

active_customers_for_new_week = df_clientes['customer_id'].unique()
active_products_for_new_week = df_productos['product_id'].unique()
df_new_predictions_base = pd.DataFrame(list(product(active_customers_for_new_week, active_products_for_new_week, [new_target_week])), columns=['customer_id', 'product_id', 'target_week'])
print(f"Tabla base para {new_target_week} creada. Dimensions: {df_new_predictions_base.shape}")

df_new_features = add_lag_features(
    base_df=df_new_predictions_base,
    transactions_hist_df=current_historical_data,
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

df_new_features = add_rolling_features(
    base_df=df_new_features,
    transactions_hist_agg_df=current_historical_data.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1),
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)
print(f"Features de Lag y Rolling añadidas. Dimensions: {df_new_features.shape}")

df_new_features = pd.merge(df_new_features, productos_sel, on='product_id', how='left')
df_new_features = pd.merge(df_new_features, clientes_sel, on='customer_id', how='left')

df_new_features = add_temporal_features(df_new_features)
print(f"Features estáticas y temporales añadidas. Dimensions: {df_new_features.shape}")

X_new_predict = df_new_features[feature_columns].copy()
print(f"X_new_predict listo para la inferencia. Dimensions: {X_new_predict.shape}")

new_probabilities = final_pipeline.predict_proba(X_new_predict)
predicted_new_purchase_proba = new_probabilities[:, 1]

new_predictions_df = df_new_features[['customer_id', 'product_id', 'target_week']].copy()
new_predictions_df['predicted_proba'] = predicted_new_purchase_proba
new_predictions_df['predicted_purchase_binary'] = (new_predictions_df['predicted_proba'] >= optimal_threshold).astype(int)

N_top = 20
print(f"\nTop {N_top} predicciones de compra para la semana '{new_target_week}' (con umbral {optimal_threshold:.2f}):")
print(new_predictions_df.sort_values(by='predicted_proba', ascending=False).head(N_top).to_string())

# --- 11. Save Predictions for 2025-W03 ---
output_filename_w03 = 'purchase_predictions_2025_W03.csv'
try:
    new_predictions_df.to_csv(output_filename_w03, index=False)
    print(f"El DataFrame de predicciones para 2025-W03 ha sido guardado exitosamente como '{output_filename_w03}'.")
except Exception as e:
    print(f"Error al guardar el archivo de predicciones para 2025-W03: {e}")
````

````python
print("--- Redefiniendo particiones de datos con historial actualizado ---")

# 1. Obtener las semanas totales únicas de `df_merged_updated` y ordénelas.
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_total_updated = len(weeks_total_updated)

# 2. Mantener las definiciones de num_weeks_test y num_weeks_val como 4 semanas cada una.
num_weeks_test = 4
num_weeks_val = 4

# 3. Calcular los índices de inicio para los conjuntos de validación y prueba
idx_val_start_updated = num_weeks_total_updated - num_weeks_test - num_weeks_val
idx_test_start_updated = num_weeks_total_updated - num_weeks_test

# 4. Crear las listas de semanas para entrenamiento, validación y prueba
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:]

print(f"Semanas para Entrenamiento actualizadas ({len(weeks_train)}): {weeks_train[0]} a {weeks_train[-1]}")
print(f"Semanas para Validación actualizadas ({len(weeks_val)}): {weeks_val[0]} a {weeks_val[-1]}")
print(f"Semanas para Prueba actualizadas ({len(weeks_test)}): {weeks_test[0]} a {weeks_test[-1]}")

# 5. Generar los nuevos DataFrames df_train, df_val y df_test
df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()

# 6. Imprimir dimensiones y rangos de fechas para verificación
print(f"\nDimensiones df_train actualizado: {df_train.shape}")
print(f"Fechas en df_train actualizado: {df_train['purchase_date'].min().strftime('%Y-%m-%d')} a {df_train['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_val actualizado: {df_val.shape}")
print(f"Fechas en df_val actualizado: {df_val['purchase_date'].min().strftime('%Y-%m-%d')} a {df_val['purchase_date'].max().strftime('%Y-%m-%d')}")

print(f"\nDimensiones df_test actualizado: {df_test.shape}")
print(f"Fechas en df_test actualizado: {df_test['purchase_date'].min().strftime('%Y-%m-%d')} a {df_test['purchase_date'].max().strftime('%Y-%m-%d')}")

print("--- Redefinición de particiones completada ---")
````

````python
print("--- Iniciando regeneración de Feature Engineering con historial actualizado ---")

# 1. Recrear el df_train_val DataFrame con los datos actualizados
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
print(f"df_train_val recreado. Dimensiones: {df_train_val.shape}")

# 2. Recrear df_train_hist_agg (historial para el conjunto de validación)
df_train_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_hist_agg = df_train_agg.copy()
print(f"df_train_hist_agg recreado. Dimensiones: {df_train_hist_agg.shape}")

# 3. Recrear df_train_val_hist_agg (historial para el conjunto de prueba)
df_train_val_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val_agg.copy()
print(f"df_train_val_hist_agg recreado. Dimensiones: {df_train_val_hist_agg.shape}")

# 4. Recrear df_base_val y df_base_test
# Se asume que `create_target_base` está definido en el contexto
# `target_weeks_val` y `target_weeks_test` se deben haber actualizado previamente en el paso de Holdout
print("\n--- Recreando bases para validación y prueba ---")
df_base_val = create_target_base(df_val, weeks_val) # Usar df_val y las semanas actualizadas
df_base_test = create_target_base(df_test, weeks_test) # Usar df_test y las semanas actualizadas

print("\n--- Aplicando Features de Lag ---")
# 5. Aplicar add_lag_features a df_base_val
df_feat_val = add_lag_features(
    base_df=df_base_val,
    transactions_hist_df=df_train_hist_agg, # Historial para validar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

# 6. Aplicar add_lag_features a df_base_test
df_feat_test = add_lag_features(
    base_df=df_base_test,
    transactions_hist_df=df_train_val_hist_agg, # Historial para probar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

print("\n--- Aplicando Features de Ventana Móvil ---")
# 7. Aplicar add_rolling_features a df_feat_val
df_feat_val = add_rolling_features(
    base_df=df_feat_val,
    transactions_hist_agg_df=df_train_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

# 8. Aplicar add_rolling_features a df_feat_test
df_feat_test = add_rolling_features(
    base_df=df_feat_test,
    transactions_hist_agg_df=df_train_val_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

print("\n--- Añadiendo Features Estáticas y Temporales ---")
# 9. Mergear features de producto con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')

# 10. Mergear features de cliente con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

# 11. Aplicar add_temporal_features a df_feat_val y df_feat_test
df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

print("--- Verificación de Features Regeneradas ---")
print(f"\nDimensiones de df_feat_val: {df_feat_val.shape}")
print("Primeras 5 filas de df_feat_val:")
print(df_feat_val.head().to_string())
print("\nNaN check en df_feat_val:")
nan_check_val = df_feat_val.isnull().sum()
print(nan_check_val[nan_check_val > 0].to_string())

print(f"\nDimensiones de df_feat_test: {df_feat_test.shape}")
print("Primeras 5 filas de df_feat_test:")
print(df_feat_test.head().to_string())
print("\nNaN check en df_feat_test:")
nan_check_test = df_feat_test.isnull().sum()
print(nan_check_test[nan_check_test > 0].to_string())

# 12. Actualizar X_val, y_val, X_test, y_test
X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()

print("\n--- Feature Engineering regenerado y conjuntos X, y actualizados ---")
````

````python
print("--- Reentrenando el Modelo LightGBM Optimizado ---")

# Obtener los mejores parámetros de la optimización anterior
best_params = study.best_trial.params

# 1. Recrear el preprocesador con los mejores parámetros
num_imputer_strategy_final = best_params['num_imputer_strategy']
numeric_imputer_re_fit = SimpleImputer(strategy=num_imputer_strategy_final)

scaler_choice_re_fit = best_params['scaler']
if scaler_choice_re_fit == 'StandardScaler':
    scaler_re_fit = StandardScaler()
elif scaler_choice_re_fit == 'MinMaxScaler':
    scaler_re_fit = MinMaxScaler()
elif scaler_choice_re_fit == 'RobustScaler':
    scaler_re_fit = RobustScaler()
else:
    scaler_re_fit = 'passthrough'

numeric_transformer_re_fit = make_pipeline(numeric_imputer_re_fit, scaler_re_fit) if scaler_re_fit != 'passthrough' else make_pipeline(numeric_imputer_re_fit)
categorical_transformer_re_fit = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))

preprocessor_re_fit = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_re_fit, numeric_features),
        ('cat', categorical_transformer_re_fit, categorical_features)
    ],
    remainder='drop'
)
print("Preprocesador recreado con los mejores parámetros.")

# 2. Recrear el clasificador LightGBM con los mejores parámetros
lgbm_best_params_re_fit = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_re_fit = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params_re_fit)
print("Clasificador LightGBM recreado con los mejores parámetros.")

# 3. Crear el final_pipeline recreado
final_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_re_fit),
    ('classifier', lgbm_re_fit)
])
print("Pipeline final recreado.")

# 4. Entrenar el final_pipeline con los datos de entrenamiento actualizados (X_val y y_val)
print("Entrenando el pipeline final con X_val (historial actualizado) y y_val...")
final_pipeline.fit(X_val, y_val)
print("Reentrenamiento completado con éxito.")

# Opcional: Evaluar rápidamente el modelo reentrenado en X_val
from sklearn.metrics import classification_report
y_pred_val_re_fit = final_pipeline.predict(X_val)
print("\nClassification Report para X_val (después del reentrenamiento):")
print(classification_report(y_val, y_pred_val_re_fit, zero_division=0))
````

````python
print("--- Preparando datos para predicción (2025-W04) ---")

# 1. Definir la nueva semana objetivo
new_target_week = '2025-W04'

# 2. Asegurarse de que `current_historical_data` contenga `df_merged_updated` (que incluye el historial hasta 2025-W03) y añadir una columna 'purchased' con valor 1.
current_historical_data = df_merged_updated.copy()
current_historical_data['purchased'] = 1

# 3. Crear un DataFrame base (`df_new_predictions_base`) con todas las combinaciones ùnicas de `customer_id` y `product_id`
active_customers_for_new_week = df_clientes['customer_id'].unique()
active_products_for_new_week = df_productos['product_id'].unique()
df_new_predictions_base = pd.DataFrame(list(product(active_customers_for_new_week, active_products_for_new_week, [new_target_week])), columns=['customer_id', 'product_id', 'target_week'])
print(f"Tabla base para {new_target_week} creada. Dimensiones: {df_new_predictions_base.shape}")

# 4. Aplicar la función `add_lag_features`
df_new_features = add_lag_features(
    base_df=df_new_predictions_base,
    transactions_hist_df=current_historical_data,
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

# 5. Aplicar la función `add_rolling_features`
df_new_features = add_rolling_features(
    base_df=df_new_features,
    transactions_hist_agg_df=current_historical_data.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1),
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)
print(f"Features de Lag y Rolling añadidas. Dimensiones: {df_new_features.shape}")

# 6. Añadir las características estáticas de producto y cliente
df_new_features = pd.merge(df_new_features, productos_sel, on='product_id', how='left')
df_new_features = pd.merge(df_new_features, clientes_sel, on='customer_id', how='left')

# 7. Aplicar la función `add_temporal_features`
df_new_features = add_temporal_features(df_new_features)
print(f"Features estáticas y temporales añadidas. Dimensiones: {df_new_features.shape}")

# 8. Seleccionar solo las `feature_columns`
X_new_predict = df_new_features[feature_columns].copy()
print(f"X_new_predict listo para la inferencia. Dimensiones: {X_new_predict.shape}")

# 9. Realizar las predicciones de probabilidad con el modelo cargado (final_pipeline)
new_probabilities = final_pipeline.predict_proba(X_new_predict)
predicted_new_purchase_proba = new_probabilities[:, 1]

# 10. Aplicar el umbral óptimo para obtener predicciones binarias
new_predictions_df = df_new_features[['customer_id', 'product_id', 'target_week']].copy()
new_predictions_df['predicted_proba'] = predicted_new_purchase_proba
new_predictions_df['predicted_purchase_binary'] = (new_predictions_df['predicted_proba'] >= optimal_threshold).astype(int)

# 11. Formatear y mostrar las top N predicciones binarias y sus probabilidades
N_top = 20
print(f"\nTop {N_top} predicciones de compra para la semana '{new_target_week}' (con umbral {optimal_threshold:.2f}):")
print(new_predictions_df.sort_values(by='predicted_proba', ascending=False).head(N_top).to_string())
````

````python
import pandas as pd
from itertools import product
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from lightgbm import LGBMClassifier
import numpy as np
import warnings
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import optuna
import joblib

print("--- Initial Data Loading and Environment Setup ---")

# --- 1. Load Initial Datasets ---

try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    df_clientes = pd.read_parquet('clientes.parquet')
    df_productos = pd.read_parquet('productos.parquet')
    print("Datasets cargados exitosamente.")
except FileNotFoundError:
    print("Error: Uno o más archivos Parquet no se encontraron. Asegúrese de que estén en el directorio correcto.")
    exit() # Exit if crucial files are not found

# --- 2. Preprocess df_transacciones and Create df_merged ---
df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date

# Group transactions by customer, product, and day to sum 'items_neto'
df_trans_agg = df_transacciones.groupby(
    ['customer_id', 'product_id', 'purchase_date']
).agg(
    items_neto=('items', 'sum')
).reset_index()

# Filter for effective purchases (items_neto > 0)
df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)

df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])
df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week

df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')
df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')
df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)
print(f"Initial df_merged created. Dimensions: {df_merged.shape}")

# --- 3. Define Helper Functions for Feature Engineering ---
def year_week_to_date(yw_str):
    try:
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        return pd.NaT

def get_lag_week(target_week_str, lag):
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    if not target_weeks:
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])
    df_final = pd.merge(df_base, target_transactions, left_on=['customer_id', 'product_id', 'target_week'], right_on=['customer_id', 'product_id', 'year_week'], how='left')
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])
    df_final['target'] = df_final['target'].fillna(0).astype(int)
    return df_final

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    df_with_lags = base_df.copy()
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()
    for lag in lag_weeks:
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())
        df_with_lags = pd.merge(df_with_lags, hist_lag[merge_cols], on=['customer_id', 'product_id', 'lookup_week'], how='left')
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])
    return df_with_lags

def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])
    for window in windows:
        hist_data_indexed = hist_data.set_index('date')
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)
        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                rolling_col = rolling_col.reset_index()
                rolling_col = rolling_col.rename(columns={feature_col: col_name})
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(), on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})
                df_with_rolling = pd.merge(df_with_rolling, rolling_col[['customer_id', 'product_id', 'target_week', col_name]], on=['customer_id', 'product_id', 'target_week'], how='left')
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)
    return df_with_rolling

def add_temporal_features(df):
    df_temp = df.copy()
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    df_temp['month'] = df_temp['date'].dt.month.astype(int)
    df_temp = df_temp.drop(columns=['date'])
    return df_temp
print("Helper functions defined.")

# --- 4. Define Global Feature Lists and Selectors ---
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]
feature_columns = numeric_features + categorical_features
print("Global feature lists and selectors set up.")
````

````python
import pandas as pd
from itertools import product
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler, OneHotEncoder
from lightgbm import LGBMClassifier
import numpy as np
import warnings
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import roc_auc_score, classification_report, confusion_matrix, ConfusionMatrixDisplay, f1_score
import matplotlib.pyplot as plt
import optuna
import joblib

print("--- Re-ejecutando pasos previos para asegurar el estado del kernel y generar predicciones para 2025-W03 ---")
print("ATENCIÓN: Asegúrese de que 'transacciones.parquet', 'clientes.parquet', 'productos.parquet', 'batch_t1.parquet', 'batch_t2.parquet', 'batch_t3.parquet' estén en el directorio de trabajo actual.")

# --- 1. Initial Data Loading & Preprocessing (up to 2024-W52) ---
try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    df_clientes = pd.read_parquet('clientes.parquet')
    df_productos = pd.read_parquet('productos.parquet')

    df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date
    df_trans_agg = df_transacciones.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
    df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)
    df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])
    df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
    df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week
    df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')
    df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')
    df_merged['year_week'] = df_merged['year'].astype(str) + '-W' + df_merged['week'].astype(str).str.zfill(2)
    print("Initial df_merged (up to 2024-W52) recreated. Dims: ", df_merged.shape)
except Exception as e:
    print(f"Error recreating initial df_merged: {e}")

# --- 2. Helper Functions for Feature Engineering ---
def year_week_to_date(yw_str):
    try:
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        return pd.NaT

def get_lag_week(target_week_str, lag):
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    if not target_weeks:
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])
    df_final = pd.merge(df_base, target_transactions, left_on=['customer_id', 'product_id', 'target_week'], right_on=['customer_id', 'product_id', 'year_week'], how='left')
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])
    df_final['target'] = df_final['target'].fillna(0).astype(int)
    return df_final

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    df_with_lags = base_df.copy()
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week'])
    hist_data = hist_data.copy()
    for lag in lag_weeks:
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())
        df_with_lags = pd.merge(df_with_lags, hist_lag[merge_cols], on=['customer_id', 'product_id', 'lookup_week'], how='left')
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns:
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])
    return df_with_lags

def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])
    for window in windows:
        hist_data_indexed = hist_data.set_index('date')
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)
        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                rolling_col = rolling_col.reset_index()
                rolling_col = rolling_col.rename(columns={feature_col: col_name})
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(), on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})
                df_with_rolling = pd.merge(df_with_rolling, rolling_col[['customer_id', 'product_id', 'target_week', col_name]], on=['customer_id', 'product_id', 'target_week'], how='left')
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)
    return df_with_rolling

def add_temporal_features(df):
    df_temp = df.copy()
    df_temp['date'] = df_temp['target_week'].apply(year_week_to_date)
    df_temp['week_of_year'] = df_temp['date'].dt.isocalendar().week.astype(int)
    df_temp['month'] = df_temp['date'].dt.month.astype(int)
    df_temp = df_temp.drop(columns=['date'])
    return df_temp
print("Helper functions re-defined.")

# --- 3. Define Common Feature Engineering Variables ---
lags_to_compute = [1, 2, 3, 4]
features_for_lag = ['purchased', 'items']
windows_to_compute = [4, 8]
features_for_rolling = ['purchased', 'items']
agg_funcs_to_apply = ['sum', 'mean']

product_features = ['brand', 'package', 'sub_category']
productos_sel = df_productos[['product_id'] + product_features].copy()
for col in product_features:
    productos_sel[col] = productos_sel[col].astype('category')

client_features = ['customer_type']
clientes_sel = df_clientes[['customer_id'] + client_features].copy()
for col in client_features:
    clientes_sel[col] = clientes_sel[col].astype('category')

numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]
feature_columns = numeric_features + categorical_features
print("Feature engineering variables and lists re-defined.")

# --- 4. Process and Integrate batch_t1 (up to 2025-W01) ---
try:
    df_batch_t1_raw = pd.read_parquet('batch_t1.parquet')
    df_batch_t1_raw['purchase_date'] = pd.to_datetime(df_batch_t1_raw['purchase_date']).dt.date
    df_batch_t1_agg = df_batch_t1_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t1_effective_purchases = df_batch_t1_agg[df_batch_t1_agg['items_neto'] > 0].copy()
    df_batch_t1_effective_purchases_renamed = df_batch_t1_effective_purchases.rename(columns={'items_neto': 'items'})

    # Fix: Ensure purchase_date is datetime and extract year/week for batch_t1
    df_batch_t1_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t1_effective_purchases_renamed['purchase_date'])
    df_batch_t1_effective_purchases_renamed['year'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t1_effective_purchases_renamed['week'] = df_batch_t1_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t1_expanded = pd.merge(df_batch_t1_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t1_expanded = pd.merge(df_batch_t1_expanded, df_productos, on='product_id', how='left')
    df_batch_t1_expanded['year_week'] = df_batch_t1_expanded['year'].astype(str) + '-W' + df_batch_t1_expanded['week'].astype(str).str.zfill(2)

    # Initialize df_merged_updated with df_merged plus batch_t1. For subsequent batches, this will be updated.
    columns_to_align = df_merged.columns.tolist()
    missing_in_batch_t1 = set(columns_to_align) - set(df_batch_t1_expanded.columns)
    for col in missing_in_batch_t1:
        df_batch_t1_expanded[col] = np.nan # Use np.nan for missing columns to maintain dtype compatibility
    df_batch_t1_aligned = df_batch_t1_expanded[columns_to_align].copy()

    df_merged_updated = pd.concat([df_merged, df_batch_t1_aligned], ignore_index=True)
    df_merged_updated.drop_duplicates(inplace=True)
    df_merged_updated.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated.reset_index(drop=True, inplace=True)
    print("batch_t1 integrated into df_merged_updated. Dims: ", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t1: {e}")

# --- 5. Redefine Data Splits after batch_t1 ---
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_test = 4
num_weeks_val = 4
idx_val_start_updated = len(weeks_total_updated) - num_weeks_test - num_weeks_val
idx_test_start_updated = len(weeks_total_updated) - num_weeks_test
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:] # Contains 2025-W01 for evaluation

df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()
print("Data splits re-defined after batch_t1.")

# --- 6. Regenerate Feature Engineering after batch_t1 ---
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
df_train_hist_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)

df_base_val = create_target_base(df_val, weeks_val)
df_base_test = create_target_base(df_test, weeks_test)

df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

X_val = df_feat_val[feature_columns].copy() # Global X_val
y_val = df_feat_val['target'].copy() # Global y_val
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()
print("Feature engineering re-generated after batch_t1.")

# --- 7. Re-run Optuna Optimization and Model Training (first run) ---
warnings.filterwarnings('ignore', category=optuna.exceptions.ExperimentalWarning)
warnings.filterwarnings('ignore', category=UserWarning)

def objective(trial):
    num_imputer_strategy = trial.suggest_categorical('num_imputer_strategy', ['mean', 'median'])
    numeric_imputer = SimpleImputer(strategy=num_imputer_strategy)
    scaler_choice = trial.suggest_categorical('scaler', ['StandardScaler', 'MinMaxScaler', 'RobustScaler', 'None'])
    if scaler_choice == 'StandardScaler': scaler = StandardScaler()
    elif scaler_choice == 'MinMaxScaler': scaler = MinMaxScaler()
    elif scaler_choice == 'RobustScaler': scaler = RobustScaler()
    else: scaler = 'passthrough'
    numeric_transformer = make_pipeline(numeric_imputer, scaler) if scaler != 'passthrough' else make_pipeline(numeric_imputer)
    categorical_transformer = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    preprocessor = ColumnTransformer(transformers=[('num', numeric_transformer, numeric_features), ('cat', categorical_transformer, categorical_features)], remainder='drop')
    lgbm_params = {
        'objective': 'binary', 'metric': 'f1', 'random_state': 42, 'n_jobs': -1,
        'n_estimators': trial.suggest_int('n_estimators', 200, 1500, step=100),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
        'num_leaves': trial.suggest_int('num_leaves', 20, 100),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-3, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-3, 10.0, log=True),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'is_unbalance': trial.suggest_categorical('is_unbalance', [True, False])
    }
    model = LGBMClassifier(**lgbm_params)
    pipeline = Pipeline(steps=[('preprocessor', preprocessor), ('classifier', model)])
    cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    try: scores = cross_val_score(pipeline, X_val, y_val, scoring='f1', cv=cv, n_jobs=-1); f1_mean = np.mean(scores); return f1_mean
    except Exception as e: return 0.0

study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=15, timeout=1800)
best_params = study.best_trial.params

num_imputer_final = SimpleImputer(strategy=best_params['num_imputer_strategy'])
scaler_choice_final = best_params['scaler']
if scaler_choice_final == 'StandardScaler': scaler_final = StandardScaler()
elif scaler_choice_final == 'MinMaxScaler': scaler_final = MinMaxScaler()
elif scaler_choice_final == 'RobustScaler': scaler_final = RobustScaler()
else: scaler_final = 'passthrough'
numeric_transformer_final = make_pipeline(num_imputer_final, scaler_final) if scaler_final != 'passthrough' else make_pipeline(num_imputer_final)
categorical_transformer_final = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
preprocessor_final = ColumnTransformer(transformers=[('num', numeric_transformer_final, numeric_features), ('cat', categorical_transformer_final, categorical_features)], remainder='drop')
lgbm_best_params = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_final = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params)
final_pipeline = Pipeline(steps=[('preprocessor', preprocessor_final), ('classifier', lgbm_final)])
final_pipeline.fit(X_val, y_val)

y_pred_proba_val = final_pipeline.predict_proba(X_val)[:, 1]
optimal_threshold = 0
best_f1_score = 0
for threshold in np.arange(0.0, 1.01, 0.01):
    y_pred_binary = (y_pred_proba_val >= threshold).astype(int)
    f1 = f1_score(y_val, y_pred_binary, pos_label=1, zero_division=0)
    if f1 > best_f1_score:
        best_f1_score = f1
        optimal_threshold = threshold
print(f"Model re-trained (after batch_t1) and optimal_threshold found: {optimal_threshold:.2f}")

# --- 6. Process and Integrate batch_t2 (up to 2025-W02) ---
try:
    df_batch_t2_raw = pd.read_parquet('batch_t2.parquet')
    df_batch_t2_raw['purchase_date'] = pd.to_datetime(df_batch_t2_raw['purchase_date']).dt.date
    df_batch_t2_agg = df_batch_t2_raw.groupby(['customer_id', 'product_id', 'purchase_date']).agg(items_neto=('items', 'sum')).reset_index()
    df_batch_t2_effective_purchases = df_batch_t2_agg[df_batch_t2_agg['items_neto'] > 0].copy()
    df_batch_t2_effective_purchases_renamed = df_batch_t2_effective_purchases.rename(columns={'items_neto': 'items'})

    # Fix: Ensure purchase_date is datetime and extract year/week for batch_t2
    df_batch_t2_effective_purchases_renamed['purchase_date'] = pd.to_datetime(df_batch_t2_effective_purchases_renamed['purchase_date'])
    df_batch_t2_effective_purchases_renamed['year'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().year
    df_batch_t2_effective_purchases_renamed['week'] = df_batch_t2_effective_purchases_renamed['purchase_date'].dt.isocalendar().week

    df_batch_t2_expanded = pd.merge(df_batch_t2_effective_purchases_renamed, df_clientes, on='customer_id', how='left')
    df_batch_t2_expanded = pd.merge(df_batch_t2_expanded, df_productos, on='product_id', how='left')
    df_batch_t2_expanded['year_week'] = df_batch_t2_expanded['year'].astype(str) + '-W' + df_batch_t2_expanded['week'].astype(str).str.zfill(2)

    # Align columns before concatenation
    columns_to_align = df_merged_updated.columns.tolist()
    missing_in_batch_t2 = set(columns_to_align) - set(df_batch_t2_expanded.columns)
    for col in missing_in_batch_t2:
        df_batch_t2_expanded[col] = np.nan # Use np.nan for missing columns to maintain dtype compatibility
    df_batch_t2_aligned = df_batch_t2_expanded[columns_to_align].copy()

    df_merged_updated_final = pd.concat([df_merged_updated, df_batch_t2_aligned], ignore_index=True)
    df_merged_updated_final.drop_duplicates(inplace=True)
    df_merged_updated_final.sort_values(by=['purchase_date', 'customer_id', 'product_id'], inplace=True)
    df_merged_updated_final.reset_index(drop=True, inplace=True)
    df_merged_updated = df_merged_updated_final.copy()
    print("batch_t2 integrated into df_merged_updated. Dims: ", df_merged_updated.shape)
except Exception as e:
    print(f"Error integrating batch_t2: {e}")

# --- 7. Redefine Data Splits after batch_t2 ---
weeks_total_updated = sorted(df_merged_updated['year_week'].unique())
num_weeks_test = 4
num_weeks_val = 4
idx_val_start_updated = len(weeks_total_updated) - num_weeks_test - num_weeks_val
idx_test_start_updated = len(weeks_total_updated) - num_weeks_test
weeks_train = weeks_total_updated[:idx_val_start_updated]
weeks_val = weeks_total_updated[idx_val_start_updated:idx_test_start_updated]
weeks_test = weeks_total_updated[idx_test_start_updated:] # Contains 2025-W02 for evaluation

df_train = df_merged_updated[df_merged_updated['year_week'].isin(weeks_train)].copy()
df_val = df_merged_updated[df_merged_updated['year_week'].isin(weeks_val)].copy()
df_test = df_merged_updated[df_merged_updated['year_week'].isin(weeks_test)].copy()
print("Data splits re-defined after batch_t2.")

# --- 8. Regenerate Feature Engineering after batch_t2 ---
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
df_train_hist_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)

df_base_val = create_target_base(df_val, weeks_val)
df_base_test = create_target_base(df_test, weeks_test)

df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()
print("Feature engineering re-generated after batch_t2.")

# --- 9. Re-train Model after batch_t2 ---
lgbm_best_params_re_fit = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_re_fit = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params_re_fit)

num_imputer_re_fit = SimpleImputer(strategy=best_params['num_imputer_strategy'])
scaler_choice_re_fit = best_params['scaler']
if scaler_choice_re_fit == 'StandardScaler': scaler_re_fit = StandardScaler()
elif scaler_choice_re_fit == 'MinMaxScaler': scaler_re_fit = MinMaxScaler()
elif scaler_choice_re_fit == 'RobustScaler': scaler_re_fit = RobustScaler()
else: scaler_re_fit = 'passthrough'
numeric_transformer_re_fit = make_pipeline(num_imputer_re_fit, scaler_re_fit) if scaler_re_fit != 'passthrough' else make_pipeline(num_imputer_re_fit)
categorical_transformer_re_fit = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))
preprocessor_final = ColumnTransformer(transformers=[('num', numeric_transformer_re_fit, numeric_features), ('cat', categorical_transformer_re_fit, categorical_features)], remainder='drop')

final_pipeline = Pipeline(steps=[('preprocessor', preprocessor_final), ('classifier', lgbm_re_fit)])

final_pipeline.fit(X_val, y_val)
print("Model re-trained with data including batch_t2.")

# Re-calculate optimal_threshold after retraining with batch_t2 data
y_pred_proba_val_current = final_pipeline.predict_proba(X_val)[:, 1]
optimal_threshold = 0
best_f1_score_current = 0
for threshold in np.arange(0.0, 1.01, 0.01):
    y_pred_binary_current = (y_pred_proba_val_current >= threshold).astype(int)
    f1_current = f1_score(y_val, y_pred_binary_current, pos_label=1, zero_division=0)
    if f1_current > best_f1_score_current:
        best_f1_score_current = f1_current
        optimal_threshold = threshold
print(f"Optimal threshold recalculated after re-training: {optimal_threshold:.2f}")

# --- 10. Prepare and Generate Predictions for 2025-W03 ---
new_target_week = '2025-W03'

current_historical_data = df_merged_updated.copy()
current_historical_data['purchased'] = 1 # Add 'purchased' column for feature generation

active_customers_for_new_week = df_clientes['customer_id'].unique()
active_products_for_new_week = df_productos['product_id'].unique()
df_new_predictions_base = pd.DataFrame(list(product(active_customers_for_new_week, active_products_for_new_week, [new_target_week])), columns=['customer_id', 'product_id', 'target_week'])
print(f"Tabla base para {new_target_week} creada. Dimensions: {df_new_predictions_base.shape}")

df_new_features = add_lag_features(
    base_df=df_new_predictions_base,
    transactions_hist_df=current_historical_data,
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

df_new_features = add_rolling_features(
    base_df=df_new_features,
    transactions_hist_agg_df=current_historical_data.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1),
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)
print(f"Features de Lag y Rolling añadidas. Dimensions: {df_new_features.shape}")

df_new_features = pd.merge(df_new_features, productos_sel, on='product_id', how='left')
df_new_features = pd.merge(df_new_features, clientes_sel, on='customer_id', how='left')

df_new_features = add_temporal_features(df_new_features)
print(f"Features estáticas y temporales añadidas. Dimensions: {df_new_features.shape}")

X_new_predict = df_new_features[feature_columns].copy()
print(f"X_new_predict listo para la inferencia. Dimensions: {X_new_predict.shape}")

new_probabilities = final_pipeline.predict_proba(X_new_predict)
predicted_new_purchase_proba = new_probabilities[:, 1]

new_predictions_df = df_new_features[['customer_id', 'product_id', 'target_week']].copy()
new_predictions_df['predicted_proba'] = predicted_new_purchase_proba
new_predictions_df['predicted_purchase_binary'] = (new_predictions_df['predicted_proba'] >= optimal_threshold).astype(int)

N_top = 20
print(f"\nTop {N_top} predicciones de compra para la semana '{new_target_week}' (con umbral {optimal_threshold:.2f}):")
print(new_predictions_df.sort_values(by='predicted_proba', ascending=False).head(N_top).to_string())

# --- 11. Save Predictions for 2025-W03 ---
output_filename_w03 = 'purchase_predictions_2025_W03.csv'
try:
    new_predictions_df.to_csv(output_filename_w03, index=False)
    print(f"El DataFrame de predicciones para 2025-W03 ha sido guardado exitosamente como '{output_filename_w03}'.")
except Exception as e:
    print(f"Error al guardar el archivo de predicciones para 2025-W03: {e}")
````

````python
print("--- Iniciando regeneración de Feature Engineering con historial actualizado ---")

# 1. Recrear el df_train_val DataFrame con los datos actualizados
df_train_val = pd.concat([df_train, df_val], ignore_index=True)
print(f"df_train_val recreado. Dimensiones: {df_train_val.shape}")

# 2. Recrear df_train_hist_agg (historial para el conjunto de validación)
df_train_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_hist_agg = df_train_agg.copy()
print(f"df_train_hist_agg recreado. Dimensiones: {df_train_hist_agg.shape}")

# 3. Recrear df_train_val_hist_agg (historial para el conjunto de prueba)
df_train_val_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1)
df_train_val_hist_agg = df_train_val_agg.copy()
print(f"df_train_val_hist_agg recreado. Dimensiones: {df_train_val_hist_agg.shape}")

# 4. Recrear df_base_val y df_base_test
# Se asume que `create_target_base` está definido en el contexto
# `target_weeks_val` y `target_weeks_test` se deben haber actualizado previamente en el paso de Holdout
print("\n--- Recreando bases para validación y prueba ---")
df_base_val = create_target_base(df_val, weeks_val) # Usar df_val y las semanas actualizadas
df_base_test = create_target_base(df_test, weeks_test) # Usar df_test y las semanas actualizadas

print("\n--- Aplicando Features de Lag ---")
# 5. Aplicar add_lag_features a df_base_val
df_feat_val = add_lag_features(
    base_df=df_base_val,
    transactions_hist_df=df_train_hist_agg, # Historial para validar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

# 6. Aplicar add_lag_features a df_base_test
df_feat_test = add_lag_features(
    base_df=df_base_test,
    transactions_hist_df=df_train_val_hist_agg, # Historial para probar
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

print("\n--- Aplicando Features de Ventana Móvil ---")
# 7. Aplicar add_rolling_features a df_feat_val
df_feat_val = add_rolling_features(
    base_df=df_feat_val,
    transactions_hist_agg_df=df_train_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

# 8. Aplicar add_rolling_features a df_feat_test
df_feat_test = add_rolling_features(
    base_df=df_feat_test,
    transactions_hist_agg_df=df_train_val_hist_agg,
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)

print("\n--- Añadiendo Features Estáticas y Temporales ---")
# 9. Mergear features de producto con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, productos_sel, on='product_id', how='left')
df_feat_test = pd.merge(df_feat_test, productos_sel, on='product_id', how='left')

# 10. Mergear features de cliente con df_feat_val y df_feat_test
df_feat_val = pd.merge(df_feat_val, clientes_sel, on='customer_id', how='left')
df_feat_test = pd.merge(df_feat_test, clientes_sel, on='customer_id', how='left')

# 11. Aplicar add_temporal_features a df_feat_val y df_feat_test
df_feat_val = add_temporal_features(df_feat_val)
df_feat_test = add_temporal_features(df_feat_test)

print("--- Verificación de Features Regeneradas ---")
print(f"\nDimensiones de df_feat_val: {df_feat_val.shape}")
print("Primeras 5 filas de df_feat_val:")
print(df_feat_val.head().to_string())
print("\nNaN check en df_feat_val:")
nan_check_val = df_feat_val.isnull().sum()
print(nan_check_val[nan_check_val > 0].to_string())

print(f"\nDimensiones de df_feat_test: {df_feat_test.shape}")
print("Primeras 5 filas de df_feat_test:")
print(df_feat_test.head().to_string())
print("\nNaN check en df_feat_test:")
nan_check_test = df_feat_test.isnull().sum()
print(nan_check_test[nan_check_test > 0].to_string())

# 12. Actualizar X_val, y_val, X_test, y_test
X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy()

print("\n--- Feature Engineering regenerado y conjuntos X, y actualizados ---")
````

````python
print("--- Reentrenando el Modelo LightGBM Optimizado ---")

# Obtener los mejores parámetros de la optimización anterior
best_params = study.best_trial.params

# 1. Recrear el preprocesador con los mejores parámetros
num_imputer_strategy_final = best_params['num_imputer_strategy']
numeric_imputer_re_fit = SimpleImputer(strategy=num_imputer_strategy_final)

scaler_choice_re_fit = best_params['scaler']
if scaler_choice_re_fit == 'StandardScaler':
    scaler_re_fit = StandardScaler()
elif scaler_choice_re_fit == 'MinMaxScaler':
    scaler_re_fit = MinMaxScaler()
elif scaler_choice_re_fit == 'RobustScaler':
    scaler_re_fit = RobustScaler()
else:
    scaler_re_fit = 'passthrough'

numeric_transformer_re_fit = make_pipeline(numeric_imputer_re_fit, scaler_re_fit) if scaler_re_fit != 'passthrough' else make_pipeline(numeric_imputer_re_fit)
categorical_transformer_re_fit = make_pipeline(SimpleImputer(strategy='most_frequent'), OneHotEncoder(handle_unknown='ignore', sparse_output=False))

preprocessor_re_fit = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer_re_fit, numeric_features),
        ('cat', categorical_transformer_re_fit, categorical_features)
    ],
    remainder='drop'
)
print("Preprocesador recreado con los mejores parámetros.")

# 2. Recrear el clasificador LightGBM con los mejores parámetros
lgbm_best_params_re_fit = {k: v for k, v in best_params.items() if k not in ['num_imputer_strategy', 'scaler']}
lgbm_re_fit = LGBMClassifier(objective='binary', metric='f1', random_state=42, n_jobs=-1, **lgbm_best_params_re_fit)
print("Clasificador LightGBM recreado con los mejores parámetros.")

# 3. Crear el final_pipeline recreado
final_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_re_fit),
    ('classifier', lgbm_re_fit)
])
print("Pipeline final recreado.")

# 4. Entrenar el final_pipeline con los datos de entrenamiento actualizados (X_val y y_val)
print("Entrenando el pipeline final con X_val (historial actualizado) y y_val...")
final_pipeline.fit(X_val, y_val)
print("Reentrenamiento completado con éxito.")

# Opcional: Evaluar rápidamente el modelo reentrenado en X_val
from sklearn.metrics import classification_report
y_pred_val_re_fit = final_pipeline.predict(X_val)
print("\nClassification Report para X_val (después del reentrenamiento):")
print(classification_report(y_val, y_pred_val_re_fit, zero_division=0))
````

````python
print("--- Preparando datos para predicción (2025-W04) ---")

# 1. Definir la nueva semana objetivo
new_target_week = '2025-W04'

# 2. Asegurarse de que `current_historical_data` contenga `df_merged_updated` (que incluye el historial hasta 2025-W03) y añadir una columna 'purchased' con valor 1.
current_historical_data = df_merged_updated.copy()
current_historical_data['purchased'] = 1

# 3. Crear un DataFrame base (`df_new_predictions_base`) con todas las combinaciones ùnicas de `customer_id` y `product_id`
active_customers_for_new_week = df_clientes['customer_id'].unique()
active_products_for_new_week = df_productos['product_id'].unique()
df_new_predictions_base = pd.DataFrame(list(product(active_customers_for_new_week, active_products_for_new_week, [new_target_week])), columns=['customer_id', 'product_id', 'target_week'])
print(f"Tabla base para {new_target_week} creada. Dimensiones: {df_new_predictions_base.shape}")

# 4. Aplicar la función `add_lag_features`
df_new_features = add_lag_features(
    base_df=df_new_predictions_base,
    transactions_hist_df=current_historical_data,
    lag_weeks=lags_to_compute,
    feature_cols=features_for_lag
)

# 5. Aplicar la función `add_rolling_features`
df_new_features = add_rolling_features(
    base_df=df_new_features,
    transactions_hist_agg_df=current_historical_data.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index().assign(purchased=1),
    windows=windows_to_compute,
    feature_cols=features_for_rolling,
    agg_funcs=agg_funcs_to_apply
)
print(f"Features de Lag y Rolling añadidas. Dimensiones: {df_new_features.shape}")

# 6. Añadir las características estáticas de producto y cliente
df_new_features = pd.merge(df_new_features, productos_sel, on='product_id', how='left')
df_new_features = pd.merge(df_new_features, clientes_sel, on='customer_id', how='left')

# 7. Aplicar la función `add_temporal_features`
df_new_features = add_temporal_features(df_new_features)
print(f"Features estáticas y temporales añadidas. Dimensiones: {df_new_features.shape}")

# 8. Seleccionar solo las `feature_columns`
X_new_predict = df_new_features[feature_columns].copy()
print(f"X_new_predict listo para la inferencia. Dimensiones: {X_new_predict.shape}")

# 9. Realizar las predicciones de probabilidad con el modelo cargado (final_pipeline)
new_probabilities = final_pipeline.predict_proba(X_new_predict)
predicted_new_purchase_proba = new_probabilities[:, 1]

# 10. Aplicar el umbral óptimo para obtener predicciones binarias
new_predictions_df = df_new_features[['customer_id', 'product_id', 'target_week']].copy()
new_predictions_df['predicted_proba'] = predicted_new_purchase_proba
new_predictions_df['predicted_purchase_binary'] = (new_predictions_df['predicted_proba'] >= optimal_threshold).astype(int)

# 11. Formatear y mostrar las top N predicciones binarias y sus probabilidades
N_top = 20
print(f"\nTop {N_top} predicciones de compra para la semana '{new_target_week}' (con umbral {optimal_threshold:.2f}):")
print(new_predictions_df.sort_values(by='predicted_proba', ascending=False).head(N_top).to_string())
````

### **Análisis de la Evolución del F1-Score**

A continuación, se presenta un resumen de la evolución del F1-Score observado en las distintas etapas del proyecto, basándose en los outputs proporcionados:

*   **Rendimiento Inicial del Modelo Optimizado (en X_val y X_test):**
    *   **F1-Score en X_val:** 0.49

    Estos valores representan el rendimiento del modelo LightGBM optimizado en los conjuntos de validación y prueba temporales generados a partir de los datos históricos iniciales, antes de la introducción de los batches de datos nuevos. Indican una capacidad inicial razonable del modelo para identificar compras en un entorno controlado.

    *   **F1-Score para 2025-W01 (evaluado contra `batch_t1.parquet`):** 0.2716
    La evaluación del modelo contra el primer batch de datos reales (`batch_t1.parquet`) muestra una disminución significativa en el F1-Score. Esto sugiere que el rendimiento del modelo en un escenario de "producción" es inferior al observado en los conjuntos de validación y prueba originales. Esta caída podría deberse a un *data drift*, donde la distribución de los datos de los nuevos batches difiere de la distribución de los datos de entrenamiento y validación iniciales, o simplemente que las compras en este batch inicial son más difíciles de predecir.

*   **Rendimiento en Batches 2, 3 y 4:**

Sin embargo, **no se ha realizado una evaluación explícita del F1-Score de estas predicciones contra sus respectivos datos reales (`batch_t2.parquet`, `batch_t3.parquet`, `batch_t4.parquet`) en los outputs disponibles del notebook.** Por lo tanto, no es posible comentar sobre la evolución del rendimiento del modelo en estos batches adicionales.

**Observaciones de la Evolución:**
La clara disminución del F1-Score del 0.49 (en X_val/X_test) al 0.2716 (en Batch 1) es un indicador importante. Resalta la necesidad de una monitorización continua del rendimiento en un entorno de producción y de estrategias para mitigar el impacto del *data drift*, como el reentrenamiento periódico del modelo con datos más recientes o la implementación de técnicas de adaptación de dominio. La falta de evaluación explícita para los batches subsiguientes impide un análisis completo de esta evolución.





### 📌 Conclusiones y Aprendizajes [1.0 puntos]

## **Conclusiones y Aprendizajes**

El desarrollo de un modelo predictivo para estimar la probabilidad semanal de compra de productos por cliente para SodAI Drinks ha permitido explorar diversas etapas del ciclo de vida del machine learning en un entorno simulado de producción.

### **Hallazgos Clave:**

1.  **Potencial Inicial del Modelo:** El modelo LightGBM optimizado mostró un rendimiento prometedor con un F1-Score de 0.49 en los conjuntos de validación y prueba iniciales. Esto indica que la metodología de ingeniería de características, la selección del modelo y la optimización de hiperparámetros fueron efectivas para aprender patrones de compra dentro del período histórico observado.
2.  **Desafío del Entorno de Producción:** La evaluación del primer batch de datos nuevos (`batch_t1.parquet` correspondiente a 2025-W01) reveló una caída significativa en el F1-Score a 0.2716. Esta discrepancia subraya la dificultad inherente de los modelos predictivos en entornos dinámicos y la posible existencia de *data drift*, donde las características de los datos en producción varían respecto a los datos de entrenamiento.
3.  **Importancia de la Reentrenamiento y Adaptación:** La estrategia de reentrenar el modelo con datos históricos actualizados (incluyendo los batches anteriores) y recalcular el umbral óptimo de F1-Score es crucial para mantener la relevancia y el rendimiento del modelo a lo largo del tiempo. Sin embargo, para los batches 2, 3 y 4, aunque se realizó el reentrenamiento y la predicción, no se evaluó explícitamente el F1-Score con los datos reales correspondientes, lo que limita la comprensión de la efectividad de esta estrategia en el tiempo.
4.  **Balance entre Precisión y Recall:** El análisis de Batch 1 mostró una precisión de 0.4924 y un recall de 0.1875. Esto indica que el modelo es moderadamente bueno en las predicciones positivas que realiza (casi la mitad son correctas), pero pierde una gran proporción de las compras reales. Este desequilibrio en las métricas resalta la importancia de elegir un umbral óptimo de manera cuidadosa, y la continua necesidad de ajustar este balance según las prioridades del negocio (¿es más importante no molestar a los clientes con falsos positivos o capturar la mayor cantidad posible de compras?).

### **Desafíos Enfrentados y Aprendizajes:**

1.  **Data Drift:** La diferencia en el rendimiento entre el entorno de desarrollo y el primer batch de datos de "producción" confirma el desafío del *data drift*. Es fundamental implementar herramientas de monitoreo de data drift y concept drift en el pipeline para detectar estos cambios tempranamente y activar reentrenamientos o ajustes de umbrales de manera proactiva.
2.  **Complejidad del Feature Engineering:** La creación de features de lag y ventana móvil fue intensiva y requirió un manejo cuidadoso de las ventanas temporales para evitar *data leakage*. La definición de funciones auxiliares (`get_lag_week`, `add_lag_features`, `add_rolling_features`, `create_target_base`, `add_temporal_features`) fue esencial para estandarizar y automatizar este proceso.
3.  **Optimización de Hiperparámetros:** El uso de Optuna fue vital para encontrar un conjunto de hiperparámetros que maximizaran el F1-Score, una métrica crucial dado el desbalance de clases. La optimización no solo consideró el modelo, sino también el preprocesador, lo que demuestra un enfoque integral para mejorar el pipeline.
4.  **Robustez del Pipeline:** El proceso de reentrenamiento y re-generación de predicciones para cada nuevo batch de datos resalta la necesidad de un pipeline robusto y automatizado. La capacidad de integrar nuevos datos, recalcular features, reentrenar el modelo y generar nuevas predicciones de manera fluida es un aprendizaje clave para la operatividad en un entorno de producción.

### **Próximos Pasos y Mejoras Potenciales:**

1.  **Evaluación Continua:** Es imperativo establecer un sistema de evaluación automática para cada nuevo batch de datos reales. Esto permitiría monitorear la degradación del modelo en tiempo real y entender cómo se adapta el rendimiento con cada reentrenamiento.
2.  **Análisis de Causas de la Caída del F1-Score:** Investigar por qué el F1-Score disminuyó en el primer batch. ¿Se debe a cambios en el comportamiento del cliente, la introducción de nuevos productos, estacionalidad no capturada, o la presencia de nuevos clientes/productos con patrones de compra atípicos?
3.  **Estrategias de Adaptación al Data Drift:** Evaluar y potencialmente implementar estrategias más sofisticadas para el data drift, como el reentrenamiento con una ventana móvil de datos, la ponderación de muestras más recientes, o modelos que se adapten continuamente.
4.  **Exploración de Modelos Adicionales:** Aunque LightGBM ha demostrado ser eficaz, la exploración de modelos de ensamblaje (e.g., CatBoost, XGBoost) o redes neuronales podría ofrecer mejoras adicionales.
5.  **Análisis de Errores:** Realizar un análisis más profundo de los Falsos Positivos y Falsos Negativos para entender los tipos de errores que comete el modelo y dónde se pueden realizar mejoras, posiblemente a través de nuevas características o una comprensión más profunda del dominio del negocio.
6.  **Interpretabilidad del Modelo:** Integrar técnicas de interpretabilidad (como SHAP o LIME) para entender mejor qué características influyen más en las predicciones, lo que puede proporcionar insights valiosos para el negocio y ayudar a identificar las causas de la variación en el rendimiento.

En general, el proyecto ha sido una valiosa experiencia en la construcción de un sistema de recomendación robusto y adaptable, destacando la importancia de la ingeniería de características, la optimización de modelos y la continua evaluación en un entorno de datos en evolución.
