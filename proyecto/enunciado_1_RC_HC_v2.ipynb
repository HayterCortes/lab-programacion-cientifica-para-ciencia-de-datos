Para esto, pueden aplicar procesamientos como:

````python
# Importar bibliotecas pandas
import pandas as pd

print("--- Cargando Datos ---")
# --- Carga de Datos ---

try:
    df_transacciones = pd.read_parquet('transacciones.parquet')
    print("Archivo transacciones.parquet cargado exitosamente.")
    print("Primeras 5 filas de transacciones:")
    print(df_transacciones.head())
    print("\nInformación del DataFrame de transacciones:")
    df_transacciones.info()
except FileNotFoundError:
    print("Error: El archivo transacciones.parquet no se encontró.")

try:
    df_clientes = pd.read_parquet('clientes.parquet')
    print("\nArchivo clientes.parquet cargado exitosamente.")
    print("Primeras 5 filas de clientes:")
    print(df_clientes.head())
    print("\nInformación del DataFrame de clientes:")
    df_clientes.info()
except FileNotFoundError:
    print("Error: El archivo clientes.parquet no se encontró.")

try:
    df_productos = pd.read_parquet('productos.parquet')
    print("\nArchivo productos.parquet cargado exitosamente.")
    print("Primeras 5 filas de productos:")
    print(df_productos.head())
    print("\nInformación del DataFrame de productos:")
    df_productos.info()
except FileNotFoundError:
    print("Error: El archivo productos.parquet no se encontró.")

# --- Transformaciones Iniciales y Agregación Diaria ---
if 'df_transacciones' in locals() and 'df_clientes' in locals() and 'df_productos' in locals():
    print("\n--- Realizando Transformaciones y Agregación Diaria ---")

    # Convertir 'purchase_date' a tipo datetime y extraer solo la fecha (sin la hora)
    df_transacciones['purchase_date'] = pd.to_datetime(df_transacciones['purchase_date']).dt.date

    # Agrupar transacciones por cliente, producto y día.
    # Se suman los 'items' para netear compras y devoluciones del mismo día.
    print("Agregando transacciones a nivel diario para netear compras y devoluciones...")
    df_trans_agg = df_transacciones.groupby(
        ['customer_id', 'product_id', 'purchase_date']
    ).agg(
        items_neto=('items', 'sum')
    ).reset_index()

    print(f"Filas antes de la agregación: {len(df_transacciones)}")
    print(f"Filas después de la agregación diaria: {len(df_trans_agg)}")

    # Filtrar para mantener solo las compras efectivas (items_neto > 0)
    df_compras_efectivas = df_trans_agg[df_trans_agg['items_neto'] > 0].copy()
    print(f"Filas después de filtrar por compras efectivas (items_neto > 0): {len(df_compras_efectivas)}")

    # --- Cruce de Información (Merge) con los datos ya agregados ---
    print("\n--- Realizando Cruce de Datos (Merge) ---")
    # Renombrar columna para claridad en los pasos siguientes
    df_compras_efectivas.rename(columns={'items_neto': 'items'}, inplace=True)
    df_compras_efectivas['purchase_date'] = pd.to_datetime(df_compras_efectivas['purchase_date'])

    # Crear columnas 'year' y 'week' a partir de la fecha de compra
    df_compras_efectivas['year'] = df_compras_efectivas['purchase_date'].dt.isocalendar().year
    df_compras_efectivas['week'] = df_compras_efectivas['purchase_date'].dt.isocalendar().week

    # Cruzar con información de clientes
    print("Cruzando con clientes...")
    df_merged = pd.merge(df_compras_efectivas, df_clientes, on='customer_id', how='left')

    # Cruzar el resultado con información de productos
    print("Cruzando con productos...")
    df_merged = pd.merge(df_merged, df_productos, on='product_id', how='left')

    print(f"Cruce finalizado. Dimensiones del DataFrame 'df_merged': {df_merged.shape}")

    # --- Verificación Final de Nulos y Duplicados Post-Proceso ---
    print("\n--- Verificación Post-Merge ---")
    print(f"Número de filas duplicadas encontradas: {df_merged.duplicated().sum()}")
    nulos = df_merged.isnull().sum()
    print("\nValores nulos por columna:")
    print(nulos[nulos > 0])
    print("Primeras 5 filas del DataFrame combinado:")
    print(df_merged.head())
    print("\nInformación del DataFrame combinado:")
    df_merged.info()    

    print("\n--- Pre-procesamiento Completado ---")
    print(f"Dimensiones finales: {df_merged.shape}")

else:
    print("\nNo se pudieron cargar todos los DataFrames necesarios. El pre-procesamiento no pudo continuar.")
````

### Justificación del Pre-procesamiento

El objetivo principal de esta etapa es transformar los datos crudos, distribuidos en tres archivos distintos (`transacciones.parquet`, `clientes.parquet`, `productos.parquet`), en un único conjunto de datos consolidado y limpio, adecuado para el análisis exploratorio y posterior modelamiento. Los pasos realizados fueron los siguientes:

1.  **Carga de Datos**: Se cargaron los tres archivos Parquet utilizando la biblioteca `pandas`. Este es el primer paso esencial para poder trabajar con la información proporcionada. Se verificó la carga inicial mostrando las primeras filas (`.head()`) y la información general (`.info()`) de cada DataFrame.

2.  **Agregación Diaria y Manejo de Devoluciones**:  Se identificó que los valores negativos en la columna `items` corresponden a devoluciones que anulan compras previas.
    * Para abordar esto, las transacciones se agruparon por `customer_id`, `product_id` y `purchase_date` (a nivel de día).
    * Se calculó la suma neta de `items` para cada grupo. Este proceso consolida todas las compras y devoluciones de un mismo producto por un mismo cliente en un solo día en un único registro.
    * Finalmente, se filtró el resultado para conservar únicamente los registros donde la suma neta de `items` fue mayor a cero. El DataFrame resultante, `df_compras_efectivas`, representa interacciones de compra reales y netas, eliminando así el ruido de transacciones anuladas y asegurando que el modelo aprenda de interacciones que efectivamente ocurrieron.

3.  **Extracción de Componentes Temporales (`year`, `week`)**: Dado que el objetivo del proyecto es predecir la probabilidad de compra *semanalmente*, se extrajeron el año y el número de la semana (según el estándar ISO) de la columna `purchase_date`. Estas nuevas columnas (`year`, `week`) sirven para la posterior creación de features y para la estrategia de validación temporal.

4.  **Cruce de Información (`merge`)**: Para poder analizar y modelar las compras en función de las características del cliente y del producto, fue necesario combinar los tres DataFrames.
    * Se realizó un `left join` desde `df_compras_efectivas` hacia `df_clientes` usando `customer_id` como clave, para añadir los atributos de cada cliente a sus transacciones.
    * Luego, se realizó otro `left join` desde el resultado anterior hacia `df_productos` usando `product_id` como clave, para incorporar las características de cada producto.

5.  **Verificación Final**:  Se realizó una última comprobación para asegurar la calidad de los datos. Como resultado del proceso de agregación, no se encontraron filas duplicadas. Además, la verificación de valores nulos confirmó que los cruces no generaron datos faltantes, lo que simplifica los pasos posteriores.

El objetivo es aplicar una serie de transformaciones en un orden definido para asegurar que los datos estén listos para los modelos a entrenar. Cada una de estas transformaciones debe ser justificada en función de su relevancia para el problema y los datos, y es importante evaluar su impacto en el rendimiento del modelo. Además, el pipeline debe ser flexible y modular para poder probar diferentes configuraciones de preprocesamiento.

````python
from itertools import product

def create_target_base(transactions_df: pd.DataFrame, target_weeks: list) -> pd.DataFrame:
    """
    Genera la tabla base con todas las combinaciones (customer_id, product_id, target_week)
    para las semanas objetivo dadas y calcula el target (si hubo compra o no).

    Args:
        transactions_df (pd.DataFrame): DataFrame con las transacciones (debe tener
                                         customer_id, product_id, year_week).
                                         Usado para determinar compras reales.
        target_weeks (list): Lista de semanas (formato 'YYYY-WNN') para las cuales
                             se generará la base y el target.

    Returns:
        pd.DataFrame: DataFrame con columnas [customer_id, product_id, target_week, target].
                      Contiene todas las combinaciones relevantes y si la compra ocurrió.
    """
    if not target_weeks:
        print("Warning: No target weeks provided.")
        return pd.DataFrame(columns=['customer_id', 'product_id', 'target_week', 'target'])

    print(f"Generando base para target_weeks: {target_weeks[0]} a {target_weeks[-1]}")

    # Filtrar transacciones relevantes para el target
    # Seleccionar solo las columnas necesarias y eliminar duplicados a nivel de semana
    target_transactions = transactions_df[transactions_df['year_week'].isin(target_weeks)][['customer_id', 'product_id', 'year_week']].drop_duplicates()
    target_transactions['target'] = 1 # Marcar que estas compras ocurrieron

    # --- Determinar el universo de clientes y productos activos ---
    # Se usan todos los clientes/productos presentes en el dataframe de transacciones de entrada
    active_customers = transactions_df['customer_id'].unique()
    active_products = transactions_df['product_id'].unique()

    print(f"Clientes activos considerados: {len(active_customers)}")
    print(f"Productos activos considerados: {len(active_products)}")

    # --- Crear todas las combinaciones posibles para las semanas objetivo ---
    base_tuples = list(product(active_customers, active_products, target_weeks))
    df_base = pd.DataFrame(base_tuples, columns=['customer_id', 'product_id', 'target_week'])

    print(f"Tamaño base inicial (combinaciones): {df_base.shape}")

    # --- Unir con las transacciones reales para obtener el target ---
    # Usar left_on y right_on para manejar nombres de columna diferentes
    df_final = pd.merge(
        df_base,
        target_transactions,
        left_on=['customer_id', 'product_id', 'target_week'], # Claves de df_base
        right_on=['customer_id', 'product_id', 'year_week'],   # Claves correspondientes en target_transactions
        how='left'
    )

    # Eliminar la columna 'year_week' redundante que viene del merge
    if 'year_week' in df_final.columns:
         df_final = df_final.drop(columns=['year_week'])

    # Rellenar con 0 donde no hubo compra
    df_final['target'] = df_final['target'].fillna(0).astype(int)

    print(f"Tamaño final base + target: {df_final.shape}")
    # Asegurarse que el target existe antes de calcular value_counts
    if 'target' in df_final.columns and not df_final.empty:
        print(f"Distribución del target:\n{df_final['target'].value_counts(normalize=True)}")
    else:
        print("Target column not found or DataFrame is empty after merge.")


    return df_final

# --- Aplicar la función para crear las bases de Validación y Test ---

print("\n--- Creando Base de Validación ---")
target_weeks_val = sorted(df_val['year_week'].unique())
df_base_val = create_target_base(df_val, target_weeks_val)

print("\n--- Creando Base de Test ---")
target_weeks_test = sorted(df_test['year_week'].unique())
df_base_test = create_target_base(df_test, target_weeks_test)


# Se muestran las primeras filas de una de las bases creadas
print("\nEjemplo df_base_val:")
# Asegurarse que df_base_val no está vacío antes de llamar a head()
if not df_base_val.empty:
    print(df_base_val.head())
else:
    print("df_base_val está vacío.")
````

````python
# --- Funciones Auxiliares para Fechas y Lags ---

def year_week_to_date(yw_str):
    """Convierte 'YYYY-WNN' a la fecha del lunes de esa semana ISO."""
    try:
        # %G: Año ISO, %V: Semana ISO (01-53), %u: Día semana ISO (1=Lunes)
        return pd.to_datetime(yw_str + '-1', format='%G-W%V-%u')
    except ValueError:
        # Manejar posibles errores de formato
        print(f"Error convirtiendo: {yw_str}")
        return pd.NaT

def get_lag_week(target_week_str, lag):
    """Calcula la semana 'YYYY-WNN' correspondiente a 'lag' semanas antes."""
    target_date = year_week_to_date(target_week_str)
    if pd.isna(target_date):
        return None
    lag_date = target_date - pd.DateOffset(weeks=lag)
    # Convertir de nuevo a formato 'YYYY-WNN'
    iso_year = lag_date.isocalendar().year
    iso_week = lag_date.isocalendar().week
    return f"{iso_year}-W{str(iso_week).zfill(2)}"

# --- Función para Añadir Features de Lag ---

def add_lag_features(base_df, transactions_hist_df, lag_weeks, feature_cols):
    """
    Añade features de lag (semanas anteriores) a la tabla base.

    Args:
        base_df (pd.DataFrame): Tabla base (customer_id, product_id, target_week, [target]).
        transactions_hist_df (pd.DataFrame): Transacciones históricas para buscar datos de lag.
                                            Debe tener customer_id, product_id, year_week,
                                            y las columnas en feature_cols.
        lag_weeks (list): Lista de enteros para los lags (ej., [1, 2, 4]).
        feature_cols (list): Lista de nombres de columnas de transactions_hist_df
                             cuyos valores se usarán como features de lag (ej., ['purchased', 'items']).

    Returns:
        pd.DataFrame: base_df con las nuevas columnas de features de lag añadidas.
    """
    df_with_lags = base_df.copy()

    # Preparar datos históricos (seleccionar columnas relevantes)
    hist_cols_needed = ['customer_id', 'product_id', 'year_week'] + feature_cols
    hist_data = transactions_hist_df[hist_cols_needed].drop_duplicates(subset=['customer_id', 'product_id', 'year_week']) 
    hist_data = hist_data.copy() 

    for lag in lag_weeks:
        print(f"  Generando lags de {lag} semana(s)...")
        # Calcular la semana de búsqueda (lookup_week) para esta pasada de lag
        df_with_lags['lookup_week'] = df_with_lags['target_week'].apply(lambda w: get_lag_week(w, lag))

        # Preparar datos históricos para el merge de este lag
        hist_lag = hist_data.rename(columns={'year_week': 'lookup_week'})

        # Definir nombres para las nuevas columnas de lag
        lag_col_names_map = {feat: f'lag_{lag}_{feat}' for feat in feature_cols}
        hist_lag = hist_lag.rename(columns=lag_col_names_map)

        # Columnas a seleccionar de hist_lag para el merge
        merge_cols = ['customer_id', 'product_id', 'lookup_week'] + list(lag_col_names_map.values())

        # Unir los datos históricos del lag correspondiente
        df_with_lags = pd.merge(
            df_with_lags,
            hist_lag[merge_cols],
            on=['customer_id', 'product_id', 'lookup_week'],
            how='left'
        )

        # Rellenar NaNs para las nuevas columnas de lag con 0
        for lag_col in lag_col_names_map.values():
            if lag_col in df_with_lags.columns: 
                 df_with_lags[lag_col] = df_with_lags[lag_col].fillna(0)
            else:
                 print(f"Advertencia: La columna de lag '{lag_col}' no se encontró después del merge para lag {lag}.")


        # Eliminar columna temporal
        df_with_lags = df_with_lags.drop(columns=['lookup_week'])

    return df_with_lags

# --- Aplicación para generar Features de Lag ---

# Combinar train y validation para obtener el historial completo para el Test set
print("Combinando df_train y df_val para historial de Test...")
df_train_val = pd.concat([df_train, df_val], ignore_index=True)

# Crear una columna 'purchased' = 1 en los datos históricos para facilitar el lag de compra S/N
df_train_hist = df_train.copy()
df_train_hist['purchased'] = 1
df_train_val_hist = df_train_val.copy()
df_train_val_hist['purchased'] = 1

# Definir lags y features a calcular
lags_to_compute = [1, 2, 3, 4] 
features_for_lag = ['purchased', 'items'] # Calcular lag para compra (S/N) y cantidad

# --- Generar features de lag para el Validation set ---
# Usar df_base_val como base y df_train_hist como historial
print("\nGenerando Features de Lag para Validation Set...")
df_feat_val = add_lag_features(df_base_val, df_train_hist, lags_to_compute, features_for_lag)

# --- Generar features de lag para el Test set ---
# Usar df_base_test como base y df_train_val_hist como historial
print("\nGenerando Features de Lag para Test Set...")
df_feat_test = add_lag_features(df_base_test, df_train_val_hist, lags_to_compute, features_for_lag)


# --- Mostrar Resultados ---
print("\nValidation Set con Features de Lag (primeras filas):")
print(df_feat_val.head())

print("\nTest Set con Features de Lag (primeras filas):")
print(df_feat_test.head())

# Verificar NaNs 
print("\nNaN check en df_feat_val:")
print(df_feat_val.isnull().sum())

print("\nNaN check en df_feat_test:")
print(df_feat_test.isnull().sum())
````

````python
# --- REFINAR DATOS HISTÓRICOS: Agregar 'items' por semana ---
print("Refinando datos históricos (agregando 'items' por semana)...")

# Para df_train_hist
df_train_agg = df_train.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index()
df_train_agg['purchased'] = 1 # Añadir indicador de compra
df_train_hist_agg = df_train_agg.copy()

# Para df_train_val_hist
df_train_val_agg = df_train_val.groupby(['customer_id', 'product_id', 'year_week'], observed=True)['items'].sum().reset_index()
df_train_val_agg['purchased'] = 1 # Añadir indicador de compra
df_train_val_hist_agg = df_train_val_agg.copy()

print("Datos históricos agregados.")


# --- RE-GENERAR Features de Lag con datos agregados ---

# --- Generar features de lag para el Validation set ---
print("\nRe-Generando Features de Lag para Validation Set (con items agregados)...")
df_feat_val = add_lag_features(df_base_val, df_train_hist_agg, lags_to_compute, features_for_lag)

# --- Generar features de lag para el Test set ---
print("\nRe-Generando Features de Lag para Test Set (con items agregados)...")
df_feat_test = add_lag_features(df_base_test, df_train_val_hist_agg, lags_to_compute, features_for_lag)

print("\nFeatures de Lag re-generadas.")

# --- Mostrar Resultados (verificar items enteros o suma correcta) ---
print("\nTest Set con Features de Lag Re-Generadas (primeras filas):")
print(df_feat_test.head())


# Features de Ventana Móvil 
def add_rolling_features(base_df, transactions_hist_agg_df, windows, feature_cols, agg_funcs):
    """
    Añade features de ventana móvil a la tabla base.
    Calcula agregados sobre N semanas ANTERIORES a la target_week.

    Args:
        base_df (pd.DataFrame): Tabla base (customer_id, product_id, target_week, ...).
        transactions_hist_agg_df (pd.DataFrame): Transacciones históricas AGREGADAS
                                                 (1 fila por cust-prod-week). Debe tener
                                                 customer_id, product_id, year_week, y feature_cols.
        windows (list): Lista de enteros para tamaños de ventana (ej., [4, 8]).
        feature_cols (list): Columnas sobre las que calcular los agregados (ej., ['purchased', 'items']).
        agg_funcs (list): Lista de funciones de agregación a aplicar (ej., ['sum', 'mean']).

    Returns:
        pd.DataFrame: base_df con las nuevas columnas de features de ventana móvil.
    """
    df_with_rolling = base_df.copy()
    hist_data = transactions_hist_agg_df.copy()

    # Asegurar orden temporal para rolling calculations
    hist_data['date'] = hist_data['year_week'].apply(year_week_to_date)
    hist_data = hist_data.sort_values(by=['customer_id', 'product_id', 'date'])

    print("\nGenerando Features de Ventana Móvil...")

    for window in windows:
        print(f"  Calculando para ventana de {window} semanas...")

        # Crear multi-índice temporal para rolling por grupo
        hist_data_indexed = hist_data.set_index('date')

        # Agrupar por cliente y producto
        grouped = hist_data_indexed.groupby(['customer_id', 'product_id'], observed=True)

        for feature_col in feature_cols:
             for agg_func in agg_funcs:
                col_name = f'rolling_{window}w_{agg_func}_{feature_col}'
                print(f"    Generando: {col_name}")

                # --- Enfoque simplificado con shift en semanas ---
                # Calcular rolling sin shift
                rolling_col = grouped[feature_col].rolling(window=window, min_periods=1).agg(agg_func)
                # Aplanar el multi-index resultante del groupby y rolling
                rolling_col = rolling_col.reset_index()
                 # Renombrar la columna resultante
                rolling_col = rolling_col.rename(columns={feature_col: col_name})

                # SHIFT para evitar Data Leakage
                rolling_col = pd.merge(rolling_col, hist_data[['customer_id', 'product_id', 'date', 'year_week']].drop_duplicates(),
                                       on=['customer_id', 'product_id', 'date'], how='left')
                rolling_col = rolling_col.sort_values(by=['customer_id', 'product_id', 'year_week'])
                rolling_col[col_name] = rolling_col.groupby(['customer_id', 'product_id'], observed=True)[col_name].shift(1)

                # Preparar para merge con base_df
                rolling_col = rolling_col.rename(columns={'year_week': 'target_week'})

                # Unir a la tabla base
                df_with_rolling = pd.merge(
                    df_with_rolling,
                    rolling_col[['customer_id', 'product_id', 'target_week', col_name]],
                    on=['customer_id', 'product_id', 'target_week'],
                    how='left'
                )
                # Rellenar NaNs
                df_with_rolling[col_name] = df_with_rolling[col_name].fillna(0)


    return df_with_rolling

# --- Aplicación para generar Features de Ventana Móvil ---

# Definir ventanas, features y agregaciones
windows_to_compute = [4, 8] 
features_for_rolling = ['purchased', 'items'] 
agg_funcs_to_apply = ['sum', 'mean'] 

# --- Generar features de ventana móvil para Validation set ---
# Usar df_feat_val como base y df_train_hist_agg como historial
df_feat_val = add_rolling_features(df_feat_val, df_train_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)

# --- Generar features de ventana móvil para Test set ---
# Usar df_feat_test como base y df_train_val_hist_agg como historial
df_feat_test = add_rolling_features(df_feat_test, df_train_val_hist_agg, windows_to_compute, features_for_rolling, agg_funcs_to_apply)


# --- Mostrar Resultados ---
print("\nValidation Set con Features de Lag y Rolling (primeras filas):")
print(df_feat_val.head())
print("\nTest Set con Features de Lag y Rolling (primeras filas):")
print(df_feat_test.head())

# Verificar columnas nuevas y NaNs
print("\nColumnas en df_feat_val:")
print(df_feat_val.columns)
print("\nNaN check en df_feat_val:")
print(df_feat_val.isnull().sum().tail(15))
````

````python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

print("--- Definiendo Pipelines de Preprocesamiento ---")

# 1. Identificar Tipos de Columnas para Features

# Columnas Numéricas
numeric_features = [
    'lag_1_purchased', 'lag_1_items', 'lag_2_purchased', 'lag_2_items',
    'lag_3_purchased', 'lag_3_items', 'lag_4_purchased', 'lag_4_items',
    'rolling_4w_sum_purchased', 'rolling_4w_mean_purchased', 'rolling_4w_sum_items', 'rolling_4w_mean_items',
    'rolling_8w_sum_purchased', 'rolling_8w_mean_purchased', 'rolling_8w_sum_items', 'rolling_8w_mean_items',
    'week_of_year', 'month'
]

# Columnas Categóricas
categorical_features = [
    'brand', 'package', 'sub_category', 'customer_type'
]

all_features = numeric_features + categorical_features
expected_cols = ['customer_id', 'product_id', 'target_week', 'target'] + all_features
missing_cols = set(expected_cols) - set(df_feat_val.columns)
extra_cols = set(df_feat_val.columns) - set(expected_cols)
if not missing_cols and not extra_cols:
    print("Verificación de columnas OK.")
else:
    print(f"WARN: Columnas faltantes en df_feat_val: {missing_cols}")
    print(f"WARN: Columnas extras en df_feat_val: {extra_cols}")

# 2. Crear Pipeline Numérico
#    - Imputación: 'median' porque es más robusta a outliers que 'mean'.
#    - Escalado: StandardScaler centra los datos (media 0, std 1).
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
print("Pipeline numérico definido (Imputer(median) -> StandardScaler).")

# 3. Crear Pipeline Categórico
#    - Imputación: 'most_frequent' para categóricas
#    - Codificación: OneHotEncoder es estándar para variables nominales sin orden inherente.
#                   handle_unknown='ignore' evita errores si aparecen nuevas categorías en test.
#                   sparse_output=False, fácil de inspeccionar.
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])
print("Pipeline categórico definido (Imputer(most_frequent) -> OneHotEncoder).")


# 4. Combinar con ColumnTransformer
#    Aplica el transformer correspondiente a cada grupo de columnas.
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ],
    remainder='drop' 
)

print("ColumnTransformer definido para combinar pipelines numérico y categórico.")
print("\nEstructura del Preprocesador:")
print(preprocessor)

# 5. Preparar los datos X e y
print("\n--- Preparando Conjuntos X, y ---")

# Columnas a usar como features
feature_columns = numeric_features + categorical_features

# Conjunto de Validación
X_val = df_feat_val[feature_columns].copy()
y_val = df_feat_val['target'].copy()

# Conjunto de Test
X_test = df_feat_test[feature_columns].copy()
y_test = df_feat_test['target'].copy() 
print(f"Dimensiones X_val: {X_val.shape}, y_val: {y_val.shape}")
print(f"Dimensiones X_test: {X_test.shape}, y_test: {y_test.shape}")

# 6. Probar el Preprocesador 
print("\n--- Probando el Preprocesador (fit en val, transform en val y test) ---")
try:
    # Ajustar el preprocesador con los datos de validación/entrenamiento
    preprocessor.fit(X_val)
    print("Preprocessor 'fit' ejecutado en X_val.")

    # Transformar los datos de validación
    X_val_processed = preprocessor.transform(X_val)
    print("Preprocessor 'transform' ejecutado en X_val.")
    print(f"Dimensiones de X_val procesado: {X_val_processed.shape}")

    # Transformar los datos de test (usando el preprocesador ya ajustado)
    X_test_processed = preprocessor.transform(X_test)
    print("Preprocessor 'transform' ejecutado en X_test.")
    print(f"Dimensiones de X_test procesado: {X_test_processed.shape}")

    # Intentar obtener nombres de features después de la transformación
    try:
        feature_names_out = preprocessor.get_feature_names_out()
        print(f"\nNúmero de features después del preprocesamiento: {len(feature_names_out)}")
        print("Algunos nombres de features procesadas:")
        print(feature_names_out[:10]) 
        print("...")
        print(feature_names_out[-10:])

    except Exception as e:
        print(f"\nNo se pudieron obtener los nombres de las features procesadas: {e}")

except Exception as e:
    print(f"\nError durante la prueba del preprocesador: {e}")
````

**Justificaciones:**

* **Imputación**: Se incluye `SimpleImputer` como una medida de robustez para manejar posibles datos faltantes en el futuro o en diferentes particiones de datos, aunque actualmente no haya. Se usa `median` para numéricos (menos sensible a outliers) y `most_frequent` para categóricos.
* **Escalado**: `StandardScaler` es fundamental para muchos algoritmos (regresión logística, SVM, redes neuronales, algoritmos basados en distancia) ya que asegura que todas las features numéricas tengan una escala comparable (media 0, desviación estándar 1).
* **Codificación Categórica**: `OneHotEncoder` es la elección estándar para variables categóricas nominales (sin un orden intrínseco) como `brand`, `package`, etc., ya que no introduce relaciones ordinales falsas. `handle_unknown='ignore'` es crucial para que el modelo no falle si encuentra una categoría en los datos de test que no vio durante el entrenamiento.
* **ColumnTransformer**: Permite aplicar las transformaciones correctas a los subconjuntos de columnas adecuados de manera organizada y reproducible.
* **Fit/Transform**: Es vital ajustar (`fit`) el preprocesador *únicamente* con los datos de entrenamiento/validación (`X_val`) para evitar fuga de datos (data leakage) del conjunto de test. Luego, se usa el mismo preprocesador ajustado para transformar (`transform`) tanto `X_val` como `X_test`.
